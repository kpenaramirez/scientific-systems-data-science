```python
def read_time_data():
    df = result_on_sky
    df['time_start'] = pd.to_datetime(df['time_start'])
    df['time_end'] = pd.to_datetime(df['time_end'])
    return df
    
def get_time_range(df, day_obs):
    day_data = df[df['day_obs'] == day_obs].iloc[0]
    return day_data['time_start'], day_data['time_end']
```

### Telemetry retrieval at 1min sampling:

```python
###ESS
###ESS inside dome
###Temperature
async def get_temperature_113_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get ESS temperature data for in-dome(M1M3) for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(temperatureItem0) AS temperature_113_Dome
    FROM "lsst.sal.ESS.temperature"
    WHERE salIndex = 113
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)

###Pressure
async def get_pressure_113_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get ESS pressure data for in-dome(M1M3) for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(pressureItem0) AS pressure_113_Dome
    FROM "lsst.sal.ESS.pressure"
    WHERE salIndex = 113
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)

###RH
async def get_RH_113_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get ESS RH data for in-dome(M1M3) for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(relativeHumidityItem) AS RH_113_Dome
    FROM "lsst.sal.ESS.relativeHumidity"
    WHERE salIndex = 113
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)



###ESS outside dome
###Temperature
async def get_temperature_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get ESS temperature data for Outside Dome for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(temperatureItem0) AS temperature_301_outDome
    FROM "lsst.sal.ESS.temperature"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)

####Pressure
async def get_pressure_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get pressure data for Outside Dome for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(pressureItem0) AS pressure_301_outDome
    FROM "lsst.sal.ESS.pressure"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)    

####Relative Humidity
async def get_RH_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get relative humidity data for Outside Dome for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(relativeHumidityItem) AS RH_301_outDome
    FROM "lsst.sal.ESS.relativeHumidity"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)



###Environmental general  
###ESS dewpoint
async def get_dewPoint_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get ESS data for dewpoint for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(dewPointItem) AS dewPoint_301
    FROM "lsst.sal.ESS.dewPoint"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)
  
####Sun Altitude
async def get_Sun_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get sun altitude data for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(sunAltitude) AS SunAltitude
    FROM "lsst.sal.MTPtg.skyEnvironment"
    WHERE time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)

####Wind speed
async def get_wind_speed_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get wind speed data for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(speed) AS WindSpeed_301
    FROM "lsst.sal.ESS.airFlow"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)


####Wind direction
async def get_wind_direction_301_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get wind direction data for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(direction) AS WindDirection_301
    FROM "lsst.sal.ESS.airFlow"
    WHERE salIndex = 301
    AND time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)



#####Dome Shutter
async def get_dome_shutter_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get dome shutter data for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(positionActual0) AS Dome_Shutter_Actual0,
        mean(positionActual1) AS Dome_Shutter_Actual1
    FROM "lsst.sal.MTDome.apertureShutter"
    WHERE time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)


#####Dome Azimuth
async def get_dome_azimuth_range(client, start_day_obs, end_day_obs, sampling="1m"):
    """Get dome azimuth data for a date range."""
    df = read_time_data()
    start_time, _ = get_time_range(df, int(start_day_obs))
    _, end_time = get_time_range(df, int(end_day_obs))
    
    query = f"""
    SELECT 
        mean(positionActual) AS Dome_Azimuth_Actual
    FROM "lsst.sal.MTDome.azimuth"
    WHERE time > '{start_time.isoformat()}Z'
    AND time < '{end_time.isoformat()}Z'
    GROUP BY time({sampling})
    """
    results = await client.influx_client.query(query)
    return pd.DataFrame(results)

```

```python
#### Assisted by gpt-5 and claude4.5sonet.
#################Units and normalization.
def normalize_units_robust(df):
    out = df.copy()
    #RH telemetry is in percentages.
    #RH: % -> fraction.
    for col in out.columns:
        if col.lower().startswith('rh_'):
            med = out[col].dropna().median()
            if med is not None and med > 2.0:
                out[col] = out[col] / 100.0
            out[col] = out[col].clip(0.0, 1.0)
    #Constants for H need kPa. It seems telemetry came at 1e5 units. Pa.
    #Pressure -> kPa.
    def to_kPa(series):
        med = series.dropna().median()
        if med is None or not np.isfinite(med):
            return pd.Series(73.9, index=series.index)  # At ~2600 m, pressure is around 70kPa.
        if 2_000 < med < 300_000:   # Pa
            return series / 1000.0
        if 500 < med < 1200:        # hPa (mbar)
            return series / 10.0
        if 50 < med < 120:          # kPa
            return series
        if 0.5 < med < 2.0:         # bar
            return series * 100.0
        # Unknown: fallback
        return pd.Series(73.9, index=series.index)

    for pcol in ['pressure_113_Dome', 'pressure_301_outDome']:
        if pcol in out.columns:
            out[pcol + '_kPa'] = to_kPa(out[pcol])

    return out

#########Equations for DeltaH.
##Saturation vapor pressure over liquid water f(T). T needs to be in C. The output pressure is in kPa.
##Here we used the Magnus relation which is valid in the T=[-45,60]C. And alternative is the Hyland-Wexler relation.
def es_kPa(T_C):
    return 0.61094 * np.exp((17.625 * T_C) / (T_C + 243.04))

###Humidity ratio w f(T,RH,P). The equation is: w = 0.62198 * e / (P - e), with e = RH * es(T).
def humidity_ratio_from_T_RH_P(T_C, RH_frac, P_kPa):
    e = RH_frac * es_kPa(T_C)
    return 0.62198 * e / np.maximum(P_kPa - e, 1e-6)

###Humidity ratio w f(T,P) from dewpoint: at dewpoint RH=1 -> e = es(Td).
def humidity_ratio_from_Td_P(Td_C, P_kPa):
    e = es_kPa(Td_C)
    return 0.62198 * e / np.maximum(P_kPa - e, 1e-6)

####Moist-air enthalpy per kg dry air (ASHRAE approx) f(T,w) so f(T,RH,P) or f(T,P).
###h_dry = 1.006*T + w*(2501 + 1.86*T) [kJ/kg_dry]
def enthalpy_kJ_per_kg_dry(T_C, w):
    return 1.006*T_C + w*(2501.0 + 1.86*T_C)

###Effective moist-air specific heat per kg of moist air f(w) so f(T,RH,P) or f(T,P), using constant cp at near-ambient.
###The /(1+w) factor is to keep all in moist air.
###c_p,moist = (c_pa + w*c_pv) / (1 + w)
###c_pa ≈ 1.006, c_pv ≈ 1.86  [kJ/(kg·K)]    
def cp_moist_kJ_per_kg_moistK(w):
    return (1.006 + 1.86*w) / (1.0 + w)

###########Apply the equations to the data.
def compute_equations(df, smooth_roll_min=10, vent_persist_min=5):
    ###Apply the normalization function.
    out = normalize_units_robust(df).copy()
    #Aliases to calculate the functions.
    #In-dome telemetry.
    T_in  = out['temperature_113_Dome']
    RH_in = out['RH_113_Dome']
    #In case we change the detector and we do not have pressure_XXX_Dome the default value would be 73.9 kPa.
    P_in  = out.get('pressure_113_Dome_kPa', pd.Series(73.9, index=out.index))
    #Out-dome telemetry.
    T_out  = out['temperature_301_outDome']
    RH_out = out.get('RH_301_outDome', None)
    #In case there are available data of outside pressure the default value would be 73.9 kPa.
    P_out  = out.get('pressure_301_outDome_kPa', pd.Series(73.9, index=out.index))
    Td_out = out['dewPoint_301']

    ###Functions calculated.
    #Humidity ratios w f(T,RH,P).
    w_in  = humidity_ratio_from_T_RH_P(T_in, RH_in, P_in)
    w_out = humidity_ratio_from_Td_P(Td_out, P_out)
    if RH_out is not None:
        w_out = w_out.fillna(humidity_ratio_from_T_RH_P(T_out, RH_out, P_out))

    #Enthalpy per kg dry air.
    h_in_dry  = enthalpy_kJ_per_kg_dry(T_in,  w_in)
    h_out_dry = enthalpy_kJ_per_kg_dry(T_out, w_out)

    #Convert enthalpy to per kg *moist air*.
    h_in_moist  = h_in_dry  / (1.0 + w_in)
    h_out_moist = h_out_dry / (1.0 + w_out)

    ###Data storage and deltas.
    #Store the results as columns in the dataframe.
    out['w_in'] = w_in
    out['w_out'] = w_out
    out['h_in_kJkgdry']  = h_in_dry
    out['h_out_kJkgdry'] = h_out_dry
    out['h_in_kJkgmoist']  = h_in_moist
    out['h_out_kJkgmoist'] = h_out_moist
    #Δh and ΔT
    out['dhe_kJkg_moist'] = h_in_moist - h_out_moist  #>0 means outside is energetically cooler.
    out['dTe_C']          = T_in - T_out #>0 means outside is cooler than inside.
    #Condensation margin: change here in case we want a colder surface to check condensation.
    out['dew_margin_C']   = T_in - Td_out

    ###Temporal data: smoothing.
    #Smooth (sm) the temporal data using the rolling function applying the mean with a window given by smooth_roll_min.
    win = max(int(smooth_roll_min), 5)     #<<<<<<<<<<----------- SAMPLING
    for c in [
        'h_in_kJkgmoist','h_out_kJkgmoist','dhe_kJkg_moist','dTe_C','dew_margin_C',
        'h_in_kJkgdry','h_out_kJkgdry','w_in','w_out',
        'temperature_113_Dome','temperature_301_outDome'
    ]:
        out[c + '_sm'] = out[c].rolling(win, min_periods=1).mean()

    ####Venting flag.
    #Venting flag plus persistence given in vent_persist_min (in normal dome shutter operations full openning happens in 2min20sec).
    raw_vent = (out['Dome_Shutter_Actual0'] >= 50.0) & (out['Dome_Shutter_Actual1'] >= 50.0)
    persist = max(int(vent_persist_min), 1)
    #True counts as 1; we require persist consecutive Trues to consider it is actually venting.
    venting = raw_vent.rolling(persist, min_periods=persist).sum().eq(persist).fillna(False)
    out['venting'] = venting
    #Rising edge: False -> True marks the first minute the dome is fully open.
    out['vent_open_event'] = venting & (~venting.shift(1, fill_value=False))
    #CHECK
    out.to_csv('venting_flags_events.csv')
    return out

####Visualization.
##Marking Sun altitude at 0deg and -12deg. Shading venting/dome open period.
##The smooth_window is a rolling mean window (in minutes) to reduce noise.
##min_separation: minimum time between reported crossings.
def _find_sun_crossings(sun_series, threshold_deg, smooth_window=5, min_separation='5min'):
    s = sun_series.copy()
    if smooth_window and smooth_window > 1:
        s = s.rolling(int(smooth_window), center=True, min_periods=1).mean()
    y = s - float(threshold_deg)
    idx = s.index
    times, last_kept = [], None

    for i in range(len(s) - 1):
        y0, y1 = y.iloc[i], y.iloc[i+1]
        if not np.isfinite(y0) or not np.isfinite(y1):
            continue
        if y0 == 0:
            t_cross = idx[i]
        elif y0 * y1 < 0 or y1 == 0:
            denom = (y0 - y1)
            if denom == 0:
                continue
            f = y0 / denom
            if f < 0 or f > 1:
                continue
            dt = idx[i+1] - idx[i]
            t_cross = idx[i] + pd.Timedelta(seconds=f * dt.total_seconds())
        else:
            continue
        if last_kept is None or (t_cross - last_kept) >= pd.Timedelta(min_separation):
            times.append(t_cross)
            last_kept = t_cross
    return times

def _add_sun_vlines(axs, times_0deg, times_m12deg,
                    color_0='red', color_m12='green',
                    ls_0='-', ls_m12='--', lw=1.2, alpha=0.9):
    if not isinstance(axs, (list, tuple, np.ndarray)):
        axs = [axs]
    labeled0 = labeled12 = True
    for ax in axs:
        for t in times_m12deg:
            ax.axvline(t, color=color_m12, linestyle=ls_m12, lw=lw, alpha=alpha,
                       label='Sun −12°')
            labeled12 = True
        for t in times_0deg:
            ax.axvline(t, color=color_0, linestyle=ls_0, lw=lw, alpha=alpha,
                       label='Sun 0°')
            labeled0 = True

def _shade_venting(axs, time_index, venting_bool):
    if not isinstance(axs, (list, tuple, np.ndarray)):
        axs = [axs]
    vm = pd.Series(venting_bool, index=time_index).astype(int)
    for ax in axs:
        ax.fill_between(time_index, 0, 1, where=vm>0, color='tab:gray', alpha=0.10,
                        transform=ax.get_xaxis_transform(), label='Venting')



def _sun_state_masks(sun: pd.Series, nautical_thr=-12.0):
    day_mask      = sun > 0.0
    nautical_mask = (sun <= 0.0) & (sun > float(nautical_thr))
    night_mask    = sun <= float(nautical_thr)
    return day_mask, nautical_mask, night_mask

def _state_labels():
    return [('Day (>0°)', 'day'), ('Twilight (−12…0°)', 'twilight'), ('Night (≤−12°)', 'night')]

    
###Time Series visualization. Smoothed by the rolling mean are plotted.
def plot_time_series_with_sun_facets(feat: pd.DataFrame, sun_col='SunAltitude',
                                     width_ratios=(1, 1, 3), figsize=None):
    if sun_col not in feat.columns:
        raise KeyError(f"Column '{sun_col}' not found.")
    sun = feat[sun_col]
    day_m, twi_m, night_m = _sun_state_masks(sun)
    states = [(day_m, 'Day (>0°)'), (twi_m, 'Twilight (−12…0°)'), (night_m, 'Night (≤−12°)')]

    # 3 rows: [Temperatures], [Enthalpies], [Δh & ΔT]
    if figsize is None:
        figsize = (18, 8)
    fig = plt.figure(figsize=figsize, constrained_layout=True)
    gs = fig.add_gridspec(nrows=3, ncols=3, width_ratios=width_ratios)

    # Collect series once
    T_in  = feat.get('temperature_113_Dome_sm', feat.get('temperature_113_Dome'))
    T_out = feat.get('temperature_301_outDome_sm', feat.get('temperature_301_outDome'))
    Td_out = feat.get('dewPoint_301')

    h_in  = feat.get('h_in_kJkgmoist_sm', feat.get('h_in_kJkgmoist'))
    h_out = feat.get('h_out_kJkgmoist_sm', feat.get('h_out_kJkgmoist'))

    dhe = feat.get('dhe_kJkg_moist_sm', feat.get('dhe_kJkg_moist'))
    dTe = feat.get('dTe_C_sm', feat.get('dTe_C'))

    # Row 1: Temps
    axes_row1 = []
    for j, (mask, title) in enumerate(states):
        ax = fig.add_subplot(gs[0, j])
        if T_in is not None:  ax.plot(feat.index[mask], T_in[mask], label='T_in')
        if T_out is not None: ax.plot(feat.index[mask], T_out[mask], label='T_out')
        if Td_out is not None: ax.plot(feat.index[mask], Td_out[mask], label='T_dew_out', alpha=0.8)
        ax.set_title(title)
        ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('Temperature [°C]')
        axes_row1.append(ax)
    axes_row1[0].legend(loc='best')

    # Row 2: Enthalpies
    axes_row2 = []
    for j, (mask, _) in enumerate(states):
        ax = fig.add_subplot(gs[1, j], sharey=(axes_row2[0] if axes_row2 else None))
        if h_in is not None:  ax.plot(feat.index[mask], h_in[mask], label='h_in')
        if h_out is not None: ax.plot(feat.index[mask], h_out[mask], label='h_out')
        ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('h [kJ/kg moist]')
        axes_row2.append(ax)
    axes_row2[0].legend(loc='best')

    # Row 3: Δh and ΔT
    axes_row3 = []
    for j, (mask, _) in enumerate(states):
        ax = fig.add_subplot(gs[2, j], sharey=(axes_row3[0] if axes_row3 else None))
        if dhe is not None: ax.plot(feat.index[mask], dhe[mask], label='Δh (moist)')
        if dTe is not None: ax.plot(feat.index[mask], dTe[mask], label='ΔT')
        ax.axhline(0.0, color='k', lw=0.8)
        ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('Δh [kJ/kg moist], ΔT [°C]')
        ax.set_xlabel('Time (UTC)')
        axes_row3.append(ax)
    axes_row3[0].legend(loc='best')

    # Optional: venting shading per panel (only where mask is True)
    vent = feat.get('venting', None)
    if vent is not None:
        for row_axes in (axes_row1, axes_row2, axes_row3):
            for ax, (mask, _) in zip(row_axes, states):
                vm = pd.Series(vent, index=feat.index).astype(bool)
                vm = vm & mask
                if vm.any():
                    ax.fill_between(feat.index[mask], 0, 1, where=vm[mask],
                                    color='tab:gray', alpha=0.10,
                                    transform=ax.get_xaxis_transform())

    # Sun crossing lines aren’t necessary in faceted panels because each panel is a single sun state,
    # but you can add them to the Day/Twilight borders if desired.

    return fig

###Lower level diagnostics.
##Current dew point marging is 5°C.
def plot_low_level_diagnostics_with_sun_facets(feat: pd.DataFrame, sun_col='SunAltitude',
                                               dew_margin_thr_C=5.0,
                                               width_ratios=(1, 1, 3), figsize=None):
    if sun_col not in feat.columns:
        raise KeyError(f"Column '{sun_col}' not found.")
    sun = feat[sun_col]
    day_m, twi_m, night_m = _sun_state_masks(sun)
    states = [(day_m, 'Day (>0°)'), (twi_m, 'Twilight (−12…0°)'), (night_m, 'Night (≤−12°)')]

    if figsize is None:
        figsize = (18, 8)
    fig = plt.figure(figsize=figsize, constrained_layout=True)
    gs = fig.add_gridspec(nrows=3, ncols=3, width_ratios=width_ratios)

    dm  = feat.get('dew_margin_C_sm', feat.get('dew_margin_C'))
    w_in = feat.get('w_in_sm', feat.get('w_in'))
    w_out= feat.get('w_out_sm', feat.get('w_out'))

    # Precompute ratio
    w_avg = (feat.get('w_in_sm', feat.get('w_in'))) + (feat.get('w_out_sm', feat.get('w_out')))
    if w_avg is not None:
        w_avg = w_avg / 2.0
        cp_mavg = cp_moist_kJ_per_kg_moistK(w_avg)
    else:
        cp_mavg = None
    dT = feat.get('dTe_C_sm', feat.get('dTe_C'))
    dT_safe = dT.where(dT.abs() >= 1e-1, np.nan) if dT is not None else None
    ratio = (feat.get('dhe_kJkg_moist_sm', feat.get('dhe_kJkg_moist')) / (cp_mavg * dT_safe)
             if (cp_mavg is not None and dT_safe is not None) else None)

    # Row 1: Dew margin
    axes1 = []
    for j, (mask, title) in enumerate(states):
        ax = fig.add_subplot(gs[0, j])
        if dm is not None:
            ax.plot(feat.index[mask], dm[mask], label='Dew margin')
            ax.axhline(dew_margin_thr_C, color='r', ls='--', lw=0.9, label=f'{dew_margin_thr_C:.0f}°C threshold')
        ax.set_title(title); ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('Dew margin [°C]')
        axes1.append(ax)
    axes1[0].legend(loc='best')

    # Row 2: Humidity ratios
    axes2 = []
    for j, (mask, _) in enumerate(states):
        ax = fig.add_subplot(gs[1, j], sharey=(axes2[0] if axes2 else None))
        if w_in is not None:  ax.plot(feat.index[mask], w_in[mask],  label='w_in')
        if w_out is not None: ax.plot(feat.index[mask], w_out[mask], label='w_out')
        ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('w [kg/kg dry]')
        axes2.append(ax)
    axes2[0].legend(loc='best')

    # Row 3: Ratio R
    axes3 = []
    for j, (mask, _) in enumerate(states):
        ax = fig.add_subplot(gs[2, j], sharey=(axes3[0] if axes3 else None))
        if ratio is not None:
            ax.plot(feat.index[mask], ratio[mask], label='R = Δh_moist / (c_p,moist_avg · ΔT)')
            ax.axhline(1.0, color='k', lw=0.8)
            ax.set_ylim(-5, 5)
        ax.grid(True, alpha=0.3)
        if j == 0: ax.set_ylabel('Ratio R')
        ax.set_xlabel('Time (UTC)')
        axes3.append(ax)
    axes3[0].legend(loc='best')

    # Venting shading
    vent = feat.get('venting', None)
    if vent is not None:
        for row_axes in (axes1, axes2, axes3):
            for ax, (mask, _) in zip(row_axes, states):
                vm = pd.Series(vent, index=feat.index).astype(bool)
                vm = vm & mask
                if vm.any():
                    ax.fill_between(feat.index[mask], 0, 1, where=vm[mask],
                                    color='tab:gray', alpha=0.10,
                                    transform=ax.get_xaxis_transform())

    return fig


#Ratio between Δh and effective moist-air specific heat (c_p)· ΔT. It is an approximation from 'Δh / (1.006·ΔT)'. Values around one shows very similar Δh,ΔT values. Very low and similar Δh,ΔT values are indeterminated.
    #c_p,moist_avg per timestamp using w_avg.


    
###Quadrant evaluation for enthalpy and temperature.
#Δh (moist) vs ΔT scatter colored by sun-altitude. The data points taken where venting are outlined.   
def plot_scatter_dh_dT_facets(
    feat: pd.DataFrame,
    sun_col='SunAltitude',
    mode='hex',                 # 'hex' | 'hist2d' | 'points'
    norm='log',                 # 'log' | 'linear'
    gridsize=50,                # for hex
    bins=80,                    # for hist2d
    cmap='cividis',             # colorblind-friendly
    mincnt=3,                   # suppress very sparse bins
    clip_percentiles=(1, 99),   # robust axis limits
    overlay_open=True,          # plot vent_open_event as stars
    overlay_vent=True,          # plot venting points as 'x'
    overlay_max_points=5000,    # subsample overlays if huge
    alpha_points=0.25,          # if mode='points'
    s_points=8,                 # if mode='points'
    show_colorbar=True,
    figsize=(15, 4.8)
):
    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    from matplotlib.colors import LogNorm, Normalize

    if sun_col not in feat.columns:
        raise KeyError(f"Column '{sun_col}' not found.")
    # Coerce to numeric
    x = pd.to_numeric(feat.get('dTe_C'), errors='coerce')
    y = pd.to_numeric(feat.get('dhe_kJkg_moist'), errors='coerce')
    valid_all = x.notna() & y.notna()

    if not valid_all.any():
        fig, axes = plt.subplots(1, 3, figsize=figsize, constrained_layout=True)
        for ax in np.atleast_1d(axes):
            ax.text(0.5, 0.5, "No data", ha='center', va='center'); ax.axis('off')
        return fig, axes

    # Sun-state masks
    sun = feat[sun_col]
    day_mask      = sun > 0.0
    twilight_mask = (sun <= 0.0) & (sun > -12.0)
    night_mask    = sun <= -12.0
    masks = [day_mask, twilight_mask, night_mask]
    titles = ['Day (>0°)', 'Twilight (−12…0°)', 'Night (≤−12°)']

    # Robust shared limits
    xv = x[valid_all].to_numpy(float)
    yv = y[valid_all].to_numpy(float)
    p_lo, p_hi = clip_percentiles
    try:
        xlim = (np.nanpercentile(xv, p_lo), np.nanpercentile(xv, p_hi))
        ylim = (np.nanpercentile(yv, p_lo), np.nanpercentile(yv, p_hi))
    except Exception:
        xlim = (float(np.nanmin(xv)), float(np.nanmax(xv)))
        ylim = (float(np.nanmin(yv)), float(np.nanmax(yv)))

    # Norm
    use_log = (norm == 'log')
    density_norm = (LogNorm() if use_log else Normalize())

    fig, axes = plt.subplots(1, 3, figsize=figsize, sharex=True, sharey=True, constrained_layout=True)
    axes = np.atleast_1d(axes)
    last_mappable = None  # colorbar handle

    # Overlays
    open_mask = feat.get('vent_open_event', pd.Series(False, index=feat.index)).astype(bool)
    vent_mask = feat.get('venting', pd.Series(False, index=feat.index)).astype(bool)

    for ax, state_mask, title in zip(axes, masks, titles):
        m = state_mask & valid_all
        if not m.any():
            ax.text(0.5, 0.5, "No data", ha='center', va='center')
            ax.set_title(title); ax.grid(True, alpha=0.3)
            continue

        xm = x[m].to_numpy(float); ym = y[m].to_numpy(float)

        # Clip to shared limits to prevent extreme values from compressing the view
        keep = (xm >= xlim[0]) & (xm <= xlim[1]) & (ym >= ylim[0]) & (ym <= ylim[1])
        xm = xm[keep]; ym = ym[keep]
        n_panel = int(xm.size)

        if n_panel == 0:
            ax.text(0.5, 0.5, "No data in range", ha='center', va='center')
            ax.set_title(title); ax.grid(True, alpha=0.3)
            continue

        if mode == 'hex':
            hb = ax.hexbin(xm, ym, gridsize=gridsize, mincnt=mincnt, cmap=cmap, norm=density_norm)
            last_mappable = hb
        elif mode == 'hist2d':
            h, _, _, img = ax.hist2d(xm, ym, bins=bins, cmap=cmap, norm=density_norm)
            last_mappable = img
        else:
            ax.scatter(xm, ym, s=s_points, alpha=alpha_points, c='tab:blue', edgecolors='none')

        # Overlays: openings and venting on top, with distinct markers
        # Subsample overlays to keep things readable
        if overlay_open:
            mo = (m & open_mask & valid_all)
            if mo.any():
                xs = x[mo].to_numpy(float); ys = y[mo].to_numpy(float)
                # Clip overlays to same limits
                ko = (xs >= xlim[0]) & (xs <= xlim[1]) & (ys >= ylim[0]) & (ys <= ylim[1])
                xs, ys = xs[ko], ys[ko]
                if xs.size > overlay_max_points:
                    idx = np.random.default_rng(0).choice(xs.size, size=overlay_max_points, replace=False)
                    xs, ys = xs[idx], ys[idx]
                ax.scatter(xs, ys, marker='*', s=60, facecolors='none', edgecolors='red',
                           linewidths=1.0, alpha=0.9, zorder=3, label='opening')

        # Cosmetics
        ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
        ax.set_title(f"{title} (n={n_panel})")
        ax.grid(True, alpha=0.3)

    # Shared axes/labels
    for ax in axes:
        ax.set_xlim(*xlim); ax.set_ylim(*ylim)
        ax.set_xlabel('ΔT = T_in − T_out [°C]')
    axes[0].set_ylabel('Δh [kJ/kg moist]')

    # Legend once (if overlays present)
    handles, labels = axes[-1].get_legend_handles_labels()
    if handles:
        axes[-1].legend(loc='best')

    # Single colorbar (density modes only)
    if show_colorbar and last_mappable is not None and mode in ('hex','hist2d'):
        cb = fig.colorbar(last_mappable, ax=axes.ravel().tolist(), fraction=0.035, pad=0.02)
        cb.set_label('counts (log scale)' if use_log else 'counts')

    return fig, axes

####Outputs for checking.
def additional_checks(feat, n=3):
    cols = ['RH_113_Dome','pressure_113_Dome_kPa',
            'RH_301_outDome','pressure_301_outDome_kPa',
            'w_in_sm','w_out_sm',
            'h_in_kJkgdry','h_in_kJkgmoist','h_in_kJkgmoist_sm','h_out_kJkgdry','h_out_kJkgmoist','h_out_kJkgmoist_sm'
            'dTe_C','dTe_C_sm']
    cols = [c for c in cols if c in feat.columns]
    print(feat[cols].head(n).to_string(float_format=lambda x: f"{x:.4f}"))
```

```python
####Based on nightly_report_aos_eas notebook.
####Asisted bt gpt-5 and claude4.5sonet.

###Getting data ranges.
def _make_dayobs_clause(dayobs_start=None, dayobs_end=None, dayobs_list=None):
    if dayobs_list is not None:
        vals = ",".join(str(int(d)) for d in sorted(set(dayobs_list)))
        return f"AND e.day_obs IN ({vals})"
    if dayobs_start is not None and dayobs_end is not None:
        return f"AND e.day_obs >= {int(dayobs_start)} AND e.day_obs <= {int(dayobs_end)}"
    if dayobs_start is not None:
        return f"AND e.day_obs = {int(dayobs_start)}"
    raise ValueError("Provide dayobs_list or dayobs_start (and optionally dayobs_end).")

###Data query from LSSTCam ccdvisit1, visit1 and exposure.
def get_iq_range(cdb_client, dayobs_start=None, dayobs_end=None, dayobs_list=None, detectors=(191,192,195,196,199,200,203,204)):
    day_clause = _make_dayobs_clause(dayobs_start, dayobs_end, dayobs_list)
    det_list = ",".join(str(int(d)) for d in detectors)

    query = f"""
        SELECT
            e.airmass AS airmass,
            e.dimm_seeing AS dimm,
            e.altitude AS elevation,
            e.azimuth AS azimuth,
            e.exposure_id AS visit_id,
            e.physical_filter AS band,
            e.day_obs AS day_obs,
            e.exp_midpt AS time,
            e.dimm_seeing AS seeing,
            e.science_program AS science_program,
            cql.psf_sigma,
            ccdvisit1.detector AS detector,
            q.psf_sigma_median AS psf_sigma_median,
            q.psf_sigma_min AS psf_sigma_min,
            q.psf_sigma_max AS psf_sigma_max,
            q.aos_fwhm AS aos_fwhm,
            q.donut_blur_fwhm AS donut_blur_fwhm,
            e.obs_end,
            e.obs_start,
            e.seq_num AS seq
        FROM
            cdb_lsstcam.ccdvisit1_quicklook AS cql,
            cdb_lsstcam.ccdvisit1 AS ccdvisit1,
            cdb_lsstcam.visit1 AS visit1,
            cdb_lsstcam.visit1_quicklook AS q,
            cdb_lsstcam.exposure AS e
        WHERE
            ccdvisit1.detector IN ({det_list})
            AND ccdvisit1.ccdvisit_id = cql.ccdvisit_id
            AND ccdvisit1.visit_id = visit1.visit_id
            AND ccdvisit1.visit_id = q.visit_id
            AND ccdvisit1.visit_id = e.exposure_id
            AND e.img_type IN ('science')
            {day_clause}
            AND e.airmass > 0
            AND e.physical_filter <> 'none'
    """
    return cdb_client.query(query).to_pandas()

#AND e.img_type IN ('science','acq','cwfs')

##Creation of the IQ sampling choosing one source of IQ input data.
def build_iq_timeseries_range(
    cdb_table,
    resample_freq='1min',              ##<<<<<<<<<<<<<<<<<<<<<------------ SAMPLING
    iq_priority=('aos_fwhm','donut_blur_fwhm','psf_fwhm','dimm'),    ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< WARNING. Order of priority.
    mix_sources=False,          #Default no mixing in IQ input data. 
    chosen_source='auto'         
):
    df = cdb_table.copy()
    #Data ranges
    df['obs_start'] = pd.to_datetime(df['obs_start'], utc=True, errors='coerce')
    df = df.dropna(subset=['obs_start']).set_index('obs_start').sort_index()
    #Calculte psf_fwhm from psf_sigma_median: LSSTCam FWHM = psf_sigma_median*2.355*0.2. Plate scale 0.2"/pix, constant 2.355=2 sqrt(2 ln(2))
    df['psf_fwhm'] = 2.355 * df['psf_sigma_median'] * float(0.2)
    
    #Aggregations per obs_start across the eight detectors.
    agg = {}
    for col in ['psf_fwhm','aos_fwhm','donut_blur_fwhm','dimm','band','science_program','seq','visit_id']:
        if col in df.columns:
            #Aggregations for categorical data using just the first field per detector.
            if col in ['band','science_program','seq','visit_id']:
                agg[col] = 'first'
            #Aggregations for the numerical data using the median.    
            else:
                agg[col] = 'median'
    
    # Make 'first' deterministic by sorting within groups
    df_sorted = df.sort_values(['visit_id'], kind='stable')
    
    # Collapse to one record per visit
    df_collapsed= (
        df_sorted
        .groupby(df.index)
        .agg(agg)
        .sort_index()
    )
       
    #per_exp = df.groupby(df.index).agg(agg).sort_index()
    per_exp = df_collapsed

    #Unified IQ [arcsec] using a priority chosen (to control AOS and Dome seeing contributions to IQ)
    #Checking data is present
    available = [c for c in iq_priority if c in per_exp.columns]
    if not available:
        raise ValueError("No IQ-like columns available after aggregation. "
                         "Expected one of: " + ", ".join(iq_priority))

    #IQ according to the priority chosen. Mixed and strict approaches. The strict approach does not search for another IQ column if missing data.
    if mix_sources:
        #Mixed: first non-null by priority.
        per_exp['IQ'] = np.nan
        for c in available:
            per_exp['IQ'] = per_exp['IQ'].where(per_exp['IQ'].notna(), per_exp[c])
        per_exp['IQ_source'] = 'mixed'
    else:
        #Strict: single-source: pick one column for the entire dataset.
        if chosen_source == 'auto':
            chosen = available[0]  #First available in priority order.
        else:
            if chosen_source not in per_exp.columns:
                raise ValueError(f"Chosen IQ source '{chosen_source}' not present. "
                                 f"Available: {available}")
            chosen = chosen_source

        #Getting the IQ for the chosen column.
        per_exp['IQ'] = per_exp[chosen]
        per_exp['IQ_source'] = chosen

        #Coverage check.
        coverage = per_exp['IQ'].notna().mean()
        print(f"\nBuilding IQ time series using as strict data source '{chosen}', with a data coverage of={coverage:.1%}")
    
    #This deals with the interpolation for missing values. This leaves a lot of NANs since there are minute gaps.
    #Soft interpolation for the sampling of IQ.
    #Resample to 1 min and optionally interpolate gaps.
    #iq_1m = per_exp[['IQ']].resample(resample_freq).median()
    #Minutes to interpolate across small gaps by 'interp_limit'.
    #if allow_time_interp:
    #    iq_1m['IQ'] = iq_1m['IQ'].interpolate(method='time', limit=int(interp_limit), limit_direction='both')
    return per_exp

###Interpolation for dealing NaN values in the 1min IQ data.
##Strategy 1: median uses only data inside the minute of sampling. This is the soft interpolation.
##Strategy 2: nearest pick the single closest exposure within a maximum tolerance (tol_s seconds).
def to_minute_series_strict(per_exp, strategy='nearest', tol_s=120):
    minute_index = pd.date_range(per_exp.index.min(), per_exp.index.max(), freq='1min', tz='UTC')    ###>>>>>>>>>>>>>>>>>>. WARNING Interpolation.
    if strategy == 'median':
        iq_1m = per_exp[['IQ']].resample('1min').median()
    elif strategy == 'nearest':
        iq_1m = per_exp[['IQ']].reindex(minute_index, method='nearest', tolerance=pd.Timedelta(seconds=tol_s))
    else:
        raise ValueError("strategy must be 'median', 'nearest'")
    return iq_1m
    

##Execution
if __name__ == "__main__":
    #ConsDB client
    os.environ["no_proxy"] = os.environ.get("no_proxy", "") + ",.consdb"
    cdb_client = ConsDbClient('http://consdb-pq.consdb:8080/consdb')

    #IQ time range request.
    dayobs_list = (
    pd.date_range('2025-07-01', '2025-09-21', freq='D')   ###>>>>>>>-----------DATA RANGE
      .strftime('%Y%m%d').astype(int).tolist()   # use .astype(str) if you prefer strings
    )

    #Check the range requested.
    print('\nRange requested:')
    print(len(dayobs_list), dayobs_list[:3], dayobs_list[-3:])
    
    #Data query for the night(range) of interest.
    raw = get_iq_range(cdb_client, dayobs_list=dayobs_list)

    #Raw per-night exposure counts.
    #print("\nRaw exposure counts by day_obs:")
    #print(raw.groupby('day_obs').size())
    raw.to_csv('raw.csv')
    print("\nRaw dataframe:")
    print(raw.sort_values(['obs_start'], kind='stable').head(10))
    
    #IQ time series.
    per_exp = build_iq_timeseries_range(
    raw,
    resample_freq='1min',       ##<<<<<<<<<<<<<<<<<<<-------- SAMPLING
    #iq_priority=('aos_fwhm','donut_blur_fwhm','psf_fwhm','dimm'),
    iq_priority=('donut_blur_fwhm','aos_fwhm','psf_fwhm','dimm'),
    mix_sources=False,
    #chosen_source='aos_fwhm')
    chosen_source='donut_blur_fwhm')
    #chosen_source='psf_fwhm')
        
    #IQ interpolation.
    iq_1m = to_minute_series_strict(per_exp, strategy='nearest', tol_s=120)  ######<<<<<<<<<<<<SD<<<<<<<<<<<<<<<<<<<<<<<<< WARNING IQ 1min interpolation.
    print('\nThe iq_1m dataset AFTER interpolation looks like:')
    print(iq_1m.head(10))
    print(iq_1m.tail(10))
    print('\nSize of iq_1m dataset:')
    print(iq_1m.size)
    print("\nnon-NaN IQ count:", iq_1m['IQ'].count())
    iq_1m.to_csv('iq_1m.csv')
    
    ####Outputs for checking.
    #Bounds.
    print("\nTime span with on-sky data (UTC):")
    print(iq_1m.index.min(), "→", iq_1m.index.max())
    

    #Unique seq count per science_program
    #cnt = (raw.groupby(['day_obs', 'science_program'])['seq']
    #         .nunique(dropna=True)            
    #         .rename('unique_seq_count')
    #         .reset_index())

    #pd.set_option('display.max_rows', None)
    #print(cnt)
    
    #Minute bins with one field or more.
    minute_has_exposure = per_exp.resample('1min').size() > 0  ##<<<<<<<<<<<<<<<<<<<-------- SAMPLING
    print("\nMinute bins with ≥1 exposure:", minute_has_exposure.mean()*100, "%")
    print(minute_has_exposure.head(10))
    
    #Minute sampling NaN range in IQ column.
    minute_non_nan_rate = iq_1m['IQ'].notna().mean()
    print(f"\nPer-minute non-NaN rate after interpolation with tolerance (tol_s): {minute_non_nan_rate:.1%}")
```

```python
####Asisted bt gpt-5 and claude4.5sonet.
#=========================
#Final check for having all the data in UTC.
#=========================
def ensure_utc_index(df, name='DataFrame'):
    if not isinstance(df.index, pd.DatetimeIndex):
        raise TypeError(f"{name} must have a DatetimeIndex. Got {type(df.index)}")
    if df.index.tz is None or str(df.index.tz) != 'UTC':
        # Confirm UTC.
        df = df.tz_localize('UTC') if df.index.tz is None else df.tz_convert('UTC')
    return df.sort_index()


#=========================
#Starting of the night definition as SunAltitude at -12 deg.
#This is needed to have a common point to compare different nights.
#=========================
def time_since_minus_twelve_start(feat, sun_col='SunAltitude'):
    if sun_col not in feat.columns:
        raise KeyError(f"Column '{sun_col}' not found in features.")
    sun = feat[sun_col]
    #SunAltitude at -12 deg mask.
    mask_night = (sun <= -12.0)
    #Number of nights (block with cumsum) with a SunAltitude at -12 deg event.
    block_id = (mask_night.astype(int).diff() == 1).cumsum()
    #Outside SunAltitude=-12 deg range data is NaN.
    block_id = block_id.where(mask_night, np.nan)

    t_night_min = pd.Series(np.nan, index=feat.index)
    for bid in sorted(block_id.dropna().unique()):
        block_mask = (block_id == bid)
        if not block_mask.any():
            continue
        #Here it is the t0 event for SunAltitude=-12deg for each night block.
        t0 = feat.index[block_mask][0]
        #Time since -12deg: Current time - t0 per night block in minutes.
        t_night_min.loc[block_mask] = (feat.index[block_mask] - t0).total_seconds() / 60.0

    return t_night_min, block_id

#=========================
#Combination of features for building the ventilation report.
#The input is Δh,ΔT,wind,SunAltitude and the venting events (from the dome shutter telemetry).
#Dew margin for safety considerations. It uses the IQ column selected previously at 1-minute cadence.
#tau_pre_min is the window time considered as pre-venting. tau_pre_min=0 means we open the dome and right way we start taking data.
#If t_iq_start − t_open >= tau_pre_min → scenario = Pre-venting (we vent some time before starting IQ).
#If t_iq_start − t_open <  tau_pre_min → scenario = Simultaneous: On-sky + venting (IQ started essentially while venting ramps).
#=========================
def build_vent_events(
    feat,
    iq_1m,
    tau_pre_min=20,                      #Minimum time you require between vent opening (t_open) and the first IQ measurement (t_iq_start) to classify an event as “Pre-venting”
    window_pre_min=60,                   #How far back (before t_open) we do data extraction for each event. The trajectory time grid is [t_open − window_pre_min, t_open + window_post_min] at 1-minute steps.
    window_post_min=120,                 #How far forward (after t_open) we do data extraction for each event. The trajectory time grid is [t_open − window_pre_min, t_open + window_post_min] at 1-minute steps.
    dew_thr_C=5.0,                       #October 2025 threshold is 5C.
    use_dew_gate_if_available=True,
    vent_persist_min=None,               #2 min for normal operations.
    vent_threshold=50.0,                  #In case we check venting at dome shutter opening less that 50%.
    max_delta_min_filter=300,     # NEW: cap Δstart (minutes)
    require_same_night=True,      # NEW: only accept first IQ from same nautical night
    night_buffer_min=15           # NEW: small grace after night end (minutes)
):

    #Final check on UTC indices.
    feat = ensure_utc_index(feat, 'feat')
    iq_1m = ensure_utc_index(iq_1m, 'iq_1m')
    
        # Right after the UTC checks, before using opens_idx
    try:
        # Reuse your existing helper on feat to get per-timestamp night blocks
        _, _night_id = time_since_minus_twelve_start(feat, 'SunAltitude')
    except KeyError:
        _night_id = pd.Series(np.nan, index=feat.index)

    
    #Derive vent_open_event if only 'venting' exists
    if 'vent_open_event' not in feat.columns:
        if 'venting' in feat.columns:
            feat['vent_open_event'] = feat['venting'] & (~feat['venting'].shift(1, fill_value=False))
        else:
            raise KeyError("build_vent_events: need 'vent_open_event' or 'venting' to detect openings.")

    #Dew point gate for safety opening.
    dew_ok_series = pd.Series(True, index=feat.index)
    if use_dew_gate_if_available and 'dew_margin_C_sm' in feat.columns:
        dew_ok_series = feat['dew_margin_C_sm'] >= float(dew_thr_C)

    #Pick open events (rising edges) during dew-safe times.
    opens_idx = feat.index[feat['vent_open_event'].fillna(False) & dew_ok_series.reindex(feat.index, fill_value=False)]

    #Fill missing values with NaN (ensure columns exist).
    for c in ['dhe_kJkg_moist_sm', 'dTe_C_sm', 'dew_margin_C_sm', 'WindSpeed_301', 'SunAltitude', 'venting']:
        if c not in feat.columns:
            feat[c] = np.nan

    records = []
    traj_rows = []

    #Define the start of on-sky and the time to consider the pre-venting with delta_min_start via tau_pre_min.
    for t_open in opens_idx:
        #First valid IQ at/after t_open.
        # Determine search cutoff for first IQ
        cut_time = t_open + pd.Timedelta(minutes=float(max_delta_min_filter))
        
        # --- inside build_vent_events(), in the for t_open in opens_idx loop ---
        
        # 1) Bound the search window
        cut_time = t_open + pd.Timedelta(minutes=float(max_delta_min_filter))
        
        if require_same_night and (t_open in _night_id.index):
            bid = _night_id.loc[t_open]
            if pd.notna(bid):
                block_mask = (_night_id == bid)
                if block_mask.any():
                    t_night_end = _night_id.index[block_mask][-1]
                    cut_time = min(cut_time, t_night_end + pd.Timedelta(minutes=float(night_buffer_min)))
        
        # 2) First IQ within the allowed window (causal or nearest per your choice)
        iq_after = iq_1m.loc[t_open:cut_time]
        iq_valid = iq_after['IQ'].dropna()
        
        # 3) If none, skip this open event
        if iq_valid.empty:
            # Optionally record exclusion reason here
            # excluded.append({'t_open': t_open, 'reason': 'no_IQ_in_window'})
            continue
        
        # 4) Now it's safe to assign and use t_iq_start
        t_iq_start = iq_valid.index[0]
        delta_min_start = (t_iq_start - t_open).total_seconds() / 60.0
        
        # 5) Final guard (redundant given cut_time, but harmless)
        if delta_min_start > float(max_delta_min_filter):
            # excluded.append({'t_open': t_open, 'reason': 'delta_exceeds_cap'})
            continue
        
        # 6) Scenario label
        scenario = 'Pre-venting' if delta_min_start >= float(tau_pre_min) else 'On-sky'
        #Initial conditions at t_open.
        row0 = feat.loc[t_open]
        init = {
            't_open': t_open,
            't_iq_start': t_iq_start,
            'scenario': scenario,
            'delta_min_start': float(delta_min_start),
            'dh0_kJkg_moist': float(row0['dhe_kJkg_moist_sm']) if pd.notna(row0['dhe_kJkg_moist_sm']) else np.nan,
            'dT0_C': float(row0['dTe_C_sm']) if pd.notna(row0['dTe_C_sm']) else np.nan,
            'M0_C': float(row0['dew_margin_C_sm']) if pd.notna(row0['dew_margin_C_sm']) else np.nan,
            'wind0_mps': float(row0['WindSpeed_301']) if pd.notna(row0['WindSpeed_301']) else np.nan,
            'sun0_deg': float(row0['SunAltitude']) if pd.notna(row0['SunAltitude']) else np.nan,
        }
        records.append(init)

        #Build aligned blocks: [t_open - pre, t_open + post] at 1-min grid (UTC).
        t0_pre = t_open - pd.Timedelta(minutes=int(window_pre_min))
        t1_post = t_open + pd.Timedelta(minutes=int(window_post_min))
        grid = pd.date_range(t0_pre, t1_post, freq='1min', tz='UTC')

        blk = pd.DataFrame(index=grid)
        #Append environmental data from feat (left join).
        blk = blk.join(feat[['dhe_kJkg_moist_sm','dTe_C_sm','dew_margin_C_sm','WindSpeed_301','SunAltitude','venting']], how='left')
        #Append IQ by nearest-with-tolerance to handle sparse cadence and fractional seconds in IQ.
        blk['IQ'] = iq_1m['IQ'].reindex(grid, method='nearest', tolerance=pd.Timedelta(seconds=120))  ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< WARNING INTERPOLATION IN IQ.
        
        #Time since dome shutter open and since first IQ.
        blk['s_open_min'] = (blk.index - t_open).total_seconds() / 60.0     #Time since dome opened. 
        blk['s_iq_min'] = (blk.index - t_iq_start).total_seconds() / 60.0   #Time since first IQ.

        #Hard-mask IQ before first IQ (prevents any pre-IQ trend from appearing anywhere) <<<<<-----Check the ACQ images!!!!!
        blk.loc[blk['s_iq_min'] < 0, 'IQ'] = np.nan

        #Baseline IQ during a specific period: median in 10 min before first IQ (or IQ at start).
        pre_iq_window = iq_1m.loc[max(t_open, t_iq_start - pd.Timedelta(minutes=10)):t_iq_start]
        if pre_iq_window['IQ'].notna().any():
            iq_base = float(pre_iq_window['IQ'].median())
        else:
            iq_base = float(iq_1m.loc[t_iq_start, 'IQ'])
        blk['IQ_base'] = iq_base

        #ΔIQ: IQ - baseline (since IQ at the beginning of the night usually is bad, negative ΔIQ values means IQ improvement).
        blk['dIQ'] = blk['IQ'] - iq_base

        #Recording the open event and the scenario.
        blk['event_open'] = t_open
        blk['scenario'] = scenario

        traj_rows.append(blk.reset_index().rename(columns={'index':'time'}))

    #Per-event table indexed by t_open in UTC.
    events = pd.DataFrame.from_records(records).set_index('t_open').sort_index() if records else pd.DataFrame(columns=[
        't_iq_start','scenario','delta_min_start','dh0_kJkg_moist','dT0_C','M0_C','wind0_mps','sun0_deg'
    ])
    traj = pd.concat(traj_rows, axis=0).sort_values(['event_open','time']) if traj_rows else pd.DataFrame()

    #Reporting.
    print(f"\nDetected {len(opens_idx)} dome shutter open event(s) (dew-point safe).")
    #if len(opens_idx) > 0:
    #    print("First open time(s):", [str(t) for t in list(opens_idx[:25])]) ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< DEBUGGING.
    #CHECK
    events.to_csv('events.csv')
    return events, traj


#=========================
#Error bands definition for IQ and ΔIQ (p16/p50/p84)
#=========================
def _bands_safe(df_grp):
    iq = df_grp['IQ'].to_numpy(dtype=float)
    iq = iq[~np.isnan(iq)]
    d  = df_grp['dIQ'].to_numpy(dtype=float)
    d  = d[~np.isnan(d)]

    if iq.size:
        IQ_med = float(np.median(iq)); IQ_lo = float(np.percentile(iq, 16)); IQ_hi = float(np.percentile(iq, 84))
    else:
        IQ_med = IQ_lo = IQ_hi = np.nan

    if d.size:
        dIQ_med = float(np.median(d)); dIQ_lo = float(np.percentile(d, 16)); dIQ_hi = float(np.percentile(d, 84))
    else:
        dIQ_med = dIQ_lo = dIQ_hi = np.nan

    return pd.Series({'IQ_med': IQ_med, 'IQ_lo': IQ_lo, 'IQ_hi': IQ_hi,
                      'dIQ_med': dIQ_med, 'dIQ_lo': dIQ_lo, 'dIQ_hi': dIQ_hi})


#=========================
#Visualization of IQ trajectories (with bands) by scenario
#=========================
def plot_iq_trajectories(traj, by='s_open_min', window_pre_min=None, window_post_min=None,
                         iq_metric_name='IQ', tau_pre_min=None,
                         show=True, return_fig=True):
     #Checking empty data.
    if traj.empty:
        print("No trajectories to plot."); 
        return None if return_fig else None
    if by not in ['s_open_min','s_iq_min']:
        raise ValueError("by must be 's_open_min' or 's_iq_min'")

    #In plot_iq_trajectories before banding:
    tmp = traj[['scenario', by, 'IQ', 'dIQ']].dropna(subset=[by]).copy()
    if by == 's_open_min':
        tmp.loc[tmp['s_open_min'] < 0, ['IQ','dIQ']] = np.nan
    tmp[by] = tmp[by].round(0)
    
    agg = tmp.groupby(['scenario', by], observed=True).apply(_bands_safe).reset_index()

    '''##Both figures. Trajectory and ΔIQ
    fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=False)
    for scen, df_s in agg.groupby('scenario', observed=True):
        axs[0].plot(df_s[by], df_s['IQ_med'], label=f'{scen}')
        axs[0].fill_between(df_s[by], df_s['IQ_lo'], df_s['IQ_hi'], alpha=0.2)
        axs[1].plot(df_s[by], df_s['dIQ_med'], label=f'{scen}')
        axs[1].fill_between(df_s[by], df_s['dIQ_lo'], df_s['dIQ_hi'], alpha=0.2)

    axs[0].set_ylabel('IQ [arcsec]'); axs[0].grid(True, alpha=0.3); axs[0].legend(loc='best')
    axs[1].axhline(0, color='k', lw=0.8)
    axs[1].set_ylabel('ΔIQ vs baseline [arcsec]')
    axs[1].set_xlabel('Minutes since ' + ('vent open' if by=='s_open_min' else 'first IQ'))
    axs[1].grid(True, alpha=0.3)
    '''
    #Single figure
    fig, axs = plt.subplots(1, 1, figsize=(10, 4))
    for scen, df_s in agg.groupby('scenario', observed=True):
        axs.plot(df_s[by], df_s['IQ_med'], label=f'{scen}')
        axs.fill_between(df_s[by], df_s['IQ_lo'], df_s['IQ_hi'], alpha=0.2)
    axs.set_ylabel('IQ [arcsec]'); axs.grid(True, alpha=0.3); axs.legend(loc='best')
    axs.set_xlabel('Minutes since ' + ('dome shutter open' if by=='s_open_min' else 'first IQ'))
    axs.grid(True, alpha=0.3)
    axs.set_ylim(0.7,2.5)

    title_bits = []
    if iq_metric_name:
        title_bits.append(f"IQ metric: {iq_metric_name}")
    if tau_pre_min is not None:
        title_bits.append(f"Venting time (vt) ={tau_pre_min} min")
    if title_bits:
        fig.suptitle(" | ".join(title_bits), y=0.98, fontsize=11)

    if show:
        plt.tight_layout(); plt.show()
    else:
        plt.tight_layout()

    return fig if return_fig else None


    
#=========================
#Time it takes to improve IQ certain X arcsec from baseline.
#Out dataframe calculates the improvement depending on the scenario in place.
#If by == 's_open_min' a column 't_hit_s_open_min' is created.
#If by == 's_iq_min'   a column 't_hit_s_iq_min' is created.
#=========================
def time_to_improvement(traj, by='s_open_min', improve_arcsec=0.2):   ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< WARNING 0.1 arcsec improvement either direction.
    out_col = 't_hit_s_open_min' if by == 's_open_min' else 't_hit_s_iq_min'
    #Checking empty data.
    if traj.empty:
        return pd.DataFrame(columns=[out_col])

    if by not in ['s_open_min','s_iq_min']:
        raise ValueError("by must be 's_open_min' or 's_iq_min'")

    results = []
    #Group by event and scenario. observed=True avoids phantom groups for categories with no data.
    for (t_open, scen), df_e in traj.groupby(['event_open','scenario'], observed=True):
        df_e = df_e.sort_values(by)
        #Improvement: dIQ <= -abs(threshold).
        mask = (df_e['dIQ'] <= -abs(improve_arcsec))
        t_hit = float(df_e.loc[mask, by].iloc[0]) if mask.any() else np.nan
        results.append({'t_open': t_open, 'scenario': scen, out_col: t_hit})
    out = pd.DataFrame(results).set_index('t_open').sort_index()
    return out

    
#============================
#Reformatting: 't_hit_s_open_min_min' -> 't_hit_s_open_min'.
#============================
def _normalize_t_hit_columns(df):
    rename_map = {}
    for c in df.columns:
        if c.endswith('_min_min'):
            rename_map[c] = c[:-4]  #Drop the extra '_min'.
    return df.rename(columns=rename_map) if rename_map else df


#=========================
#Reformatting: Adds minutes since SunAltitude at -12 deg) per night, and a night_id to the events table.
#=========================
def attach_time_of_night_to_events(events, feat, max_tolerance='5min'):
    feat = ensure_utc_index(feat, 'feat')
    t_night_min, night_id = time_since_minus_twelve_start(feat, 'SunAltitude')
    tno = t_night_min.reindex(events.index, method='nearest', tolerance=pd.Timedelta(max_tolerance))
    nid = night_id.reindex(events.index, method='nearest', tolerance=pd.Timedelta(max_tolerance))
    out = events.copy()
    out['t_night_min_at_open'] = tno
    out['night_id'] = nid
    return out


# =========================
# Per-night summaries
# =========================
def summarize_by_night(events, t_open_hit, t_iq_hit):
    """
    Build a per-night (night_id) summary:
      - counts, scenario mix
      - median delta_min_start
      - median t_hit_s_open_min, t_hit_s_iq_min
      - medians of initial conditions (dh0, dT0, M0, wind0)
    Requires 'night_id' in events and t_open_hit / t_iq_hit having standardized columns.
    """
    ev = events.copy()

    # Normalize accidental names if needed
    t_open_hit = _normalize_t_hit_columns(t_open_hit)
    t_iq_hit   = _normalize_t_hit_columns(t_iq_hit)

    cols_open = [c for c in t_open_hit.columns if c.startswith('t_hit_')]
    cols_iq   = [c for c in t_iq_hit.columns   if c.startswith('t_hit_')]

    # Join only missing t_hit columns to avoid overlap errors
    miss_open = [c for c in cols_open if c not in ev.columns]
    if miss_open:
        ev = ev.join(t_open_hit[miss_open], how='left')
    miss_iq = [c for c in cols_iq if c not in ev.columns]
    if miss_iq:
        ev = ev.join(t_iq_hit[miss_iq], how='left')

    # Guarantee expected columns exist
    for col in ['t_hit_s_open_min', 't_hit_s_iq_min']:
        if col not in ev.columns:
            ev[col] = np.nan

    if 'night_id' not in ev.columns:
        print("summarize_by_night: 'night_id' not found in events. Call attach_time_of_night_to_events first.")
        return pd.DataFrame()

    def scen_prop(g):
        vc = g['scenario'].value_counts(normalize=True)
        return pd.Series({
            'n_events': len(g),
            'p_pre_venting': float(vc.get('Pre-venting', 0.0)),
            'p_on_sky_venting': float(vc.get('On-sky', 0.0)),
            'delta_min_start_med': float(np.nanmedian(g['delta_min_start'])),
            't_hit_s_open_min_med': float(np.nanmedian(g['t_hit_s_open_min'])),
            't_hit_s_iq_min_med': float(np.nanmedian(g['t_hit_s_iq_min'])),
            'dh0_med': float(np.nanmedian(g['dh0_kJkg_moist'])),
            'dT0_med': float(np.nanmedian(g['dT0_C'])),
            'M0_med': float(np.nanmedian(g['M0_C'])),
            'wind0_med': float(np.nanmedian(g['wind0_mps'])),
        })

    night_summary = ev.groupby('night_id', dropna=False).apply(scen_prop).reset_index()
    return night_summary

# =========================
# tau value
# =========================
def suggest_tau_from_cdf(events, night_window=(0, 120), quantile=0.80, tno_col='t_night_min_at_open'):
    if 'delta_min_start' not in events.columns:
        raise KeyError("events must have delta_min_start. Build events first.")
    if tno_col not in events.columns:
        print(f"Warning: {tno_col} not in events; using all events.")
        sub = pd.to_numeric(events['delta_min_start'], errors='coerce').dropna()
    else:
        mask = (events[tno_col] >= night_window[0]) & (events[tno_col] < night_window[1])
        sub = pd.to_numeric(events.loc[mask, 'delta_min_start'], errors='coerce').dropna()
    if sub.empty:
        return np.nan
    return float(np.nanpercentile(sub, quantile*100.0))

def plot_tau_hist_cdf(events, night_window=(0, 120), tno_col='t_night_min_at_open', tau_lines=(10,20,30), binsize=1.0):
    if 'delta_min_start' not in events.columns:
        raise KeyError("events must have delta_min_start. Build events first.")
    if tno_col in events.columns:
        mask = (events[tno_col] >= night_window[0]) & (events[tno_col] < night_window[1])
        sub = pd.to_numeric(events.loc[mask, 'delta_min_start'], errors='coerce').dropna()
    else:
        sub = pd.to_numeric(events['delta_min_start'], errors='coerce').dropna()
    if sub.empty:
        print("No events to plot.")
        return
    vmax = max(5.0, float(np.nanpercentile(sub, 99)))
    edges = np.arange(0.0, vmax + binsize, binsize)
    fig, axs = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)
    axs[0].hist(sub, bins=edges, color='tab:blue', alpha=0.75, edgecolor='white')
    axs[0].set_xlabel('Δstart = t_iq_start − t_open (min)'); axs[0].set_ylabel('Count'); axs[0].grid(True, alpha=0.3)
    s = np.sort(sub.values); cdf = np.arange(1, len(s)+1)/len(s)
    axs[1].plot(s, cdf, color='tab:purple', lw=2); axs[1].set_ylim(0, 1.01)
    axs[1].set_xlabel('Δstart (min)'); axs[1].set_ylabel('CDF'); axs[1].grid(True, alpha=0.3)
    for tau in (tau_lines or []):
        axs[0].axvline(tau, color='red', ls='--', lw=1.2)
        axs[1].axvline(tau, color='red', ls='--', lw=1.2)
    plt.show()

def reclassify_for_tau(events, traj, tau):
    ev = events.copy()
    ev['scenario'] = np.where(pd.to_numeric(ev['delta_min_start'], errors='coerce') >= float(tau),
                              'Pre-venting', 'On-sky')
    if traj is None or traj.empty:
        return ev, traj
    m = ev['scenario'].to_dict()  # key=t_open, value=scenario
    tr = traj.copy()
    tr['scenario'] = tr['event_open'].map(m)
    return ev, tr


def _median_diq_at_minutes(traj_tau, by='s_open_min', minutes=(30, 60)):
    out = {}
    if traj_tau.empty:
        for m in minutes: out[f'dIQ_med_diff_{int(m)}'] = np.nan
        return out
    df = traj_tau.copy()
    df = df[[by, 'scenario', 'dIQ']].dropna(subset=[by]).copy()
    df = df[df[by] >= 0]
    df[by] = df[by].round(0)
    grp = df.groupby(['scenario', by])['dIQ'].median().unstack(0)  # columns by scenario
    for m in minutes:
        if m in grp.index and {'Pre-venting','On-sky'}.issubset(grp.columns):
            out[f'dIQ_med_diff_{int(m)}'] = float(grp.loc[m, 'Pre-venting'] - grp.loc[m, 'On-sky'])
        else:
            out[f'dIQ_med_diff_{int(m)}'] = np.nan
    return out

def _median_t_hit_by_scenario(traj_tau, improve_arcsec=0.1):
    if traj_tau.empty:
        return {'t_hit_open_pre_med': np.nan, 't_hit_open_on_med': np.nan,
                't_hit_iq_pre_med': np.nan,   't_hit_iq_on_med': np.nan}
    t_open_hit = time_to_improvement(traj_tau, by='s_open_min', improve_arcsec=improve_arcsec)
    t_iq_hit   = time_to_improvement(traj_tau, by='s_iq_min',   improve_arcsec=improve_arcsec)
    res = {}
    for scen, lab in [('Pre-venting','Pre-venting'), ('On-sky','On-sky')]:
        res[f't_hit_open_{lab}_med'] = float(np.nanmedian(t_open_hit.loc[t_open_hit['scenario']==scen, 't_hit_s_open_min'])) if not t_open_hit.empty else np.nan
        res[f't_hit_iq_{lab}_med']   = float(np.nanmedian(t_iq_hit.loc[t_iq_hit['scenario']==scen, 't_hit_s_iq_min']))     if not t_iq_hit.empty else np.nan
    return res

def _counts_by_scenario(events_tau):
    vc = events_tau['scenario'].value_counts()
    return {
        'n_pre': int(vc.get('Pre-venting', 0)),
        'n_on':  int(vc.get('On-sky', 0)),
        'n_all': int(len(events_tau)),
    }

def sweep_tau(events, traj, tau_grid, eval_minutes=(30, 60), improve_arcsec=0.1, min_per_scen=5):
    rows = []
    for tau in tau_grid:
        ev_tau, tr_tau = reclassify_for_tau(events, traj, tau)
        sep = _median_diq_at_minutes(tr_tau, by='s_open_min', minutes=eval_minutes)
        thit = _median_t_hit_by_scenario(tr_tau, improve_arcsec=improve_arcsec)
        cnt = _counts_by_scenario(ev_tau)
        # Simple score: prioritize larger separation at 30/60 (more negative = larger improvement for Pre-venting if negative),
        # but keep scenarios sufficiently populated.
        penalty = 0.0
        if cnt['n_pre'] < min_per_scen or cnt['n_on'] < min_per_scen:
            penalty = 1.0  # penalize low counts
        score = 0.0
        for m in eval_minutes:
            v = sep.get(f'dIQ_med_diff_{int(m)}', np.nan)
            # We want Pre-venting to have more negative dIQ than On-sky (i.e., v is more negative).
            # Convert to positive "goodness": -(v). Ignore NaN.
            if np.isfinite(v): score += -v
        score -= penalty  # apply penalty
        row = {'tau': float(tau), 'score': float(score)}
        row.update(cnt); row.update(sep); row.update(thit)
        rows.append(row)
    return pd.DataFrame(rows).sort_values('tau')

def plot_tau_sweep(df_scores, eval_minutes=(30,60)):
    if df_scores.empty:
        print("No sweep results to plot.")
        return
    fig, axs = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)
    ax = axs[0,0]
    ax.plot(df_scores['tau'], df_scores['score'], marker='o')
    ax.set_title('Score vs tau'); ax.set_xlabel('tau_pre_min [min]'); ax.grid(True, alpha=0.3)

    ax = axs[0,1]
    for m in eval_minutes:
        col = f'dIQ_med_diff_{int(m)}'
        if col in df_scores.columns:
            ax.plot(df_scores['tau'], df_scores[col], marker='o', label=f'Δ(dIQ_med) @ {m} min')
    ax.axhline(0, color='k', lw=0.8); ax.grid(True, alpha=0.3); ax.legend(loc='best')
    ax.set_title('Median Δ(dIQ) (Pre − On)')

    ax = axs[1,0]
    ax.plot(df_scores['tau'], df_scores['n_pre'], label='Pre-venting', color='tab:orange')
    ax.plot(df_scores['tau'], df_scores['n_on'],  label='On-sky', color='tab:blue')
    ax.set_title('Counts by scenario vs tau'); ax.set_xlabel('Ventilation time [min]'); ax.set_ylabel('N Dome shutter openings')
    ax.grid(True, alpha=0.3); ax.legend(loc='best')

    ax = axs[1,1]
    for c, lab in [('t_hit_open_pre_med','open pre'), ('t_hit_open_on_med','open on'),
                   ('t_hit_iq_pre_med','iq pre'),   ('t_hit_iq_on_med','iq on')]:
        if c in df_scores.columns:
            ax.plot(df_scores['tau'], df_scores[c], marker='o', label=lab)
    ax.set_title('Median time-to-improvement vs tau'); ax.set_xlabel('tau_pre_min [min]')
    ax.grid(True, alpha=0.3); ax.legend(loc='best')
    plt.show()

def plot_tau_grid_trajectories(
    traj,
    tau_grid,
    window_post_min=120,
    by='s_open_min',
    include_ci=True,                 # draw p16–p84 bands
    colors=None,                     # optional color override: {'Pre-venting': ..., 'On-sky + venting': ...}
    figsize=None,                    # default scales with number of taus
    show=True
):
    """
    Faceted grid: one subplot per tau, with BOTH scenarios overlapped in each subplot.
    - Classifies scenario per-event via delta_min_start >= tau
    - Aggregates IQ by rounded minute: median and p16–p84
    - Plots both scenarios in the same axes for direct comparison

    Requires traj to have columns: ['event_open', 'delta_min_start', by, 'IQ'].
    If 'delta_min_start' is missing, merge it from events before calling.
    """

    if traj.empty:
        print("No trajectories.")
        return None, None

    if 'delta_min_start' not in traj.columns:
        raise ValueError("traj missing 'delta_min_start'; merge it from events (left_on='event_open', right_index=True).")

    if by not in ('s_open_min', 's_iq_min'):
        raise ValueError("by must be 's_open_min' or 's_iq_min'")

    # Colors
    colors = colors or {'Pre-venting': 'tab:orange', 'On-sky': 'tab:blue'}
    scen_order = ['Pre-venting', 'On-sky']

    n = len(tau_grid)
    if figsize is None:
        figsize = (max(20, 3.6*n), 4)

    fig, axes = plt.subplots(1, n, figsize=figsize, sharex=True, sharey=True, constrained_layout=True)
    if n == 1:
        axes = np.array([axes])

    # Common x-range limits
    xlo, xhi = 0.0, float(window_post_min)

    for j, tau in enumerate(tau_grid):
        ax = axes[j]

        # Reclassify per tau at the EVENT level, then map to traj rows
        tr = traj.copy()
        # Ensure numeric
        dms = pd.to_numeric(tr['delta_min_start'], errors='coerce')
        tr['scenario'] = np.where(dms >= float(tau), 'Pre-venting', 'On-sky')

        # Count events per scenario (not rows)
        ev_tau = tr[['event_open', 'scenario']].drop_duplicates()
        vc = ev_tau['scenario'].value_counts()
        n_pre = int(vc.get('Pre-venting', 0))
        n_on  = int(vc.get('On-sky', 0))

        # Prepare data by time
        tmp = tr[[by, 'scenario', 'IQ']].dropna(subset=[by]).copy()
        tmp = tmp[tmp[by].between(0, window_post_min)]
        tmp[by] = tmp[by].round(0)

        # Aggregate: median and 16/84 percentiles
        if tmp.empty:
            ax.set_title(f"tau={int(tau)} (no data)")
            ax.grid(True, alpha=0.3)
            continue

        agg = (
            tmp.groupby(['scenario', by], observed=True)['IQ']
               .agg(IQ_med='median',
                    IQ_lo=lambda x: np.nanpercentile(x, 16),
                    IQ_hi=lambda x: np.nanpercentile(x, 84))
               .reset_index()
        )

        # Plot both scenarios overlapped
        any_plotted = False
        for scen in scen_order:
            g = agg[agg['scenario'] == scen]
            if g.empty:
                continue
            any_plotted = True
            c = colors.get(scen, None)
            ax.plot(g[by], g['IQ_med'], label=scen, color=c, lw=2)
            if include_ci:
                ax.fill_between(g[by], g['IQ_lo'], g['IQ_hi'], color=c, alpha=0.20, linewidth=0)

        ax.set_xlim(xlo, xhi)
        if by == 's_open_min':
            ax.set_xlabel('Minutes since vent open')
        else:
            ax.set_xlabel('Minutes since first IQ')

        # Title shows tau and event counts in each scenario
        ax.set_title(f"tau={int(tau)} | Pre={n_pre}, On={n_on}")
        ax.grid(True, alpha=0.3)

        # Only add legend on the first subplot that actually plotted data
        if j == 0 and any_plotted:
            ax.legend(loc='best')

    # Y label on the leftmost subplot
    axes[0].set_ylabel('IQ [arcsec]')
    if show:
        plt.show()
    return fig, axes



#=========================
#Main function.
#=========================
def run_venting(feat, iq_1m, tau_pre_min=20, window_pre_min=60, window_post_min=120,
                dew_thr_C=5.0, iq_metric_name='aos_fwhm'):
    #Checking common UTC.
    feat = ensure_utc_index(feat, 'feat')
    iq_1m = ensure_utc_index(iq_1m, 'iq_1m')
    print("Features index tz:", feat.index.tz, " | iq_1m index tz:", iq_1m.index.tz)

    #Time since Sun at -12deg. The column is stored as t_night_min.
    try:
        feat['t_night_min'], _ = time_since_minus_twelve_start(feat, 'SunAltitude')
    except KeyError:
        feat['t_night_min'] = np.nan

    #Build events and trajectories.
    events, traj = build_vent_events(
        feat, iq_1m,
        tau_pre_min=tau_pre_min,
        window_pre_min=window_pre_min,
        window_post_min=window_post_min,
        dew_thr_C=dew_thr_C,
        use_dew_gate_if_available=True
    )

    #Checking empty data (closed nights).
    if events.empty:
        print("No dome shutter open events found under dew-safe conditions.")
        return events, traj, pd.DataFrame(), pd.DataFrame()

    #Adds minutes since SunAltitude at -12 deg) per night, and a night_id to the events table summaries.
    events_with_tno = attach_time_of_night_to_events(events, feat, max_tolerance='2min')

    print("\nEvent counts by scenario:")
    print(events_with_tno['scenario'].value_counts(dropna=True))

    print("\nEvent initial conditions (head):")
    pd.set_option('display.max_rows', None)
    #print(events_with_tno[['scenario','delta_min_start','dh0_kJkg_moist','dT0_C','M0_C','wind0_mps','sun0_deg']].head(25))  ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< DEBUGGING.
    print(events_with_tno[['scenario','delta_min_start','dh0_kJkg_moist','dT0_C','M0_C','wind0_mps','sun0_deg']])  ######<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<< DEBUGGING.
    
    #Plot trajectories vs time since vent open and since first IQ (with labels).
    plot_iq_trajectories(traj, by='s_open_min',
                         window_pre_min=window_pre_min, window_post_min=window_post_min,
                         iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min)

    plot_iq_trajectories(traj, by='s_iq_min',
                         window_pre_min=window_pre_min, window_post_min=window_post_min,
                         iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min)


    #Time-to-improvement (0.1").
    t_open_hit = _normalize_t_hit_columns(time_to_improvement(traj, by='s_open_min', improve_arcsec=0.1))
    t_iq_hit   = _normalize_t_hit_columns(time_to_improvement(traj, by='s_iq_min',   improve_arcsec=0.1))

    #Keep only the numeric columns.
    cols_open = [c for c in t_open_hit.columns if c.startswith('t_hit_')]
    cols_iq   = [c for c in t_iq_hit.columns   if c.startswith('t_hit_')]

    summary = (
        events_with_tno
        .join(t_open_hit[cols_open], how='left')
        .join(t_iq_hit[cols_iq],     how='left'))

    #Guarantee expected columns exist.
    for col in ['t_hit_s_open_min','t_hit_s_iq_min']:
        if col not in summary.columns:
            summary[col] = np.nan

    print("\nMedian time-to-improvement (0.1\") by scenario:")
    med = summary.groupby('scenario')[['t_hit_s_open_min','t_hit_s_iq_min']].median()
    print(med)

    #Per-night summaries (informative) — pass events_with_tno (not summary) to avoid re-joining overlaps
    night_summary = summarize_by_night(events_with_tno, t_open_hit, t_iq_hit)
    if not night_summary.empty:
        print("\nPer-night summary:")
        print(night_summary)

    #Distribution of IQ at acquisition (s_iq_min=0) per scenario (optional quick view)
    acq = traj[np.isclose(traj['s_iq_min'], 0.0, atol=0.5)].copy()
    if not acq.empty:
        print("\nIQ at first acquisition (arcsec) by scenario (median [p16, p84]):")
        for scen, df_s in acq.groupby('scenario', observed=True):
            vals = df_s['IQ'].dropna().values
            if vals.size:
                print(f"  {scen}: {np.nanmedian(vals):.3f} "
                      f"[{np.nanpercentile(vals,16):.3f}, {np.nanpercentile(vals,84):.3f}]")
    else:
        print("\nNo IQ samples exactly at first acquisition; skip quick summary.")

    return events_with_tno, traj, summary, night_summary
```

```python

'''




#=========================
#tau_pre_min parameter help (histogram + CDF) plus a single-value suggestion.
#Δstart = t_iq_start - t_open (minutes) for events whose t_night_min_at_open lies in [night_window[0], night_window[1]) minutes since nautical night start.
#=========================
def plot_tau_pre_min_hist_cdf(events, night_window=(0, 120), tau_overlay=None, binsize_min=1.0):
    ev = events.copy()
    if 'delta_min_start' not in ev.columns:
        if 't_iq_start' not in ev.columns:
            raise KeyError("events must have delta_min_start or t_iq_start")
        ev['delta_min_start'] = (ev['t_iq_start'] - ev.index).total_seconds()/60.0

    if 't_night_min_at_open' not in ev.columns:
        raise KeyError("events must have 't_night_min_at_open' — call attach_time_of_night_to_events first.")

    mask = (ev['t_night_min_at_open'] >= night_window[0]) & (ev['t_night_min_at_open'] < night_window[1])
    sub = ev.loc[mask, 'delta_min_start'].dropna().astype(float)
    n = len(sub)

    print(f"Selected {n} events in night window [{night_window[0]}, {night_window[1]}) min since nautical night start.")
    if n == 0:
        print("No events in this window.")
        return

    vmax = max(5.0, float(np.nanpercentile(sub, 99)))
    edges = np.arange(0.0, vmax + binsize_min, binsize_min)

    fig, axs = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)

    axs[0].hist(sub, bins=edges, color='tab:blue', alpha=0.75, edgecolor='white')
    axs[0].set_xlabel('Δstart = t_iq_start − t_open [min]')
    axs[0].set_ylabel('Count')
    axs[0].set_title('Histogram of Δstart')

    s = np.sort(sub.values)
    cdf = np.arange(1, n+1)/n
    axs[1].plot(s, cdf, color='tab:purple', lw=2)
    axs[1].set_xlabel('Δstart [min]')
    axs[1].set_ylabel('CDF')
    axs[1].set_ylim(0, 1.01)
    axs[1].grid(True, alpha=0.3)
    axs[1].set_title('Empirical CDF of Δstart')

    q_levels = [0.50, 0.80, 0.90, 0.95]
    q_vals = {int(q*100): float(np.nanpercentile(sub, q*100)) for q in q_levels}
    print("Suggested tau_pre_min by 50/80/90/95% quantile (in minutes):")
    for q, v in q_vals.items():
        print(f"  {q}% quantile: {v:.2f} min")

    for v in q_vals.values():
        axs[0].axvline(v, color='gray', ls='--', lw=0.8, alpha=0.8)
        axs[1].axvline(v, color='gray', ls='--', lw=0.8, alpha=0.8)
    if tau_overlay is not None:
        axs[0].axvline(tau_overlay, color='red', ls='-', lw=1.2, label=f'tau_pre_min={tau_overlay}')
        axs[1].axvline(tau_overlay, color='red', ls='-', lw=1.2)
        axs[0].legend(loc='best')

    plt.show()


def suggest_tau_pre_min(events, night_window=(0, 120), quantile=0.80):
    if 't_night_min_at_open' not in events.columns:
        raise KeyError("events must have 't_night_min_at_open' — call attach_time_of_night_to_events first.")
    sub = events.loc[
        (events['t_night_min_at_open'] >= night_window[0]) &
        (events['t_night_min_at_open'] < night_window[1]),
        'delta_min_start'
    ].dropna()
    if sub.empty:
        return np.nan
    return float(np.nanpercentile(sub, quantile*100.0))


# =========================
# Per-night summaries
# =========================
def summarize_by_night(events, t_open_hit, t_iq_hit):
    """
    Build a per-night (night_id) summary:
      - counts, scenario mix
      - median delta_min_start
      - median t_hit_s_open_min, t_hit_s_iq_min
      - medians of initial conditions (dh0, dT0, M0, wind0)
    Requires 'night_id' in events and t_open_hit / t_iq_hit having standardized columns.
    """
    ev = events.copy()

    # Normalize accidental names if needed
    t_open_hit = _normalize_t_hit_columns(t_open_hit)
    t_iq_hit   = _normalize_t_hit_columns(t_iq_hit)

    cols_open = [c for c in t_open_hit.columns if c.startswith('t_hit_')]
    cols_iq   = [c for c in t_iq_hit.columns   if c.startswith('t_hit_')]

    # Join only missing t_hit columns to avoid overlap errors
    miss_open = [c for c in cols_open if c not in ev.columns]
    if miss_open:
        ev = ev.join(t_open_hit[miss_open], how='left')
    miss_iq = [c for c in cols_iq if c not in ev.columns]
    if miss_iq:
        ev = ev.join(t_iq_hit[miss_iq], how='left')

    # Guarantee expected columns exist
    for col in ['t_hit_s_open_min', 't_hit_s_iq_min']:
        if col not in ev.columns:
            ev[col] = np.nan

    if 'night_id' not in ev.columns:
        print("summarize_by_night: 'night_id' not found in events. Call attach_time_of_night_to_events first.")
        return pd.DataFrame()

    def scen_prop(g):
        vc = g['scenario'].value_counts(normalize=True)
        return pd.Series({
            'n_events': len(g),
            'p_pre_venting': float(vc.get('Pre-venting', 0.0)),
            'p_on_sky_venting': float(vc.get('On-sky + venting', 0.0)),
            'delta_min_start_med': float(np.nanmedian(g['delta_min_start'])),
            't_hit_s_open_min_med': float(np.nanmedian(g['t_hit_s_open_min'])),
            't_hit_s_iq_min_med': float(np.nanmedian(g['t_hit_s_iq_min'])),
            'dh0_med': float(np.nanmedian(g['dh0_kJkg_moist'])),
            'dT0_med': float(np.nanmedian(g['dT0_C'])),
            'M0_med': float(np.nanmedian(g['M0_C'])),
            'wind0_med': float(np.nanmedian(g['wind0_mps'])),
        })

    night_summary = ev.groupby('night_id', dropna=False).apply(scen_prop).reset_index()
    return night_summary


# =========================
# Execution
# =========================
def run_venting(feat, iq_1m, tau_pre_min=20, window_pre_min=60, window_post_min=120,
                dew_thr_C=5.0, iq_metric_name='aos_fwhm'):
    # Checking common UTC
    feat = ensure_utc_index(feat, 'feat')
    iq_1m = ensure_utc_index(iq_1m, 'iq_1m')
    print("Features index tz:", feat.index.tz, " | iq_1m index tz:", iq_1m.index.tz)

    # Optional: add time since night start to feat (exploration)
    try:
        feat['t_night_min'], _ = time_since_minus_twelve_start(feat, 'SunAltitude')
    except KeyError:
        feat['t_night_min'] = np.nan

    # Build events and trajectories
    events, traj = build_vent_events(
        feat, iq_1m,
        tau_pre_min=tau_pre_min,
        window_pre_min=window_pre_min,
        window_post_min=window_post_min,
        dew_thr_C=dew_thr_C,
        use_dew_gate_if_available=True
    )

    # Checking empty data (closed nights).
    if events.empty:
        print("No dome shutter open events found under dew-safe conditions.")
        return events, traj, pd.DataFrame(), pd.DataFrame()

    # Attach time-of-night and night_id for per-night summaries
    events_with_tno = attach_time_of_night_to_events(events, feat, max_tolerance='5min')

    print("\nEvent counts by scenario:")
    print(events_with_tno['scenario'].value_counts(dropna=False))

    print("\nEvent initial conditions (head):")
    print(events_with_tno[['scenario','delta_min_start','dh0_kJkg_moist','dT0_C','M0_C','wind0_mps','sun0_deg']].head())

    # Plot trajectories vs time since vent open and since first IQ (with labels)
    plot_iq_trajectories(traj, by='s_open_min',
                         window_pre_min=window_pre_min, window_post_min=window_post_min,
                         iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min)


    # Time-to-improvement (e.g., 0.1")
    t_open_hit = _normalize_t_hit_columns(time_to_improvement(traj, by='s_open_min', improve_arcsec=0.1))
    t_iq_hit   = _normalize_t_hit_columns(time_to_improvement(traj, by='s_iq_min',   improve_arcsec=0.1))

    # Keep only the numeric columns you need
    cols_open = [c for c in t_open_hit.columns if c.startswith('t_hit_')]
    cols_iq   = [c for c in t_iq_hit.columns   if c.startswith('t_hit_')]

    summary = (
        events_with_tno
        .join(t_open_hit[cols_open], how='left')
        .join(t_iq_hit[cols_iq],     how='left')
    )

    # Guarantee expected columns exist so downstream selection never errors
    for col in ['t_hit_s_open_min','t_hit_s_iq_min']:
        if col not in summary.columns:
            summary[col] = np.nan

    print("Summary columns:", summary.columns.tolist())
    print(summary.head())

    print("\nMedian time-to-improvement (0.1\") by scenario:")
    med = summary.groupby('scenario')[['t_hit_s_open_min','t_hit_s_iq_min']].median()
    print(med)

    # Per-night summaries (informative) — pass events_with_tno (not summary) to avoid re-joining overlaps
    night_summary = summarize_by_night(events_with_tno, t_open_hit, t_iq_hit)
    if not night_summary.empty:
        print("\nPer-night summary:")
        print(night_summary)

    # Distribution of IQ at acquisition (s_iq_min=0) per scenario (optional quick view)
    acq = traj[np.isclose(traj['s_iq_min'], 0.0, atol=0.5)].copy()
    if not acq.empty:
        print("\nIQ at first acquisition (arcsec) by scenario (median [p16, p84]):")
        for scen, df_s in acq.groupby('scenario', observed=True):
            vals = df_s['IQ'].dropna().values
            if vals.size:
                print(f"  {scen}: {np.nanmedian(vals):.3f} "
                      f"[{np.nanpercentile(vals,16):.3f}, {np.nanpercentile(vals,84):.3f}]")
    else:
        print("\nNo IQ samples exactly at first acquisition; skip quick summary.")

    return events_with_tno, traj, summary, night_summary

'''
'''
###Helpers
# ---------- Signed time-of-night (optional) ----------
def attach_time_of_night_to_events(events, feat, max_tolerance='5min',
                                   threshold_deg=-12.0, signed=False):
    """
    Adds minutes since nautical night start (SunAltitude <= threshold_deg) at open time,
    and a night_id to the events table.

    - If signed=False (default): t_night_min_at_open is NaN before night starts; >=0 during night.
    - If signed=True: also compute t_night_min_signed_at_open:
        * >=0 minutes after the most recent crossing (night started)
        * negative minutes before the next crossing (night not started yet)
      Useful for windows like (-60, 60) around the nautical start.

    Returns events with:
      - t_night_min_at_open (non-negative or NaN)
      - night_id (float, NaN outside night)
      - t_night_min_signed_at_open (only when signed=True)
    """
    feat = ensure_utc_index(feat, 'feat')

    # Base “minutes since night start” and night_id using your helper (threshold is fixed to -12 in your function)
    t_night_min, night_id = time_since_minus_twelve_start(feat, 'SunAltitude')

    out = events.copy()
    # Reindex minutes since night start and night_id at each t_open (nearest)
    out['t_night_min_at_open'] = t_night_min.reindex(events.index, method='nearest',
                                                     tolerance=pd.Timedelta(max_tolerance))
    out['night_id'] = night_id.reindex(events.index, method='nearest',
                                       tolerance=pd.Timedelta(max_tolerance))

    if not signed:
        return out

    # Build a signed version:
    # 1) Identify exact crossing times into night (SunAltitude falling to <= threshold)
    sun = feat['SunAltitude']
    mask_night = (sun <= threshold_deg)
    # Start-of-night indices (day->night)
    starts = feat.index[(mask_night.astype(int).diff() == 1).fillna(False)]
    if len(starts) == 0:
        # No crossings; signed == nonsigned behavior
        out['t_night_min_signed_at_open'] = out['t_night_min_at_open']
        return out

    # 2) Build a per-minute Series of the most recent start time (backfill past crossing)
    last_start = pd.Series(index=feat.index, dtype='datetime64[ns, UTC]')
    last_start.loc[starts] = starts
    last_start = last_start.ffill()  # last start at or before any time (NaT before first start)

    # 3) Build Series of the next start time (for times before first start or between nights)
    next_start = pd.Series(index=feat.index, dtype='datetime64[ns, UTC]')
    next_start.loc[starts] = starts
    next_start = next_start[::-1].ffill()[::-1]  # next start at or after any time (NaT after last start)

    # 4) Compute signed minutes: +minutes since last start if in night; else −minutes to next start
    # Align both start series to the event times by nearest
    ls = last_start.reindex(events.index, method='nearest',
                            tolerance=pd.Timedelta(max_tolerance))
    ns = next_start.reindex(events.index, method='nearest',
                            tolerance=pd.Timedelta(max_tolerance))

    signed_minutes = []
    for t, tnmin in zip(events.index, out['t_night_min_at_open']):
        if pd.notna(tnmin):  # already in night
            t0 = ls.loc[t]
            val = (t - t0).total_seconds() / 60.0 if pd.notna(t0) else np.nan
            signed_minutes.append(val)
        else:
            t1 = ns.loc[t]
            val = -(t1 - t).total_seconds() / 60.0 if pd.notna(t1) else np.nan
            signed_minutes.append(val)

    out['t_night_min_signed_at_open'] = signed_minutes
    return out

'''
'''
# ---------- tau_pre_min picker (hist + CDF) with signed option ----------
def plot_tau_pre_min_hist_cdf(events, night_window=(0, 120), tau_overlay=None, binsize_min=1.0,
                              use_signed=False):
    """
    Plot histogram and CDF of Δstart = t_iq_start - t_open (minutes) for events whose
    time-of-night lies in [night_window[0], night_window[1]) minutes.

    - If use_signed=False: window uses t_night_min_at_open (>=0; NaN before night)
    - If use_signed=True:  window uses t_night_min_signed_at_open (can be negative before night)
      This solves the “Selected 0 events in [0, 60)” when opens occur before nautical night.

    Prints tau_pre_min suggestions at 50/80/90/95% quantiles.
    """
    ev = events.copy()
    # Δstart
    if 'delta_min_start' not in ev.columns:
        if 't_iq_start' not in ev.columns:
            raise KeyError("events must have delta_min_start or t_iq_start")
        ev['delta_min_start'] = (ev['t_iq_start'] - ev.index).total_seconds()/60.0

    col = 't_night_min_signed_at_open' if use_signed else 't_night_min_at_open'
    if col not in ev.columns:
        raise KeyError(f"events must have '{col}'. Call attach_time_of_night_to_events(..., signed={use_signed}) first.")

    mask = (ev[col] >= night_window[0]) & (ev[col] < night_window[1])
    sub = ev.loc[mask, 'delta_min_start'].dropna().astype(float)
    n = len(sub)

    print(f"Selected {n} events in time-of-night window [{night_window[0]}, {night_window[1]}) min "
          f"(col={col}).")
    if n == 0:
        print("No events in this window.")
        return

    vmax = max(5.0, float(np.nanpercentile(sub, 99)))
    edges = np.arange(0.0, vmax + binsize_min, binsize_min)

    fig, axs = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)
    axs[0].hist(sub, bins=edges, color='tab:blue', alpha=0.75, edgecolor='white')
    axs[0].set_xlabel('Δstart = t_iq_start − t_open (min)')
    axs[0].set_ylabel('Count')
    axs[0].set_title('Histogram of Δstart')

    s = np.sort(sub.values)
    cdf = np.arange(1, n+1)/n
    axs[1].plot(s, cdf, color='tab:purple', lw=2)
    axs[1].set_xlabel('Δstart (min)')
    axs[1].set_ylabel('CDF')
    axs[1].set_ylim(0, 1.01)
    axs[1].grid(True, alpha=0.3)
    axs[1].set_title('Empirical CDF of Δstart')

    q_levels = [0.50, 0.80, 0.90, 0.95]
    q_vals = {int(q*100): float(np.nanpercentile(sub, q*100)) for q in q_levels}
    print("Suggested tau_pre_min by quantile (minutes):")
    for q, v in q_vals.items():
        print(f"  {q}% quantile: {v:.2f} min")

    for v in q_vals.values():
        axs[0].axvline(v, color='gray', ls='--', lw=0.8, alpha=0.8)
        axs[1].axvline(v, color='gray', ls='--', lw=0.8, alpha=0.8)
    if tau_overlay is not None:
        axs[0].axvline(tau_overlay, color='red', ls='-', lw=1.2, label=f'tau_pre_min={tau_overlay}')
        axs[1].axvline(tau_overlay, color='red', ls='-', lw=1.2)
        axs[0].legend(loc='best')

    plt.show()


def suggest_tau_pre_min(events, night_window=(0, 120), quantile=0.80, use_signed=False):
    """
    Suggest a single tau_pre_min for the given time-of-night window as the specified quantile of Δstart.
    """
    col = 't_night_min_signed_at_open' if use_signed else 't_night_min_at_open'
    if col not in events.columns:
        raise KeyError(f"events must have '{col}'. Call attach_time_of_night_to_events(..., signed={use_signed}) first.")

    sub = events.loc[
        (events[col] >= night_window[0]) & (events[col] < night_window[1]),
        'delta_min_start'
    ].dropna()
    if sub.empty:
        return np.nan
    return float(np.nanpercentile(sub, quantile*100.0))
'''
'''
# ---------- Join-only-missing utility for safe merges ----------
def _join_only_missing(left_df, right_df, cols, how='left'):
    """
    Join only missing columns from right_df[cols] into left_df, avoiding column overlap errors.
    """
    miss = [c for c in cols if c not in left_df.columns]
    return left_df.join(right_df[miss], how=how) if miss else left_df


# =========================
# Compact per-night report
# =========================
def make_night_report(feat, iq_1m, events_with_tno, traj, summary, night_id,
                      tau_pre_min, iq_metric_name='aos_fwhm',
                      night_window_signed=(-60, 120),  # minutes around nautical start (signed)
                      out_dir=None):
    """
    Generate a compact report for a specific night_id:
      - Prints counts, scenario mix, medians
      - Plots trajectories (filtered to this night) with labels
      - Plots tau_pre_min histogram/CDF in a signed time-of-night window (e.g., (-60,120))
      - Optionally saves PNGs and CSV summary to out_dir
    """
    # Filter to the requested night
    ev_n = events_with_tno.loc[events_with_tno['night_id'] == night_id]
    if ev_n.empty:
        print(f"\nNight {night_id}: no events.")
        return

    # Subset trajectories belonging to these events
    opens = ev_n.index
    traj_n = traj[traj['event_open'].isin(opens)].copy()

    # Basic stats
    print(f"\n=== Night report: night_id={night_id} ===")
    print(f"Events: {len(ev_n)}")
    print("Scenario counts:")
    print(ev_n['scenario'].value_counts(dropna=False))
    # Initial conditions medians
    med_ic = ev_n[['delta_min_start','dh0_kJkg_moist','dT0_C','M0_C','wind0_mps']].median(numeric_only=True)
    print("Initial medians:")
    print(med_ic)

    # Trajectory plots (per-night)
    plt.close('all')
    plot_iq_trajectories(traj_n, by='s_open_min',
                         window_pre_min=None, window_post_min=None,
                         iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min)
    if out_dir:
        os.makedirs(out_dir, exist_ok=True)
        plt.gcf().savefig(os.path.join(out_dir, f"night_{int(night_id)}_traj_open.png"), dpi=130)

    plot_iq_trajectories(traj_n, by='s_iq_min',
                         window_pre_min=None, window_post_min=None,
                         iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min)
    if out_dir:
        plt.gcf().savefig(os.path.join(out_dir, f"night_{int(night_id)}_traj_iq.png"), dpi=130)

    # Tau helper (signed time-of-night)
    # Make sure we have signed time-of-night in events
    if 't_night_min_signed_at_open' not in events_with_tno.columns:
        # Generate signed column quickly (reuse feat)
        events_with_tno_signed = attach_time_of_night_to_events(events_with_tno, feat, signed=True)
    else:
        events_with_tno_signed = events_with_tno

    # Merge delta_min_start into events if needed
    if 'delta_min_start' not in events_with_tno_signed.columns and 't_iq_start' in events_with_tno_signed.columns:
        tmp = events_with_tno_signed.copy()
        tmp['delta_min_start'] = (tmp['t_iq_start'] - tmp.index).total_seconds()/60.0
        events_with_tno_signed = tmp

    ev_n_signed = events_with_tno_signed.loc[opens]
    plot_tau_pre_min_hist_cdf(ev_n_signed,
                              night_window=night_window_signed,
                              tau_overlay=tau_pre_min,
                              binsize_min=1.0,
                              use_signed=True)
    if out_dir:
        plt.gcf().savefig(os.path.join(out_dir, f"night_{int(night_id)}_tau_hist.png"), dpi=130)

    # Save CSV summary for the night
    if out_dir:
        ev_n.to_csv(os.path.join(out_dir, f"night_{int(night_id)}_events.csv"))
        traj_n.to_csv(os.path.join(out_dir, f"night_{int(night_id)}_traj.csv"))
        print(f"Saved report artifacts to {out_dir}")


def make_all_night_reports(feat,
                           iq_1m,
                           events_with_tno,
                           traj,
                           summary,
                           tau_pre_min,
                           iq_metric_name='aos_fwhm',
                           night_window_signed=(-60, 120),
                           out_root='./night_reports',
                           summarize=True,
                           t_open_hit=None,
                           t_iq_hit=None):
    """
    Generate a report bundle for each night_id:
      - trajectory plots (s_open_min and s_iq_min) with labels
      - tau_pre_min histogram/CDF in a signed time-of-night window (e.g., [-60, +120] min)
      - events/traj CSVs per night
      - optional per-night summary CSV across all nights

    Parameters
    ----------
    feat, iq_1m : DataFrames
        Original feature table and 1-min IQ series (UTC).
    events_with_tno : DataFrame
        Events with 'night_id' and (optionally) 't_night_min_signed_at_open'.
    traj : DataFrame
        Long-format trajectories from build_vent_events.
    summary : DataFrame
        Joined event table with any t_hit_* columns you already computed.
    tau_pre_min : float
        Current tau used for scenario split; displayed on plots.
    iq_metric_name : str
        Label for the IQ metric (e.g., 'aos_fwhm').
    night_window_signed : tuple(min, max)
        Time-of-night window in minutes relative to nautical start (signed),
        used in tau histogram (e.g., (-60, 120)).
    out_root : str
        Root output directory (one subdir per night).
    summarize : bool
        If True and both t_open_hit and t_iq_hit are provided, write a consolidated
        per-night summary CSV (one row per night_id).
    t_open_hit, t_iq_hit : DataFrames or None
        Outputs from time_to_improvement, used only if summarize=True.

    Returns
    -------
    list of str : list of per-night output directories written
    """
    # Ensure signed time-of-night is available
    if 't_night_min_signed_at_open' not in events_with_tno.columns:
        events_with_tno = attach_time_of_night_to_events(events_with_tno, feat, signed=True)

    # Make root
    os.makedirs(out_root, exist_ok=True)

    # Unique numeric night_ids (skip NaN which indicates pre-nautical opens)
    night_ids = events_with_tno['night_id'].dropna().unique()
    # Sort for reproducibility
    try:
        night_ids = sorted(int(n) for n in night_ids)
    except Exception:
        night_ids = sorted(night_ids)

    out_dirs = []
    for nid in night_ids:
        out_dir = os.path.join(out_root, f"night_{int(nid)}")
        os.makedirs(out_dir, exist_ok=True)
        make_night_report(feat, iq_1m, events_with_tno, traj, summary,
                          night_id=nid,
                          tau_pre_min=tau_pre_min,
                          iq_metric_name=iq_metric_name,
                          night_window_signed=night_window_signed,
                          out_dir=out_dir)
        out_dirs.append(out_dir)

    # Optional consolidated per-night summary
    if summarize and (t_open_hit is not None) and (t_iq_hit is not None):
        # Normalize names (safety) and compute per-night table
        t_open_hit = _normalize_t_hit_columns(t_open_hit)
        t_iq_hit   = _normalize_t_hit_columns(t_iq_hit)
        night_summary = summarize_by_night(events_with_tno, t_open_hit, t_iq_hit)
        if not night_summary.empty:
            night_summary_path = os.path.join(out_root, "all_nights_summary.csv")
            night_summary.to_csv(night_summary_path, index=False)
            print(f"Saved consolidated per-night summary to {night_summary_path}")

    print(f"\nWrote {len(out_dirs)} night bundles under {out_root}")
    return out_dirs

# ========== Update: allow saving figures from the trajectory plot ==========
  



# ========== Ensure events carry signed time-of-night (− before nautical, + after) ==========
def ensure_signed_time_of_night(events_with_tno, feat, max_tolerance='5min'):
    """
    Guarantee 't_night_min_signed_at_open' exists in events_with_tno by attaching signed time-of-night.
    """
    if 't_night_min_signed_at_open' in events_with_tno.columns:
        return events_with_tno
    return attach_time_of_night_to_events(events_with_tno, feat, signed=True, max_tolerance=max_tolerance)


# ========== Internal: write per-group (night/date) bundle ==========
def make_group_report(feat, iq_1m, events_with_tno, traj, summary,
                      open_index, group_label, tau_pre_min, iq_metric_name,
                      night_window_signed=(-60, 120), out_dir=None):
    """
    Shared writer for a group of events identified by open_index (Index of t_open).
    Writes trajectory PNGs, tau PNG, events/traj CSVs under out_dir/group_label_*.
    """
    if out_dir:
        os.makedirs(out_dir, exist_ok=True)

    # Slice events/traj to this group
    ev_g = events_with_tno.loc[open_index]
    if ev_g.empty:
        print(f"[{group_label}] no events.")
        return

    traj_g = traj[traj['event_open'].isin(open_index)].copy()

    # Traj plots: s_open_min
    fig = plot_iq_trajectories(traj_g, by='s_open_min',
                               iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min,
                               show=False, return_fig=True)
    if out_dir and fig:
        fig.savefig(os.path.join(out_dir, f"{group_label}_traj_open.png"), dpi=130)
        plt.close(fig)

    # Traj plots: s_iq_min
    fig = plot_iq_trajectories(traj_g, by='s_iq_min',
                               iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min,
                               show=False, return_fig=True)
    if out_dir and fig:
        fig.savefig(os.path.join(out_dir, f"{group_label}_traj_iq.png"), dpi=130)
        plt.close(fig)

    # Tau helper (signed time-of-night)
    ev_signed = ensure_signed_time_of_night(ev_g, feat)
    # Ensure delta_min_start exists
    if 'delta_min_start' not in ev_signed.columns and 't_iq_start' in ev_signed.columns:
        ev_signed = ev_signed.copy()
        ev_signed['delta_min_start'] = (ev_signed['t_iq_start'] - ev_signed.index).total_seconds()/60.0

    # Tau plots (signed window)
    plot_tau_pre_min_hist_cdf(ev_signed,
                              night_window=night_window_signed,
                              tau_overlay=tau_pre_min,
                              binsize_min=1.0,
                              use_signed=True)
    if out_dir:
        plt.gcf().savefig(os.path.join(out_dir, f"{group_label}_tau_hist.png"), dpi=130)
        plt.close()

    # CSVs
    if out_dir:
        ev_g.to_csv(os.path.join(out_dir, f"{group_label}_events.csv"))
        traj_g.to_csv(os.path.join(out_dir, f"{group_label}_traj.csv"))
        print(f"[{group_label}] wrote plots and CSVs to {out_dir}")


# ========== All-night runner with fallback to date grouping for NaN night_id ==========
def make_all_night_reports(feat,
                           iq_1m,
                           events_with_tno,
                           traj,
                           summary,
                           tau_pre_min,
                           iq_metric_name='aos_fwhm',
                           night_window_signed=(-60, 120),
                           out_root='./night_reports',
                           summarize=True,
                           t_open_hit=None,
                           t_iq_hit=None,
                           fallback_by_date=True):
    """
    For each night_id, write a bundle:
      - night_{id}_traj_open.png, night_{id}_traj_iq.png
      - night_{id}_tau_hist.png (signed window)
      - night_{id}_events.csv, night_{id}_traj.csv
    If fallback_by_date=True, also write bundles for opens with night_id NaN, grouped by UTC date.

    Also writes all_nights_summary.csv if summarize=True and t_open_hit/t_iq_hit provided.
    """
    os.makedirs(out_root, exist_ok=True)

    # Ensure signed time-of-night available for tau windows and date fallback
    events_with_tno = ensure_signed_time_of_night(events_with_tno, feat)

    # 1) Night-based bundles (numerical night_id)
    night_ids = events_with_tno['night_id'].dropna().unique()
    # Sort stable
    try:
        night_ids = sorted(int(n) for n in night_ids)
    except Exception:
        night_ids = sorted(night_ids)

    for nid in night_ids:
        opens = events_with_tno.index[events_with_tno['night_id'] == nid]
        out_dir = os.path.join(out_root, f"night_{int(nid)}")
        make_group_report(feat, iq_1m, events_with_tno, traj, summary,
                          open_index=opens,
                          group_label=f"night_{int(nid)}",
                          tau_pre_min=tau_pre_min,
                          iq_metric_name=iq_metric_name,
                          night_window_signed=night_window_signed,
                          out_dir=out_dir)

    # 2) Fallback: date-based bundles for events without night_id (pre-nautical)
    if fallback_by_date:
        mask_pre = events_with_tno['night_id'].isna()
        if mask_pre.any():
            pre_df = events_with_tno[mask_pre].copy()
            # Group by UTC date (of t_open index)
            for day, idxs in pre_df.groupby(pre_df.index.date).groups.items():
                opens = pre_df.loc[idxs].index
                out_dir = os.path.join(out_root, f"date_{pd.Timestamp(day).strftime('%Y%m%d')}_pre")
                make_group_report(feat, iq_1m, events_with_tno, traj, summary,
                                  open_index=opens,
                                  group_label=f"date_{pd.Timestamp(day).strftime('%Y%m%d')}_pre",
                                  tau_pre_min=tau_pre_min,
                                  iq_metric_name=iq_metric_name,
                                  night_window_signed=night_window_signed,
                                  out_dir=out_dir)

    # 3) Consolidated per-night summary (if requested)
    if summarize and (t_open_hit is not None) and (t_iq_hit is not None):
        t_open_hit = _normalize_t_hit_columns(t_open_hit)
        t_iq_hit   = _normalize_t_hit_columns(t_iq_hit)
        night_summary = summarize_by_night(events_with_tno, t_open_hit, t_iq_hit)
        if not night_summary.empty:
            out_csv = os.path.join(out_root, "all_nights_summary.csv")
            night_summary.to_csv(out_csv, index=False)
            print(f"Saved consolidated per-night summary to {out_csv}")
        else:
            print("No per-night summary rows to write.")



# ========== Generic bands helper for any column ==========
def _bands_safe_col(df_grp, colname):
    v = df_grp[colname].to_numpy(dtype=float)
    v = v[~np.isnan(v)]
    if v.size:
        med = float(np.median(v)); lo = float(np.percentile(v, 16)); hi = float(np.percentile(v, 84))
    else:
        med = lo = hi = np.nan
    return pd.Series({'med': med, 'lo': lo, 'hi': hi})


# ========== DeltaH / DeltaT per-night summary plot + CSV ==========
def make_deltaH_deltaT_bundle(traj_group, group_label, out_dir=None,
                              time_axis='s_open_min', time_window_post=(0, 60)):
    """
    Produce a compact summary of ΔT and Δh for a group of events (per night/date):
      - Panel A: ΔT bands vs minutes since open per scenario
      - Panel B: Δh bands vs minutes since open per scenario
      - Panel C: Δh vs ΔT scatter (all samples in group)
      - Panel D: Histograms of ΔT and Δh, with medians annotated
      - CSV with key stats in time_window_post (e.g., [0,60] minutes since open):
          * n_points
          * ΔT med/p16/p84
          * Δh med/p16/p84
          * frac(Δh>0) in window
          * frac(ΔT>0) in window
    Requirements:
      - traj_group contains 'dTe_C_sm' (or 'dTe_C') and 'dhe_kJkg_moist_sm' (or 'dhe_kJkg')
      - 'scenario' and 's_open_min' columns
    """
    if traj_group.empty:
        print(f"[{group_label}] No trajectory rows to summarize ΔT/Δh.")
        return

    df = traj_group.copy()

    # Choose smoothed columns if available
    col_dT = 'dTe_C_sm' if 'dTe_C_sm' in df.columns else ('dTe_C' if 'dTe_C' in df.columns else None)
    col_dh = 'dhe_kJkg_moist_sm' if 'dhe_kJkg_moist_sm' in df.columns else ('dhe_kJkg' if 'dhe_kJkg' in df.columns else None)
    if col_dT is None or col_dh is None:
        print(f"[{group_label}] Missing ΔT or Δh columns; found dT={col_dT}, dH={col_dh}. Skipping ΔH/ΔT bundle.")
        return

    # Round the time axis for banding
    if time_axis not in df.columns:
        print(f"[{group_label}] time axis '{time_axis}' not found in trajectories.")
        return
    df = df.dropna(subset=[time_axis]).copy()
    df[time_axis] = df[time_axis].round(0)

    # Aggregate bands per scenario and minute
    agg_T = df.groupby(['scenario', time_axis], observed=True).apply(_bands_safe_col, colname=col_dT).reset_index()
    agg_H = df.groupby(['scenario', time_axis], observed=True).apply(_bands_safe_col, colname=col_dh).reset_index()

    # Figure layout
    fig, axs = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)
    # Panel A: ΔT bands
    for scen, g in agg_T.groupby('scenario', observed=True):
        axs[0,0].plot(g[time_axis], g['med'], label=f'{scen}')
        axs[0,0].fill_between(g[time_axis], g['lo'], g['hi'], alpha=0.15)
    axs[0,0].axhline(0, color='k', lw=0.8)
    axs[0,0].set_title('ΔT vs minutes since open (median ± p16/p84)')
    axs[0,0].set_xlabel('Minutes since open')
    axs[0,0].set_ylabel('ΔT = T_in − T_out [°C]')
    axs[0,0].grid(True, alpha=0.3)
    axs[0,0].legend(loc='best')

    # Panel B: Δh bands
    for scen, g in agg_H.groupby('scenario', observed=True):
        axs[0,1].plot(g[time_axis], g['med'], label=f'{scen}')
        axs[0,1].fill_between(g[time_axis], g['lo'], g['hi'], alpha=0.15)
    axs[0,1].axhline(0, color='k', lw=0.8)
    axs[0,1].set_title('Δh vs minutes since open (median ± p16/p84)')
    axs[0,1].set_xlabel('Minutes since open')
    axs[0,1].set_ylabel('Δh [kJ/kg moist]')
    axs[0,1].grid(True, alpha=0.3)
    axs[0,1].legend(loc='best')

    # Panel C: scatter Δh vs ΔT
    axs[1,0].scatter(df[col_dT], df[col_dh], s=8, alpha=0.5, c='tab:blue')
    axs[1,0].axhline(0, color='k', lw=0.8)
    axs[1,0].axvline(0, color='k', lw=0.8)
    axs[1,0].set_title('Δh vs ΔT (all samples in group)')
    axs[1,0].set_xlabel('ΔT [°C]')
    axs[1,0].set_ylabel('Δh [kJ/kg moist]')
    axs[1,0].grid(True, alpha=0.3)

    # Panel D: histograms
    vals_T = df[col_dT].to_numpy(dtype=float)
    vals_H = df[col_dh].to_numpy(dtype=float)
    axs[1,1].hist(vals_T[~np.isnan(vals_T)], bins=40, alpha=0.6, color='tab:orange', label='ΔT', edgecolor='white')
    axs[1,1].hist(vals_H[~np.isnan(vals_H)], bins=40, alpha=0.6, color='tab:green', label='Δh', edgecolor='white')
    axs[1,1].axvline(np.nanmedian(vals_T), color='tab:orange', lw=1.2, ls='--')
    axs[1,1].axvline(np.nanmedian(vals_H), color='tab:green', lw=1.2, ls='--')
    axs[1,1].set_title('Histograms: ΔT and Δh')
    axs[1,1].set_xlabel('Value')
    axs[1,1].set_ylabel('Count')
    axs[1,1].grid(True, alpha=0.3)
    axs[1,1].legend(loc='best')

    fig.suptitle(f"{group_label} — ΔT / Δh summary", y=0.98, fontsize=11)
    if out_dir:
        fig.savefig(os.path.join(out_dir, f"{group_label}_dh_dt_summary.png"), dpi=130)
    plt.close(fig)

    # CSV stats in time_window_post (e.g., 0..60 minutes since open)
    t0, t1 = time_window_post
    win = df[(df[time_axis] >= t0) & (df[time_axis] <= t1)]
    def pct16(x): return float(np.nanpercentile(x, 16)) if np.isfinite(x).any() else np.nan
    def pct84(x): return float(np.nanpercentile(x, 84)) if np.isfinite(x).any() else np.nan

    out_stats = {
        'group_label': group_label,
        'n_points_window': int(win.shape[0]),
        'dT_med': float(np.nanmedian(win[col_dT])) if not win.empty else np.nan,
        'dT_p16': pct16(win[col_dT].to_numpy(dtype=float)) if not win.empty else np.nan,
        'dT_p84': pct84(win[col_dT].to_numpy(dtype=float)) if not win.empty else np.nan,
        'dH_med': float(np.nanmedian(win[col_dh])) if not win.empty else np.nan,
        'dH_p16': pct16(win[col_dh].to_numpy(dtype=float)) if not win.empty else np.nan,
        'dH_p84': pct84(win[col_dh].to_numpy(dtype=float)) if not win.empty else np.nan,
        'frac_dH_pos_window': float(np.nanmean(win[col_dh] > 0.0)) if not win.empty else np.nan,
        'frac_dT_pos_window': float(np.nanmean(win[col_dT] > 0.0)) if not win.empty else np.nan,
    }
    if out_dir:
        pd.DataFrame([out_stats]).to_csv(os.path.join(out_dir, f"{group_label}_dh_dt_summary.csv"), index=False)
    return out_stats


# ========== Integrate ΔH/ΔT bundle into group report ==========
def make_group_report(feat, iq_1m, events_with_tno, traj, summary,
                      open_index, group_label, tau_pre_min, iq_metric_name,
                      night_window_signed=(-60, 120), out_dir=None):
    """
    Shared writer for a group of events identified by open_index (Index of t_open).
    Writes trajectory PNGs, tau PNG, events/traj CSVs, and ΔH/ΔT bundle under out_dir/group_label_*.
    """
    if out_dir:
        os.makedirs(out_dir, exist_ok=True)

    # Slice events/traj to this group
    ev_g = events_with_tno.loc[open_index]
    if ev_g.empty:
        print(f"[{group_label}] no events.")
        return

    traj_g = traj[traj['event_open'].isin(open_index)].copy()

    # Trajectory plots (s_open_min and s_iq_min)
    fig = plot_iq_trajectories(traj_g, by='s_open_min',
                               iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min,
                               show=False, return_fig=True)
    if out_dir and fig:
        fig.savefig(os.path.join(out_dir, f"{group_label}_traj_open.png"), dpi=130)
        plt.close(fig)

    fig = plot_iq_trajectories(traj_g, by='s_iq_min',
                               iq_metric_name=iq_metric_name, tau_pre_min=tau_pre_min,
                               show=False, return_fig=True)
    if out_dir and fig:
        fig.savefig(os.path.join(out_dir, f"{group_label}_traj_iq.png"), dpi=130)
        plt.close(fig)

    # Tau helper (signed time-of-night)
    ev_signed = ensure_signed_time_of_night(ev_g, feat)
    # Ensure delta_min_start exists
    if 'delta_min_start' not in ev_signed.columns and 't_iq_start' in ev_signed.columns:
        ev_signed = ev_signed.copy()
        ev_signed['delta_min_start'] = (ev_signed['t_iq_start'] - ev_signed.index).total_seconds()/60.0

    # Tau plots (signed window)
    plot_tau_pre_min_hist_cdf(ev_signed,
                              night_window=night_window_signed,
                              tau_overlay=tau_pre_min,
                              binsize_min=1.0,
                              use_signed=True)
    if out_dir:
        plt.gcf().savefig(os.path.join(out_dir, f"{group_label}_tau_hist.png"), dpi=130)
        plt.close()

    # ΔH/ΔT bundle (plots + CSV)
    make_deltaH_deltaT_bundle(traj_g, group_label, out_dir=out_dir, time_axis='s_open_min', time_window_post=(0, 60))

     # Correlation summary (IQ vs Δh/ΔT/wind/sun) with open@0 and first-IQ median line
    make_night_correlation_plot(traj_g, ev_g, group_label, out_dir=out_dir,
                                s_axis='s_open_min')
    
    # CSVs for events and traj
    if out_dir:
        ev_g.to_csv(os.path.join(out_dir, f"{group_label}_events.csv"))
        traj_g.to_csv(os.path.join(out_dir, f"{group_label}_traj.csv"))
        print(f"[{group_label}] wrote plots and CSVs to {out_dir}")


def _to_float_series(s):
    # Coerce to numeric float; invalid parses become NaN
    return pd.to_numeric(s, errors='coerce')

def _pearsonr_safe(x, y):
    # Safe Pearson on numeric arrays only
    x = _to_float_series(x).to_numpy(float)
    y = _to_float_series(y).to_numpy(float)
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() < 2:
        return np.nan, int(m.sum())
    r = np.corrcoef(x[m], y[m])[0, 1]
    return float(r), int(m.sum())

def make_night_correlation_plot(traj_group, ev_group, group_label, out_dir=None,
                                s_axis='s_open_min',
                                col_dT_candidates=('dTe_C_sm','dTe_C'),
                                col_dh_candidates=('dhe_kJkg_moist_sm','dhe_kJkg'),
                                wind_col='WindSpeed_301',
                                sun_col='SunAltitude',
                                iq_col='IQ'):
    """
    One summary plot per night/date showing correlations between Δh, ΔT, wind, Sun altitude and IQ trajectories.
    Highlights: open at t=0 (s_open_min) and the median Δstart (t_iq_start − t_open) across events.
    Writes <group_label>_corr_summary.png in out_dir (if provided).
    """
    if traj_group.empty:
        print(f"[{group_label}] make_night_correlation_plot: empty trajectories.")
        return

    df = traj_group.copy()

    # Pick ΔT / Δh columns
    col_dT = next((c for c in col_dT_candidates if c in df.columns), None)
    col_dh = next((c for c in col_dh_candidates if c in df.columns), None)
    if col_dT is None or col_dh is None:
        print(f"[{group_label}] ΔT/Δh columns missing (ΔT={col_dT}, Δh={col_dh}); skipping correlation plot.")
        return

    # Ensure time axis exists
    if s_axis not in df.columns:
        print(f"[{group_label}] time axis '{s_axis}' missing; skipping correlation plot.")
        return

    # Coerce all used columns to float
    for c in [s_axis, iq_col, col_dT, col_dh, wind_col, sun_col]:
        if c in df.columns:
            df[c] = _to_float_series(df[c])

    # Drop NaN time rows and round minutes for banding
    df = df.dropna(subset=[s_axis]).copy()
    df[s_axis] = df[s_axis].round(0)

    # Bands for IQ
    def _p16(series):
        a = _to_float_series(series).to_numpy(float)
        return np.nanpercentile(a, 16) if np.isfinite(a).any() else np.nan
    def _p84(series):
        a = _to_float_series(series).to_numpy(float)
        return np.nanpercentile(a, 84) if np.isfinite(a).any() else np.nan

    bands = df.groupby(s_axis, observed=True)[iq_col].agg(IQ_med='median', IQ_lo=_p16, IQ_hi=_p84).reset_index()

    # Per-minute medians for Δh / ΔT
    agg_dh = df.groupby(s_axis, observed=True)[col_dh].median().reset_index().rename(columns={col_dh: 'dH_med'})
    agg_dT = df.groupby(s_axis, observed=True)[col_dT].median().reset_index().rename(columns={col_dT: 'dT_med'})

    # Median Δstart across events (for first-IQ line)
    if 'delta_min_start' not in ev_group.columns and 't_iq_start' in ev_group.columns:
        ev_group = ev_group.copy()
        ev_group['delta_min_start'] = (ev_group['t_iq_start'] - ev_group.index).total_seconds()/60.0
    med_delta = float(np.nanmedian(_to_float_series(ev_group['delta_min_start']))) if 'delta_min_start' in ev_group.columns else np.nan

    # Build figure
    fig, axs = plt.subplots(2, 3, figsize=(15, 8), constrained_layout=True)

    # A: IQ bands vs minutes since open; Δh/ΔT medians on twin y-axis
    axL = axs[0, 0]
    axL.plot(bands[s_axis], bands['IQ_med'], color='tab:blue', lw=2, label='IQ median')
    axL.fill_between(bands[s_axis], bands['IQ_lo'], bands['IQ_hi'], color='tab:blue', alpha=0.15, label='IQ p16–p84')
    axL.axvline(0.0, color='k', lw=1.2, ls='-', label='open (t=0)')
    if np.isfinite(med_delta):
        axL.axvline(med_delta, color='tab:red', lw=1.2, ls='--', label=f'first IQ (median Δstart ≈ {med_delta:.1f} m)')
    axL.set_xlabel('Minutes since open'); axL.set_ylabel('IQ [arcsec]'); axL.grid(True, alpha=0.3)

    axR = axL.twinx()
    axR.plot(agg_dh[s_axis], agg_dh['dH_med'], color='tab:green', lw=1.4, label='Δh median')
    axR.plot(agg_dT[s_axis], agg_dT['dT_med'], color='tab:orange', lw=1.4, label='ΔT median')
    axR.set_ylabel('Δh [kJ/kg moist], ΔT [°C]')

    h1, l1 = axL.get_legend_handles_labels()
    h2, l2 = axR.get_legend_handles_labels()
    axL.legend(h1+h2, l1+l2, loc='best')

    # Helper for scatter with Pearson r
    def _scatter(ax, x, y, xlabel, ylabel):
        r, n = _pearsonr_safe(x, y)
        ax.scatter(_to_float_series(x), _to_float_series(y), s=8, alpha=0.5)
        ax.grid(True, alpha=0.3)
        ax.set_xlabel(xlabel); ax.set_ylabel(ylabel)
        ax.set_title(f"Pearson r = {r:.2f} (n={n})")

    # B: IQ vs Δh
    _scatter(axs[0, 1], df[col_dh], df[iq_col], xlabel='Δh [kJ/kg moist]', ylabel='IQ [arcsec]')

    # C: IQ vs ΔT
    _scatter(axs[0, 2], df[col_dT], df[iq_col], xlabel='ΔT [°C]', ylabel='IQ [arcsec]')

    # D: IQ vs Wind (if available)
    if wind_col in df.columns:
        _scatter(axs[1, 0], df[wind_col], df[iq_col], xlabel='Wind [m/s]', ylabel='IQ [arcsec]')
    else:
        axs[1, 0].set_visible(False)

    # E: IQ vs Sun altitude (if available)
    if sun_col in df.columns:
        _scatter(axs[1, 1], df[sun_col], df[iq_col], xlabel='Sun altitude [deg]', ylabel='IQ [arcsec]')
    else:
        axs[1, 1].set_visible(False)

    # F: Δh vs ΔT colored by IQ
    x = _to_float_series(df[col_dT]).to_numpy(float)
    h = _to_float_series(df[col_dh]).to_numpy(float)
    iq = _to_float_series(df[iq_col]).to_numpy(float)
    m = np.isfinite(x) & np.isfinite(h) & np.isfinite(iq)
    sc = axs[1, 2].scatter(x[m], h[m], c=iq[m], s=10, alpha=0.6, cmap='viridis')
    axs[1, 2].axhline(0, color='k', lw=0.8); axs[1, 2].axvline(0, color='k', lw=0.8)
    axs[1, 2].grid(True, alpha=0.3)
    axs[1, 2].set_xlabel('ΔT [°C]'); axs[1, 2].set_ylabel('Δh [kJ/kg moist]')
    cb = fig.colorbar(sc, ax=axs[1, 2]); cb.set_label('IQ [arcsec]')

    fig.suptitle(f"{group_label} — IQ & environment correlations (open@0, first-IQ median@{med_delta:.1f}m)",
                 fontsize=11, y=0.98)

    if out_dir:
        os.makedirs(out_dir, exist_ok=True)
        fig.savefig(os.path.join(out_dir, f"{group_label}_corr_summary.png"), dpi=130)
    plt.close(fig)



def quick_diagnostics(events_with_tno, traj, binsize_min=1.0, improve_arcsec=0.1, show=True, return_fig=False):
    """
    Quick diagnostics:
      A) Δstart = t_iq_start − t_open histogram and CDF
      B) Per-minute coverage (number of IQ samples) for:
         - minutes since open (s_open_min)
         - minutes since first IQ (s_iq_min)

    Requirements:
      - events_with_tno contains t_iq_start (or delta_min_start)
      - traj contains 's_open_min', 's_iq_min', 'IQ'

    Returns:
      fig if return_fig=True, else None
    """
    # Δstart
    ev = events_with_tno.copy()
    if 'delta_min_start' not in ev.columns:
        if 't_iq_start' not in ev.columns:
            raise KeyError("events_with_tno must have delta_min_start or t_iq_start")
        ev['delta_min_start'] = (ev['t_iq_start'] - ev.index).total_seconds()/60.0

    dstart = _to_float_series(ev['delta_min_start']).dropna()
    n = len(dstart)

    # Coverage per-minute
    tr = traj[['s_open_min','s_iq_min','IQ']].copy()
    tr['s_open_min'] = _to_float_series(tr['s_open_min']).round(0)
    tr['s_iq_min']   = _to_float_series(tr['s_iq_min']).round(0)
    cov_open = tr.dropna(subset=['s_open_min'])[['s_open_min','IQ']].groupby('s_open_min', as_index=False)['IQ'].count().rename(columns={'IQ':'count_open'})
    cov_iq   = tr.dropna(subset=['s_iq_min'])  [['s_iq_min','IQ']].groupby('s_iq_min',   as_index=False)['IQ'].count().rename(columns={'IQ':'count_iq'})

    # Plot
    fig, axs = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)

    # A) Δstart hist + CDF
    if n > 0:
        vmax = max(5.0, float(np.nanpercentile(dstart, 99)))
        edges = np.arange(0.0, vmax + binsize_min, binsize_min)
        axs[0].hist(dstart, bins=edges, color='tab:blue', alpha=0.75, edgecolor='white')
        axs[0].set_xlabel('Δstart = t_iq_start − t_open (min)')
        axs[0].set_ylabel('Count')
        axs[0].set_title(f'Δstart (n={n})')

        s = np.sort(dstart.values); cdf = np.arange(1, n+1)/n
        ax_cdf = axs[0].twinx()
        ax_cdf.plot(s, cdf, color='tab:purple', lw=2)
        ax_cdf.set_ylabel('CDF')
        ax_cdf.set_ylim(0, 1.01)

        # Print quantiles for quick selection of tau_pre_min
        for q in [50, 80, 90, 95]:
            qv = float(np.nanpercentile(dstart, q))
            axs[0].axvline(qv, color='gray', ls='--', lw=0.8)
            axs[0].text(qv, axs[0].get_ylim()[1]*0.8, f'{q}%\n{qv:.1f}m', rotation=90, va='center', ha='right', fontsize=8)
    else:
        axs[0].text(0.5, 0.5, 'No Δstart samples', ha='center', va='center', transform=axs[0].transAxes)
        axs[0].axis('off')

    # B) Coverage lines
    axs[1].plot(cov_open['s_open_min'], cov_open['count_open'], label='coverage: s_open_min', color='tab:green')
    axs[1].plot(cov_iq['s_iq_min'],     cov_iq['count_iq'],   label='coverage: s_iq_min',   color='tab:orange')
    axs[1].set_xlabel('Minutes')
    axs[1].set_ylabel('Count of IQ samples')
    axs[1].set_title('Per-minute coverage')
    axs[1].grid(True, alpha=0.3)
    axs[1].legend(loc='best')

    if show:
        plt.show()
    else:
        plt.tight_layout()
    return fig if return_fig else None


def plot_iq_dh_dt_vs_time_of_night(traj, events_with_tno, signed=True,
                                   ton_window=(-120, 360),  # minutes
                                   iq_col='IQ',
                                   dT_candidates=('dTe_C_sm','dTe_C'),
                                   dH_candidates=('dhe_kJkg_moist_sm','dhe_kJkg'),
                                   binsize_min=1.0,
                                   show=True, return_fig=False):
    """
    IQ / Δh / ΔT vs time-of-night (minutes relative to nautical start).
    - signed=True: negative before nautical start; positive after
    - ton_window: x-axis window to display, in minutes
    - Computes time-of-night for each trajectory row as:
        t_night_min_signed_at_open + s_open_min  (if signed)
        t_night_min_at_open + s_open_min         (if unsigned, rows before night remain NaN)

    Panels:
      1) IQ bands (p16/p50/p84) vs time-of-night, with sample counts (secondary axis)
      2) Δh bands vs time-of-night
      3) ΔT bands vs time-of-night

    Returns:
      fig if return_fig=True else None
    """
    if traj.empty or events_with_tno.empty:
        print("plot_iq_dh_dt_vs_time_of_night: empty inputs."); 
        return None if return_fig else None

    # Choose ΔT/Δh columns
    col_dT = next((c for c in dT_candidates if c in traj.columns), None)
    col_dH = next((c for c in dH_candidates if c in traj.columns), None)
    if col_dT is None or col_dH is None:
        print(f"Missing ΔT/Δh columns (ΔT={col_dT}, Δh={col_dH}).")
        return None if return_fig else None

    # Build per-row time-of-night from event anchor
    ton_col_ev = 't_night_min_signed_at_open' if signed else 't_night_min_at_open'
    if ton_col_ev not in events_with_tno.columns:
        raise KeyError(f"events_with_tno must have '{ton_col_ev}'. Use attach_time_of_night_to_events(..., signed={signed}) first.")

    # Map event-level time-of-night at open to trajectory rows
    ev_map = events_with_tno[ton_col_ev].to_dict()  # {t_open_ts: minutes}
    df = traj.copy()
    df['t_open_ton'] = df['event_open'].map(ev_map)
    df['s_open_min'] = _to_float_series(df['s_open_min'])
    if signed:
        df['ton'] = _to_float_series(df['t_open_ton']) + df['s_open_min']
    else:
        # unsigned: only valid after night start (t_open_ton not NaN)
        df['ton'] = np.where(pd.notna(df['t_open_ton']), _to_float_series(df['t_open_ton']) + df['s_open_min'], np.nan)

    # Select window and coerce columns
    df = df[(df['ton'] >= ton_window[0]) & (df['ton'] <= ton_window[1])]
    df['ton']   = _to_float_series(df['ton']).round(0)
    df[iq_col]  = _to_float_series(df[iq_col])
    df[col_dT]  = _to_float_series(df[col_dT])
    df[col_dH]  = _to_float_series(df[col_dH])

    # Bands per-minute for each variable
    def bands_col(df_, col):
        a = df_[col].to_numpy(float)
        a = a[np.isfinite(a)]
        if a.size:
            med = float(np.median(a)); lo = float(np.percentile(a, 16)); hi = float(np.percentile(a, 84))
        else:
            med = lo = hi = np.nan
        return pd.Series({'med': med, 'lo': lo, 'hi': hi})

    agg_IQ = df.groupby('ton', observed=True).apply(bands_col, col=iq_col).reset_index()
    agg_H  = df.groupby('ton', observed=True).apply(bands_col, col=col_dH).reset_index()
    agg_T  = df.groupby('ton', observed=True).apply(bands_col, col=col_dT).reset_index()

    # Coverage (how many IQ samples per minute)
    cov = df.groupby('ton', observed=True)[iq_col].count().reset_index(name='n_IQ')

    # Plot
    fig, axs = plt.subplots(3, 1, figsize=(12, 10), sharex=True, constrained_layout=True)

    # 1) IQ bands + counts
    axs[0].plot(agg_IQ['ton'], agg_IQ['med'], color='tab:blue', lw=2, label='IQ median')
    axs[0].fill_between(agg_IQ['ton'], agg_IQ['lo'], agg_IQ['hi'], color='tab:blue', alpha=0.15, label='IQ p16–p84')
    axs[0].axvline(0, color='k', lw=1.2, ls='-', label='nautical start (0 min)')
    axs[0].set_ylabel('IQ [arcsec]')
    axs[0].grid(True, alpha=0.3)
    ax0b = axs[0].twinx()
    ax0b.plot(cov['ton'], cov['n_IQ'], color='tab:gray', lw=1.2, alpha=0.7, label='count (IQ)')
    ax0b.set_ylabel('count (IQ)')
    h1,l1 = axs[0].get_legend_handles_labels()
    h2,l2 = ax0b.get_legend_handles_labels()
    axs[0].legend(h1+h2, l1+l2, loc='best')

    # 2) Δh bands
    axs[1].plot(agg_H['ton'], agg_H['med'], color='tab:green', lw=1.8, label='Δh median')
    axs[1].fill_between(agg_H['ton'], agg_H['lo'], agg_H['hi'], color='tab:green', alpha=0.15)
    axs[1].axhline(0, color='k', lw=0.8); axs[1].axvline(0, color='k', lw=0.8)
    axs[1].set_ylabel('Δh [kJ/kg moist]'); axs[1].grid(True, alpha=0.3); axs[1].legend(loc='best')

    # 3) ΔT bands
    axs[2].plot(agg_T['ton'], agg_T['med'], color='tab:orange', lw=1.8, label='ΔT median')
    axs[2].fill_between(agg_T['ton'], agg_T['lo'], agg_T['hi'], color='tab:orange', alpha=0.15)
    axs[2].axhline(0, color='k', lw=0.8); axs[2].axvline(0, color='k', lw=0.8)
    axs[2].set_ylabel('ΔT [°C]'); axs[2].set_xlabel('Minutes relative to nautical start'); axs[2].grid(True, alpha=0.3); axs[2].legend(loc='best')

    fig.suptitle(f"Across nights — IQ / Δh / ΔT vs time-of-night ({'signed' if signed else 'unsigned'})", y=0.98, fontsize=11)

    if show:
        plt.show()
    else:
        plt.tight_layout()
    return fig if return_fig else None


def build_dayobs_list(start_ymd='20250415', end_ymd='20250920'):
    """Return [YYYYMMDD, ...] integers across [start, end], inclusive."""
    rng = pd.date_range(pd.to_datetime(start_ymd), pd.to_datetime(end_ymd), freq='D')
    return rng.strftime('%Y%m%d').astype(int).tolist()

def _to_dayobs_from_index(dt_index):
    """Map a DatetimeIndex (UTC) to day_obs (YYYYMMDD int) by UTC calendar date."""
    return (pd.to_datetime(dt_index)
              .tz_convert('UTC') if getattr(dt_index, 'tz', None) else pd.to_datetime(dt_index, utc=True)) \
             .strftime('%Y%m%d').astype(int)

def summarize_nights_iq_and_venting_robust(
    df_open,           # exposures table; must have ['type'] and ideally one of ['obs_start','exp_midpt','time','time_start']
    features,          # telemetry (DatetimeIndex UTC) with SunAltitude, optionally Dome_Shutter_Actual{0,1}, dew_margin_C_sm, vent_open_event
    events,            # events from build_vent_events (index = t_open)
    start_ymd='20250415',
    end_ymd='20250920',
    on_sky_types=('science','cwfs','focus','acq'),  # exclude 'flat' by default
    dew_thr_C=5.0,
    shutter_threshold=None,     # auto-detect 0.5 (0..1) or 50.0 (0..100) if None
    shutter_persist_min=2,      # require >= N consecutive minutes as “slit open”
    map_tolerance_open='2H',    # nearest tolerance when mapping events -> night_id
    promote_open_tol='12H',     # promotion window for pre-nautical events -> next night
    map_tolerance_exp='4H',     # nearest tolerance when mapping exposures -> night_id
    promote_exp_tol='12H'       # promotion window for pre-nautical exposures -> next night
):
    """
    Return:
      summary_counts: dict
      per_night: one row per astronomical night with IQ, including:
        - Times: t0_nautical_utc, t_open_first_any_utc, t_open_first_dewsafe_utc, t_iq_first_utc
        - Lists of open times (UTC) and minutes vs t0 (any & dew-safe)
        - Per-night 'axis' values:
            s_open_first_min_vs_t0, s_iq_first_min_vs_t0,
            s_open_min_at_first_iq, s_iq_min_at_first_open, iq_before_open_flag
        - Inclusion flags and reasons
        - Counts: n_exposures_night, n_opens_night
    """
    def _map_times_to_night_id_promoted(times, night_id_series, nearest_tol, promote_tol):
        times = pd.to_datetime(times, utc=True)
        if not isinstance(times, pd.DatetimeIndex):
            times = pd.DatetimeIndex(times, tz='UTC')
        times = times.sort_values().unique()
        nid_nearest = night_id_series.reindex(times, method='nearest', tolerance=pd.Timedelta(nearest_tol))
        nid_next = night_id_series.fillna(method='bfill')
        nid_bfill = nid_next.reindex(times, method='nearest', tolerance=pd.Timedelta(promote_tol))
        mapped = nid_nearest.copy()
        mask_nan = mapped.isna()
        mapped[mask_nan] = nid_bfill[mask_nan]
        return mapped

    def _first_time(mask, night):
        m = mask & feat['night_id'].eq(night)
        return feat.index[m].min() if m.any() else pd.NaT

    feat = ensure_utc_index(features.copy(), 'features')
    # Astronomical night bins + t0 (first index where night mask becomes true)
    mask_night = (feat['SunAltitude'] <= -12.0)
    _, night_id = time_since_minus_twelve_start(feat, 'SunAltitude')
    feat['night_id'] = night_id
    t0_map = (feat.loc[mask_night].reset_index()
                 .groupby('night_id')['index'].min()
                 .to_dict())

    # Slit/vent open detection
    slit_open_mask = pd.Series(False, index=feat.index)
    if {'Dome_Shutter_Actual0','Dome_Shutter_Actual1'}.issubset(feat.columns):
        s0 = pd.to_numeric(feat['Dome_Shutter_Actual0'], errors='coerce')
        s1 = pd.to_numeric(feat['Dome_Shutter_Actual1'], errors='coerce')
        if shutter_threshold is None:
            med = np.nanmedian(pd.concat([s0, s1])); thr = 0.5 if 0 <= med <= 1 else 50.0
        else:
            thr = float(shutter_threshold)
        raw_slit = (s0 >= thr) & (s1 >= thr)
        slit_open_mask = raw_slit.rolling(shutter_persist_min, min_periods=shutter_persist_min).sum().eq(shutter_persist_min).fillna(False)
    elif 'vent_open_event' in feat.columns:
        slit_open_mask = feat['vent_open_event'].fillna(False).astype(bool)

    # Dew-safe
    if 'dew_margin_C_sm' in feat.columns:
        dew_margin = pd.to_numeric(feat['dew_margin_C_sm'], errors='coerce')
        dew_safe_mask = dew_margin >= dew_thr_C
        dew_safe_slit_open = slit_open_mask & dew_safe_mask
    else:
        dew_safe_mask = pd.Series(False, index=feat.index)
        dew_safe_slit_open = slit_open_mask

    nights = feat['night_id'].dropna().unique().tolist()

    # First opens + lists of opens per night
    first_open_any, first_open_dew = {}, {}
    open_list_any, open_list_dew = {}, {}
    for nid in nights:
        m_any = slit_open_mask & feat['night_id'].eq(nid)
        m_dew = dew_safe_slit_open & feat['night_id'].eq(nid)
        opens_any = feat.index[m_any].sort_values()
        opens_dew = feat.index[m_dew].sort_values()
        first_open_any[nid] = (opens_any.min() if len(opens_any) else pd.NaT)
        first_open_dew[nid] = (opens_dew.min() if len(opens_dew) else pd.NaT)
        open_list_any[nid] = list(opens_any)
        open_list_dew[nid] = list(opens_dew)

    nights_with_open_any = {nid for nid,t in first_open_any.items() if pd.notna(t)}
    nights_with_open_dew = {nid for nid,t in first_open_dew.items() if pd.notna(t)}

    # Nights included via events
    if not events.empty:
        ev_idx = ensure_utc_index(events.copy(), 'events').index
        ev_nid = _map_times_to_night_id_promoted(ev_idx, night_id, map_tolerance_open, promote_open_tol)
        nights_included = set(pd.Series(ev_nid).dropna().unique().tolist())
    else:
        nights_included = set()

    # Nights with IQ (promoted)
    df = df_open.copy()
    if on_sky_types is not None:
        df = df[df['type'].isin(on_sky_types)]
    time_col = next((c for c in ['obs_start','exp_midpt','time','time_start','midpt'] if c in df.columns), None)

    nights_with_iq, first_iq, n_exp_night = set(), {}, {}
    if time_col is not None:
        exp_times = pd.to_datetime(df[time_col], utc=True).dropna()
        exp_nid = _map_times_to_night_id_promoted(exp_times, night_id, map_tolerance_exp, promote_exp_tol)
        df_map = pd.DataFrame({'t': pd.DatetimeIndex(exp_times), 'night_id': exp_nid.values})
        for nid, grp in df_map.dropna(subset=['night_id']).groupby('night_id'):
            nights_with_iq.add(nid)
            first_iq[nid] = grp['t'].min()
            n_exp_night[nid] = int(grp.shape[0])
    else:
        if 'day_obs' not in df.columns:
            raise KeyError("df_open must have an exposure timestamp (obs_start/exp_midpt/...) or a day_obs column.")
        df['day_obs'] = pd.to_numeric(df['day_obs'], errors='coerce').astype('Int64')
        df = df.dropna(subset=['day_obs']).copy()
        date_to_nid = (feat.dropna(subset=['night_id'])
                          .groupby(feat.index.date)['night_id']
                          .agg(lambda s: s.mode().iloc[0] if not s.mode().empty else np.nan))
        for d, grp in df.groupby(pd.to_datetime(df['day_obs'].astype(str), format='%Y%m%d').dt.date):
            nid = date_to_nid.get(d, np.nan)
            if pd.notna(nid):
                nights_with_iq.add(nid)
                n_exp_night[nid] = n_exp_night.get(nid, 0) + int(grp.shape[0])

    # day_obs_label (prefer t0, else first open any, else first IQ)
    day_obs_label = {}
    consider = set(nights_with_iq) | set(nights_with_open_any) | set(nights_with_open_dew)
    for nid in consider:
        t0 = t0_map.get(nid, pd.NaT)
        t_any = first_open_any.get(nid, pd.NaT)
        t_iqf = first_iq.get(nid, pd.NaT)
        t_lbl = t0 if pd.notna(t0) else (t_any if pd.notna(t_any) else t_iqf)
        day_obs_label[nid] = (int(pd.Timestamp(t_lbl).strftime('%Y%m%d')) if pd.notna(t_lbl) else np.nan)

    # Build rows
    rows = []
    for nid in sorted(nights_with_iq):
        t0 = t0_map.get(nid, pd.NaT)
        t_any = first_open_any.get(nid, pd.NaT)
        t_dew = first_open_dew.get(nid, pd.NaT)
        t_iqf = first_iq.get(nid, pd.NaT)

        # “time since t0” positions (these are the per-night axes you asked for)
        s_open_first_min_vs_t0 = ((t_any - t0).total_seconds()/60.0) if (pd.notna(t_any) and pd.notna(t0)) else np.nan
        s_iq_first_min_vs_t0   = ((t_iqf - t0).total_seconds()/60.0) if (pd.notna(t_iqf) and pd.notna(t0)) else np.nan

        # “IQ vs open” positions
        s_open_min_at_first_iq = ((t_iqf - t_any).total_seconds()/60.0) if (pd.notna(t_iqf) and pd.notna(t_any)) else np.nan
        s_iq_min_at_first_open = (-s_open_min_at_first_iq) if pd.notna(s_open_min_at_first_iq) else np.nan
        iq_before_open_flag    = bool(pd.notna(s_open_min_at_first_iq) and (s_open_min_at_first_iq < 0))

        # Reasons/state
        status = 'included' if nid in nights_included else 'rejected'
        telem_present = feat['night_id'].eq(nid).any()
        if status == 'included':
            reason = None
        else:
            if not telem_present:
                reason = 'telemetry_gap'
            elif nid not in nights_with_open_any:
                reason = 'no_slit_open'
            elif (nid in nights_with_open_any) and (nid not in nights_with_open_dew):
                reason = 'dew_unsafe_at_open'
            else:
                reason = 'no_iq_after_open_or_alignment'

        # Lists of open times & their minutes vs t0
        opens_any = open_list_any.get(nid, [])
        opens_dew = open_list_dew.get(nid, [])
        opens_any_from_t0 = [ (ts - t0).total_seconds()/60.0 for ts in opens_any ] if pd.notna(t0) else []
        opens_dew_from_t0 = [ (ts - t0).total_seconds()/60.0 for ts in opens_dew ] if pd.notna(t0) else []

        rows.append({
            'night_id': nid,
            'day_obs_label': day_obs_label.get(nid, np.nan),
            'has_iq': True,
            'included': (status == 'included'),
            'reason': reason,
            't0_nautical_utc': t0,
            't_open_first_any_utc': t_any,
            't_open_first_dewsafe_utc': t_dew,
            't_iq_first_utc': t_iqf,
            's_open_first_min_vs_t0': s_open_first_min_vs_t0,
            's_iq_first_min_vs_t0': s_iq_first_min_vs_t0,
            's_open_min_at_first_iq': s_open_min_at_first_iq,
            's_iq_min_at_first_open': s_iq_min_at_first_open,
            'iq_before_open_flag': iq_before_open_flag,
            'open_times_utc': opens_any,
            'open_times_dewsafe_utc': opens_dew,
            'open_times_from_t0_min': opens_any_from_t0,
            'open_times_dewsafe_from_t0_min': opens_dew_from_t0,
            'n_exposures_night': n_exp_night.get(nid, 0),
            'n_opens_night': int(feat.loc[feat['night_id'].eq(nid) & slit_open_mask].shape[0]),
        })

    per_night = pd.DataFrame(rows).sort_values('night_id')

    summary_counts = {
        'nights_with_iq': int(len(set(nights_with_iq))),
        'nights_included': int(per_night['included'].sum()),
        'nights_rejected': int((~per_night['included']).sum()),
        'rejected_breakdown': per_night.loc[~per_night['included'], 'reason'].value_counts().to_dict()
    }

    rows.append({
    'night_id': nid,
    'day_obs_label': day_obs_label.get(nid, np.nan),

    # inclusion
    'has_iq': True,
    'included': (status == 'included'),
    'reason': reason,

    # absolute anchors
    't0_nautical_utc': t0,
    't_open_first_any_utc': t_any,
    't_open_first_dewsafe_utc': t_dew,
    't_iq_first_utc': t_iqf,

    # per-night axes (preferred names)
    's_open_first_min_vs_t0': s_open_first_min_vs_t0,     # (t_open_first_any - t0)/min
    's_iq_first_min_vs_t0':   s_iq_first_min_vs_t0,       # (t_iq_first - t0)/min
    's_open_min_at_first_iq': s_open_min_at_first_iq,     # (t_iq_first - t_open_first_any)/min
    's_iq_min_at_first_open': s_iq_min_at_first_open,     # = -s_open_min_at_first_iq
    'iq_before_open_flag':    iq_before_open_flag,

    # backward-compatible aliases (so your existing print/select works)
    'delta_iq_minus_open_any_min': s_open_min_at_first_iq,
    'delta_open_any_vs_t0_min':    s_open_first_min_vs_t0,
    'delta_iq_vs_t0_min':          s_iq_first_min_vs_t0,

    # lists of open times & their minutes vs t0
    'open_times_utc': open_list_any.get(nid, []),
    'open_times_dewsafe_utc': open_list_dew.get(nid, []),
    'open_times_from_t0_min': [
        (ts - t0).total_seconds()/60.0 for ts in open_list_any.get(nid, [])
    ] if pd.notna(t0) else [],
    'open_times_dewsafe_from_t0_min': [
        (ts - t0).total_seconds()/60.0 for ts in open_list_dew.get(nid, [])
    ] if pd.notna(t0) else [],

    # counts
    'n_exposures_night': n_exp_night.get(nid, 0),
    'n_opens_night': int(feat.loc[feat['night_id'].eq(nid) & slit_open_mask].shape[0]),
})
    
    return summary_counts, per_night
    

# =========================
# Execution remains your run_venting (unchanged except for calling summarize_by_night
# on events_with_tno and using standardized t_hit names).
# ========================
'''
```

```python
def per_event_iq_at_minutes(traj, minutes=(30, 60), by='s_open_min'):
    """
    Returns a per-event DataFrame (index=event_open) with IQ_{m}m columns
    computed as the median IQ in the 1-minute bin at each requested minute.
    """
    if traj.empty:
        return pd.DataFrame()
    t = traj.copy()
    keep = [by, 'event_open', 'IQ']
    t = t.dropna(subset=[by, 'event_open', 'IQ'])
    t[by] = pd.to_numeric(t[by], errors='coerce').round(0)
    g = t.groupby(['event_open', by], observed=True)['IQ'].median().reset_index()
    rows = []
    for ev, sub in g.groupby('event_open', observed=True):
        row = {'event_open': ev}
        for m in minutes:
            v = sub.loc[sub[by] == float(m), 'IQ']
            row[f'IQ_{int(m)}m'] = float(v.iloc[0]) if not v.empty else np.nan
        rows.append(row)
    return pd.DataFrame(rows).set_index('event_open').sort_index()


def outcome_curve_by_tau(events, traj, tau_grid, minute=60, by='s_open_min',
                         min_events=8, n_boot=2000, seed=0):
    """
    For each tau in tau_grid, compute median IQ at `minute` since open
    among events with delta_min_start >= tau. Returns a DataFrame with:
      tau, n, median, lo95, hi95
    """
    # Build per-event IQ@minute table once
    iq_ev = per_event_iq_at_minutes(traj, minutes=(minute,), by=by)
    ev = events[['delta_min_start']].join(iq_ev, how='left')
    ycol = f'IQ_{int(minute)}m'
    rng = np.random.default_rng(seed)

    out = []
    for tau in tau_grid:
        sub = ev.loc[pd.to_numeric(ev['delta_min_start'], errors='coerce') >= float(tau), ycol].dropna()
        n = int(sub.size)
        if n < min_events:
            out.append({'tau': float(tau), 'n': n, 'median': np.nan, 'lo95': np.nan, 'hi95': np.nan})
            continue
        med = float(np.nanmedian(sub))
        # Bootstrap 95% CI for the median
        if n_boot and n > 1:
            idx = rng.integers(0, n, size=(n_boot, n))
            boots = np.nanmedian(sub.to_numpy()[idx], axis=1)
            lo95, hi95 = float(np.nanpercentile(boots, 2.5)), float(np.nanpercentile(boots, 97.5))
        else:
            lo95 = hi95 = np.nan
        out.append({'tau': float(tau), 'n': n, 'median': med, 'lo95': lo95, 'hi95': hi95})
    return pd.DataFrame(out).sort_values('tau')


def plot_outcome_curve(df_curve, minute=60, ax=None, title=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(7,4), constrained_layout=True)
    else:
        fig = ax.figure
    ax.plot(df_curve['tau'], df_curve['median'], color='tab:green', marker='o') #, label=f'Median IQ@{minute}m')
    if {'lo95','hi95'}.issubset(df_curve.columns):
        ax.fill_between(df_curve['tau'], df_curve['lo95'], df_curve['hi95'], color='tab:green', alpha=0.15, linewidth=0)
    ax.set_xlabel('Ventilation time [min]')
    ax.set_ylabel(f'Median IQ at {minute} min [arcsec]')
    ax.grid(True, alpha=0.3)
    if title:
        ax.set_title(title)
    # Suggested choice: smallest tau within 1% of global min median
    y = df_curve['median'].to_numpy(float)
    x = df_curve['tau'].to_numpy(float)
    if np.isfinite(y).any():
        y_min = np.nanmin(y)
        within = np.where(y <= 1.01*y_min)[0]
        if within.size:
            tau_star = x[within[0]]
            ax.axvline(tau_star, color='tab:red', ls='--', lw=1.2, label=f'Minimun value at {int(tau_star)} min')
    ax.legend(loc='best')
    return fig, ax

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import ceil
from matplotlib import cm
from matplotlib.colors import Normalize

# Assumes your ensure_utc_index(...) and time_since_minus_twelve_start(...) exist.

def compute_night_windows_from_features(features: pd.DataFrame, sun_col='SunAltitude') -> pd.DataFrame:
    feat = ensure_utc_index(features, 'features')
    if sun_col not in feat.columns:
        raise KeyError(f"{sun_col} not found in features.")
    t_night_min, night_id_series = time_since_minus_twelve_start(feat, sun_col)
    rows = []
    ids = [int(x) for x in pd.Series(night_id_series).dropna().unique()]
    ids.sort()
    for i, nid in enumerate(ids):
        mask = (night_id_series == nid)
        if not mask.any():
            continue
        t0 = feat.index[mask][0]
        if i+1 < len(ids):
            t1 = feat.index[night_id_series == ids[i+1]][0]
        else:
            t1 = t0 + pd.Timedelta(hours=24)
        rows.append({'night_id': nid, 't0_m12': t0, 't1_m12': t1})
    return pd.DataFrame(rows)

def plot_dh_dT_vectors_by_scenario_v2(
    features: pd.DataFrame,
    events: pd.DataFrame,
    dh_col='dhe_kJkg_moist_sm',
    dT_col='dTe_C_sm',
    wind_col='WindSpeed_301',
    sun_col='SunAltitude',
    scenario_col='scenario',
    mode='event',                # 'event' -> one trajectory per event; 'night' -> one per night
    anchor='t_open',             # 't_open' or 't_m12' (m12 requires night mapping)
    horizon_min=240,             # window length after anchor
    sample_every_min=5,
    invert_x=True,
    invert_y=False,
    initial_marker='D',          # diamond for initial conditions
    initial_tol_min=10,          # tolerance to sample ΔT/Δh at t_open for the marker
    figsize=(12, 5),
    alpha_points=0.7,
    arrow_alpha=0.0,
    title=" "
):
    """
    Event-mode (default): titles show event totals per scenario and drawn events. Initial markers match drawn events.
    Night-mode: one path per night (like before), titles show per-night counts.
    Colors points by wind with a shared colorbar. Inverts axes if requested.
    Returns (fig, axes, counts_dict).
    """
    feat = ensure_utc_index(features, 'features')
    for col in [dh_col, dT_col]:
        if col not in feat.columns:
            raise KeyError(f"features missing '{col}'")

    have_wind = (wind_col in feat.columns)
    # Wind color mapping (shared)
    if have_wind:
        w_all = pd.to_numeric(feat[wind_col], errors='coerce').clip(lower=0)
        wmin = float(np.nanpercentile(w_all, 2)) if w_all.notna().any() else 0.0
        wmax = float(np.nanpercentile(w_all, 98)) if w_all.notna().any() else (wmin + 1.0)
        if not np.isfinite(wmax) or wmax <= wmin:
            wmax = wmin + 1.0
        norm = Normalize(vmin=wmin, vmax=wmax)
        cmap = cm.viridis
        sm = cm.ScalarMappable(norm=norm, cmap=cmap)
        sm.set_array([])
    else:
        norm = None; cmap = None; sm = None

    scen_list = sorted(events[scenario_col].dropna().unique().tolist()) if scenario_col in events.columns else ['__all__']
    if not scen_list:
        scen_list = ['__all__']

    # Build night windows if needed (for m12 anchoring or night-mode)
    nights = None
    if mode == 'night' or anchor == 't_m12':
        nights = compute_night_windows_from_features(feat, sun_col=sun_col)
        if 'night_id' not in events.columns:
            raise KeyError("events must have 'night_id' to use mode='night' or anchor='t_m12'.")

    # Panel layout
    n_scen = len(scen_list)
    ncols = min(2, n_scen)
    nrows = int(ceil(n_scen / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize, sharex=True, sharey=True)
    if nrows * ncols == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    counts = {}
    for i, scen in enumerate(scen_list):
        ax = axes[i]
        drawn = 0
        total = 0

        if mode == 'event':
            # One trajectory per event
            ev_scen = events[events[scenario_col] == scen].copy()
            ev_scen = ev_scen.sort_index()  # index is t_open
            total = len(ev_scen)
            for t_open, row in ev_scen.iterrows():
                # Anchor time
                if anchor == 't_open':
                    t_start = t_open
                else:
                    # use nautical start from nights for this event
                    nid = row.get('night_id', np.nan)
                    if nights is None or pd.isna(nid):
                        continue
                    nn = nights[nights['night_id'] == int(nid)]
                    if nn.empty: 
                        continue
                    t_start = nn.iloc[0]['t0_m12']

                t_end = t_start + pd.Timedelta(minutes=float(horizon_min))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                if have_wind:
                    W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                else:
                    W  = pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 2:
                    continue
                drawn += 1

                dT = dT[mask]; dH = dH[mask]; W = W[mask]
                if have_wind:
                    colors = cmap(norm(W.values))
                    ax.scatter(dT.values, dH.values, s=40, c=colors, edgecolor='k', linewidth=0.3, alpha=alpha_points)
                else:
                    ax.scatter(dT.values, dH.values, s=40, color='tab:blue', edgecolor='k', linewidth=0.3, alpha=alpha_points)

                x = dT.values; y = dH.values
                dx = x[1:] - x[:-1]; dy = y[1:] - y[:-1]
                ax.quiver(x[:-1], y[:-1], dx, dy, angles='xy', scale_units='xy', scale=1.0,
                          color='gray', alpha=arrow_alpha, width=0.002, headwidth=3, headlength=4)

                # Initial marker at t_open (sample ΔT/Δh at t_open with larger tolerance)
                t0 = pd.to_datetime(t_open, utc=True)
                t_window = t0 + pd.to_timedelta([0], unit='min')
                dT0 = feat[dT_col].reindex(t_window, method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                dH0 = feat[dh_col].reindex(t_window, method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                if dT0.notna().all() and dH0.notna().all():
                    ax.scatter([float(dT0.iloc[0])], [float(dH0.iloc[0])],
                               s=90, marker=initial_marker, color='red', edgecolor='k', linewidth=0.6, zorder=5, alpha=0.4)

        else:
            # Night-mode: one trajectory per night (earliest open shown with marker)
            if nights is None:
                nights = compute_night_windows_from_features(feat, sun_col=sun_col)
            # Map earliest open per night for marker and scenario filtering
            ev = events.reset_index().rename(columns={'index':'t_open'}).sort_values('t_open')
            first = ev.groupby('night_id', as_index=False).first()[['night_id','t_open',scenario_col]]
            # Filter to nights of this scenario
            nn = nights.merge(first, on='night_id', how='left')
            nn = nn[nn[scenario_col] == scen]
            total = nn['night_id'].nunique()

            for _, r in nn.iterrows():
                t0_m12, t1_m12 = r['t0_m12'], r['t1_m12']
                t_open_first = r['t_open']
                t_start = t0_m12 if anchor == 't_m12' else (t_open_first if pd.notna(t_open_first) else t0_m12)
                t_end = min(t1_m12, t_start + pd.Timedelta(minutes=float(horizon_min)))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce') if have_wind else pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 2:
                    continue
                drawn += 1

                dT = dT[mask]; dH = dH[mask]; W = W[mask]
                if have_wind:
                    colors = cmap(norm(W.values))
                    ax.scatter(dT.values, dH.values, s=40, c=colors, edgecolor='k', linewidth=0.3, alpha=alpha_points)
                else:
                    ax.scatter(dT.values, dH.values, s=40, color='tab:blue', edgecolor='k', linewidth=0.3, alpha=alpha_points)

                x = dT.values; y = dH.values
                dx = x[1:] - x[:-1]; dy = y[1:] - y[:-1]
                ax.quiver(x[:-1], y[:-1], dx, dy, angles='xy', scale_units='xy', scale=1.0,
                          color='gray', alpha=arrow_alpha, width=0.002, headwidth=3, headlength=4)

                # Initial marker at earliest open in night
                if pd.notna(t_open_first):
                    t0 = pd.to_datetime(t_open_first, utc=True)
                    dT0 = feat[dT_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    dH0 = feat[dh_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    if dT0.notna().all() and dH0.notna().all():
                        ax.scatter([float(dT0.iloc[0])], [float(dH0.iloc[0])],
                                   s=90, marker=initial_marker, color='red', edgecolor='k', linewidth=0.6, zorder=5, alpha=0.4)

        # Axes cosmetics and titles
        if invert_x: ax.invert_xaxis()
        if invert_y: ax.invert_yaxis()
        ax.axhline(0.0, color='k', lw=0.8, alpha=0.6)
        ax.axvline(0.0, color='k', lw=0.8, alpha=0.6)
        ax.grid(True, alpha=0.3)

        # Title
        # For event mode, show event totals; for night mode, show nights
        noun = 'events' if mode == 'event' else 'nights'
        ax.set_title(f"{scen}")
        counts[scen] = dict(total=int(total), drawn=int(drawn))

    # Axis labels and shared colorbar
    for ax in axes[:len(scen_list)]:
        ax.set_xlabel("ΔT [°C]")
        ax.set_ylabel("Δh [kJ/kg moist]")
    fig.suptitle(title, y=0.98)
    fig.tight_layout(rect=[0, 0, 1, 0.96])

    for j in range(len(scen_list), len(axes)):
        axes[j].set_visible(False)

    if have_wind and sm is not None:
        cbar = fig.colorbar(sm, ax=axes[:len(scen_list)], fraction=0.02, pad=0.02)
        cbar.set_label("Wind [m/s]")

    # Legend for initial marker
    axes[0].scatter([], [], s=80, marker=initial_marker, color='red', edgecolor='k', linewidth=0.6, label='Initial conditions')
    axes[0].legend(loc='best', frameon=True)
        # Legend for initial marker
    axes[1].scatter([], [], s=80, marker=initial_marker, color='red', edgecolor='k', linewidth=0.6, label='Initial conditions')
    axes[0].legend(loc='best', frameon=True)

    return fig, axes[:len(scen_list)], counts
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def report_open_timing_histograms(
    events: pd.DataFrame,
    feat: pd.DataFrame,
    # Columns
    sun_col: str = 'SunAltitude',
    wind_speed_col: str = 'WindSpeed_301',
    wind_dir_col: str = 'WindDirection_301',
    dome_az_col: str = 'Dome_Azimuth_Actual',
    delta_min_start_col: str = 'delta_min_start',
    # Binning configuration
    bin_size_min: float = 5.0,
    bin_size_deg: float = 1.0,
    bin_size_dT: float = 0.2,
    bin_size_dH: float = 1.0,
    bin_size_wind: float = 0.5,
    bin_size_dir_deg: float = 10.0,
    # Extra sun-altitude thresholds (deg) to compute t_open − crossing histograms for (besides 0° and −12°)
    altitudes_deg = (-6.0, -18.0),
    # Matching tolerances
    sun_match_tolerance: str = '10min',
    env_match_tolerance: str = '10min',
    wind_match_tolerance: str = '10min',
    dome_az_match_tolerance: str = '10min',
    # ΔT and Δh candidates
    dT_col_candidates = ('dTe_C_sm', 'dTe_C'),
    dH_col_candidates = ('dhe_kJkg_moist_sm', 'dhe_kJkg'),
    # Output controls
    show: bool = True,
    return_figs: bool = False,
):
    """
    Builds histograms and summaries for:
      Minutes:
        - delta_min_start = t_iq_start − t_open
        - t_open − sunset (0° descending)
        - t_open − (−12°) after that sunset
        - t_open − (extra altitude crossings you specify)
      Degrees:
        - Sun altitude at open
      Environment at open:
        - ΔT0 (°C)
        - Δh0 (kJ/kg moist)
      Wind:
        - Wind speed (m/s) at open
        - Wind speed (m/s) at IQ start
        - Wind direction (deg) at open (circular stats + histogram mode)
        - Wind direction (deg) at IQ start (circular stats + histogram mode)
      Dome orientation:
        - Dome azimuth at open (deg, circular stats + histogram mode)

    Inputs
    - events: DataFrame indexed by t_open (DatetimeIndex, UTC or naive). Must include either:
        delta_min_start (minutes) OR t_iq_start (Datetime-like).
    - feat: DataFrame with a DatetimeIndex containing:
        SunAltitude, WindSpeed_301, WindDirection_301, Dome_Azimuth_Actual,
        and optionally ΔT/Δh columns (see dT_col_candidates, dH_col_candidates).

    Returns
    - summary: dict with sub-dicts for each histogram containing:
        n, median/p16/p84/mode (linear) OR circ_mean_deg/circ_std_deg/mode_deg (circular),
        mode_bin, counts, bin_edges
      Keys include:
        'delta_min_start'
        'open_minus_sunset_min'
        'open_minus_minus12_min'
        'open_minus_altitude_min' (dict keyed by extra altitudes)
        'open_sun_altitude_deg'
        'dT0_C' (if available)
        'dH0_kJkg_moist' (if available)
        'wind_speed_open_mps'
        'wind_speed_iq_start_mps'
        'wind_dir_open_deg'
        'wind_dir_iq_start_deg'
        'dome_az_open_deg'
    - figs: list of matplotlib Figures if return_figs=True
    """
    # ---------- helpers ----------
    def _ensure_utc_index(df, name):
        if not isinstance(df.index, pd.DatetimeIndex):
            raise TypeError(f"{name} must have a DatetimeIndex.")
        return df.tz_localize('UTC') if df.index.tz is None else df.tz_convert('UTC')

    def _to_float_series(s):
        return pd.to_numeric(s, errors='coerce')

    def _crossings_down(s: pd.Series, thr: float) -> pd.DatetimeIndex:
        """
        Downward crossings of s across thr (prev > thr and curr <= thr), with linear interpolation.
        """
        s = _to_float_series(s)
        idx = s.index
        vals = s.to_numpy(dtype=float)
        out = []
        for i in range(1, len(s)):
            v0, v1 = vals[i-1], vals[i]
            if not (np.isfinite(v0) and np.isfinite(v1)):
                continue
            if v0 > thr and v1 <= thr:
                t0, t1 = idx[i-1], idx[i]
                alpha = 1.0 if v1 == v0 else (thr - v0) / (v1 - v0)
                alpha = float(np.clip(alpha, 0.0, 1.0))
                t_cross = t0 + (t1 - t0) * alpha
                out.append(pd.Timestamp(t_cross).tz_convert(idx.tz) if idx.tz is not None else pd.Timestamp(t_cross))
        return pd.DatetimeIndex(out, tz=idx.tz).sort_values()

    def _first_on_or_after(times: pd.DatetimeIndex, t: pd.Timestamp):
        if len(times) == 0:
            return pd.NaT
        pos = times.searchsorted(t, side='left')
        return times[pos] if pos < len(times) else pd.NaT

    def _last_on_or_before(times: pd.DatetimeIndex, t: pd.Timestamp):
        if len(times) == 0:
            return pd.NaT
        pos = times.searchsorted(t, side='right') - 1
        return times[pos] if pos >= 0 else pd.NaT

    def _make_edges(v: np.ndarray, binsize: float):
        v = np.asarray(v, dtype=float)
        v = v[np.isfinite(v)]
        if v.size == 0:
            return np.linspace(-binsize, binsize, 3)
        vmin, vmax = np.nanpercentile(v, [1, 99])
        if not np.isfinite(vmin) or not np.isfinite(vmax):
            vmin, vmax = np.nanmin(v), np.nanmax(v)
        if vmin == vmax:
            vmin -= binsize
            vmax += binsize
        left = np.floor(vmin / binsize) * binsize
        right = np.ceil(vmax / binsize) * binsize
        nb = max(5, int(np.ceil((right - left) / binsize)))
        edges = left + np.arange(nb + 1) * binsize
        return edges

    def _hist_summary(vals: pd.Series, binsize: float):
        v = _to_float_series(vals).to_numpy(dtype=float)
        v = v[np.isfinite(v)]
        out = {'n': int(v.size)}
        if v.size == 0:
            out.update({
                'median': np.nan, 'p16': np.nan, 'p84': np.nan,
                'mode': np.nan, 'mode_bin': (np.nan, np.nan),
                'counts': np.array([], dtype=int), 'bin_edges': np.array([], dtype=float),
            })
            return out
        edges = _make_edges(v, binsize)
        counts, bin_edges = np.histogram(v, bins=edges)
        mode_idx = int(np.argmax(counts)) if counts.size else 0
        left, right = bin_edges[mode_idx], bin_edges[mode_idx + 1]
        out.update({
            'median': float(np.nanmedian(v)),
            'p16': float(np.nanpercentile(v, 16)),
            'p84': float(np.nanpercentile(v, 84)),
            'mode': float(0.5 * (left + right)),
            'mode_bin': (float(left), float(right)),
            'counts': counts,
            'bin_edges': bin_edges
        })
        return out

    def _hist_summary_direction(vals_deg: pd.Series, bin_size_deg: float):
        """
        Direction histogram in degrees (0..360). Provides circular mean/std and a histogram mode.
        """
        v = _to_float_series(vals_deg).to_numpy(dtype=float)
        v = v[np.isfinite(v)]
        out = {'n': int(v.size)}
        if v.size == 0:
            out.update({
                'circ_mean_deg': np.nan, 'circ_std_deg': np.nan,
                'mode_deg': np.nan, 'mode_bin': (np.nan, np.nan),
                'counts': np.array([], dtype=int), 'bin_edges': np.array([], dtype=float),
            })
            return out
        v = np.mod(v, 360.0)
        # Histogram edges from 0 to 360
        edges = np.arange(0.0, 360.0 + bin_size_deg, bin_size_deg, dtype=float)
        if edges[-1] > 360.0 + 1e-9:
            edges[-1] = 360.0
        counts, bin_edges = np.histogram(v, bins=edges, range=(0.0, 360.0))
        mode_idx = int(np.argmax(counts)) if counts.size else 0
        left, right = bin_edges[mode_idx], bin_edges[mode_idx + 1]
        # Circular mean/std
        theta = np.deg2rad(v)
        C = np.nanmean(np.cos(theta))
        S = np.nanmean(np.sin(theta))
        mean_angle = np.arctan2(S, C)
        if mean_angle < 0:
            mean_angle += 2 * np.pi
        Rbar = float(np.hypot(C, S))
        Rbar = max(Rbar, 1e-12)  # guard log(0)
        circ_std_rad = np.sqrt(-2.0 * np.log(Rbar))
        out.update({
            'circ_mean_deg': float(np.rad2deg(mean_angle)),
            'circ_std_deg': float(np.rad2deg(circ_std_rad)),
            'mode_deg': float(0.5 * (left + right)),
            'mode_bin': (float(left), float(right)),
            'counts': counts,
            'bin_edges': bin_edges
        })
        return out

    def _rename_with_unit(summary, unit_suffix):
        # Convert generic keys to unit-specific (e.g., _min, _deg, _C, _kJkg, _mps)
        if summary['n'] == 0:
            return {**summary,
                    f'median_{unit_suffix}': np.nan,
                    f'p16_{unit_suffix}': np.nan,
                    f'p84_{unit_suffix}': np.nan,
                    f'mode_{unit_suffix}': np.nan}
        return {
            'n': summary['n'],
            f'median_{unit_suffix}': summary['median'],
            f'p16_{unit_suffix}': summary['p16'],
            f'p84_{unit_suffix}': summary['p84'],
            f'mode_{unit_suffix}': summary['mode'],
            'mode_bin': summary['mode_bin'],
            'counts': summary['counts'],
            'bin_edges': summary['bin_edges']
        }

    # ---------- prepare inputs ----------
    events = _ensure_utc_index(events.copy(), 'events')
    feat = _ensure_utc_index(feat.copy(), 'feat')

    for col, msg in [
        (sun_col, "Sun altitude"),
        (wind_speed_col, "wind speed"),
        (wind_dir_col, "wind direction"),
        (dome_az_col, "dome azimuth"),
    ]:
        if col not in feat.columns:
            raise KeyError(f"feat must contain column '{col}' ({msg}).")

    sun = feat[sun_col].sort_index()

    # delta_min_start if missing
    if delta_min_start_col not in events.columns:
        if 't_iq_start' not in events.columns:
            raise KeyError(f"events must have '{delta_min_start_col}' or 't_iq_start'.")
        events[delta_min_start_col] = (events['t_iq_start'] - events.index).total_seconds() / 60.0

    # Compute/normalize t_iq_start series (for wind at IQ start)
    if 't_iq_start' in events.columns:
        t_iq_start_series = pd.to_datetime(events['t_iq_start'])
        if t_iq_start_series.dt.tz is None:
            t_iq_start_series = t_iq_start_series.dt.tz_localize('UTC')
        else:
            t_iq_start_series = t_iq_start_series.dt.tz_convert('UTC')
    else:
        t_iq_start_series = events.index + pd.to_timedelta(pd.to_numeric(events[delta_min_start_col], errors='coerce'), unit='m')

    # ---------- anchors: sunset (0° down) and -12° after that sunset ----------
    sunset_times = _crossings_down(sun, 0.0)
    minus12_times = _crossings_down(sun, -12.0)

    t_open = events.index
    open_minus_sunset = []
    open_minus_minus12 = []

    for t in t_open:
        ss = _last_on_or_before(sunset_times, t)
        if pd.isna(ss):
            open_minus_sunset.append(np.nan)
            open_minus_minus12.append(np.nan)
            continue
        # sunset diff
        dt_ss_min = (t - ss).total_seconds() / 60.0
        open_minus_sunset.append(dt_ss_min)
        # -12° after that sunset
        m12_after_ss = _first_on_or_after(minus12_times, ss)
        if pd.isna(m12_after_ss):
            # fallback: nearest -12° to t
            i = minus12_times.searchsorted(t, side='left')
            cand = []
            if i > 0:
                cand.append(minus12_times[i-1])
            if i < len(minus12_times):
                cand.append(minus12_times[i])
            m12_after_ss = min(cand, key=lambda x: abs((t - x).total_seconds())) if cand else pd.NaT
        dt_m12_min = (t - m12_after_ss).total_seconds() / 60.0 if pd.notna(m12_after_ss) else np.nan
        open_minus_minus12.append(dt_m12_min)

    events = events.assign(
        open_minus_sunset_min=np.array(open_minus_sunset, dtype=float),
        open_minus_minus12_min=np.array(open_minus_minus12, dtype=float),
    )

    # ---------- extra altitude crossings (optional) ----------
    alt_extras = [float(a) for a in altitudes_deg if a not in (0.0, -12.0)]
    open_minus_alt = {}
    if len(alt_extras) > 0:
        crossings = {a: _crossings_down(sun, a) for a in alt_extras}
        for a in alt_extras:
            diffs = []
            times_a = crossings[a]
            for t in t_open:
                ss = _last_on_or_before(sunset_times, t)
                if pd.isna(ss) or len(times_a) == 0:
                    diffs.append(np.nan)
                    continue
                a_after_ss = _first_on_or_after(times_a, ss)
                if pd.isna(a_after_ss):
                    # fallback: nearest crossing to t
                    i = times_a.searchsorted(t, side='left')
                    cand = []
                    if i > 0:
                        cand.append(times_a[i-1])
                    if i < len(times_a):
                        cand.append(times_a[i])
                    a_after_ss = min(cand, key=lambda x: abs((t - x).total_seconds())) if cand else pd.NaT
                dt_a_min = (t - a_after_ss).total_seconds() / 60.0 if pd.notna(a_after_ss) else np.nan
                diffs.append(dt_a_min)
            open_minus_alt[a] = np.array(diffs, dtype=float)

    # ---------- sun altitude at open ----------
    open_sun_alt = sun.reindex(t_open, method='nearest',
                               tolerance=pd.Timedelta(sun_match_tolerance))

    # ---------- ΔT0 and Δh0 at open ----------
    dT_col = next((c for c in dT_col_candidates if c in feat.columns), None)
    dH_col = next((c for c in dH_col_candidates if c in feat.columns), None)
    dT0 = feat[dT_col].reindex(t_open, method='nearest',
                               tolerance=pd.Timedelta(env_match_tolerance)) if dT_col else pd.Series(index=t_open, dtype=float)
    dH0 = feat[dH_col].reindex(t_open, method='nearest',
                               tolerance=pd.Timedelta(env_match_tolerance)) if dH_col else pd.Series(index=t_open, dtype=float)

    # ---------- wind speed/direction at open and at IQ start ----------
    t_iq_valid = pd.to_datetime(t_iq_start_series, errors='coerce')

    wind_speed_open = feat[wind_speed_col].reindex(t_open, method='nearest',
                                                   tolerance=pd.Timedelta(wind_match_tolerance))
    wind_speed_iq = feat[wind_speed_col].reindex(t_iq_valid, method='nearest',
                                                 tolerance=pd.Timedelta(wind_match_tolerance))
    wind_speed_iq.index = events.index

    wind_dir_open = feat[wind_dir_col].reindex(t_open, method='nearest',
                                               tolerance=pd.Timedelta(wind_match_tolerance))
    wind_dir_iq = feat[wind_dir_col].reindex(t_iq_valid, method='nearest',
                                             tolerance=pd.Timedelta(wind_match_tolerance))
    wind_dir_iq.index = events.index

    # ---------- dome azimuth at open ----------
    dome_az_open = feat[dome_az_col].reindex(t_open, method='nearest',
                                             tolerance=pd.Timedelta(dome_az_match_tolerance))

    # ---------- histogram summaries ----------
    h_delta = _rename_with_unit(_hist_summary(events[delta_min_start_col], bin_size_min), 'min')
    h_sunset = _rename_with_unit(_hist_summary(events['open_minus_sunset_min'], bin_size_min), 'min')
    h_m12 = _rename_with_unit(_hist_summary(events['open_minus_minus12_min'], bin_size_min), 'min')

    h_open_alt = _rename_with_unit(_hist_summary(open_sun_alt, bin_size_deg), 'deg')
    h_dT = _rename_with_unit(_hist_summary(dT0, bin_size_dT), 'C') if dT_col else None
    h_dH = _rename_with_unit(_hist_summary(dH0, bin_size_dH), 'kJkg') if dH_col else None

    h_alt_extras = {}
    for a, vals in open_minus_alt.items():
        h_alt_extras[a] = _rename_with_unit(_hist_summary(pd.Series(vals, index=t_open), bin_size_min), 'min')

    h_wind_speed_open = _rename_with_unit(_hist_summary(wind_speed_open, bin_size_wind), 'mps')
    h_wind_speed_iq = _rename_with_unit(_hist_summary(wind_speed_iq, bin_size_wind), 'mps')

    h_wind_dir_open = _hist_summary_direction(wind_dir_open, bin_size_dir_deg)
    h_wind_dir_iq = _hist_summary_direction(wind_dir_iq, bin_size_dir_deg)

    h_dome_az_open = _hist_summary_direction(dome_az_open, bin_size_dir_deg)

    summary = {
        'delta_min_start': h_delta,
        'open_minus_sunset_min': h_sunset,
        'open_minus_minus12_min': h_m12,
        'open_minus_altitude_min': h_alt_extras,  # dict keyed by altitude (deg)
        'open_sun_altitude_deg': h_open_alt,
        'dT0_C': h_dT,
        'dH0_kJkg_moist': h_dH,
        'wind_speed_open_mps': h_wind_speed_open,
        'wind_speed_iq_start_mps': h_wind_speed_iq,
        'wind_dir_open_deg': h_wind_dir_open,
        'wind_dir_iq_start_deg': h_wind_dir_iq,
        'dome_az_open_deg': h_dome_az_open,
    }

    # ---------- plotting ----------
    figs = []
    if show or return_figs:
        # Main figure: 2x3 grid
        fig, axs = plt.subplots(2, 3, figsize=(15, 8), constrained_layout=True)

        def _plot_hist(ax, entry, title, xlabel, zero_line=False):
            if entry is None or entry.get('n', 0) == 0:
                ax.set_title(title + ' (no data)')
                ax.set_xlabel(xlabel)
                ax.set_ylabel('Count')
                ax.grid(True, alpha=0.3)
                return
            counts, edges = entry['counts'], entry['bin_edges']
            centers = (edges[:-1] + edges[1:]) / 2.0
            ax.hist(centers, bins=edges, weights=counts, alpha=0.85, color='tab:blue', edgecolor='white')
            if zero_line:
                ax.axvline(0, color='k', lw=0.8, alpha=0.8)
            # Mode line (linear or circular)
            mode_key = next((k for k in entry.keys() if k.startswith('mode_') or k == 'mode'), None)
            if mode_key and np.isfinite(entry.get(mode_key, np.nan)):
                ax.axvline(entry[mode_key], color='tab:red', ls='--', lw=1.2, label=f"mode ≈ {entry[mode_key]:.2f}")
            # Circular mean line if provided
            #if 'circ_mean_deg' in entry and np.isfinite(entry.get('circ_mean_deg', np.nan)):
            #    ax.axvline(entry['circ_mean_deg'], color='tab:orange', ls=':', lw=1.2, label=f"circ mean ≈ {entry['circ_mean_deg']:.1f}°")
            #if 'circ_std_deg' in entry and np.isfinite(entry.get('circ_std_deg', np.nan)):
                #ax.set_title(f"{title}\n(circ std ≈ {entry['circ_std_deg']:.1f}°)")
            ax.set_title(f"{title}")
            ax.legend(loc='best')
            ax.set_xlabel(xlabel)
            ax.set_ylabel('Count')
            ax.grid(True, alpha=0.3)

        _plot_hist(axs[0, 0], h_delta, 'IQ(t0) − Dome open(to)', 'Minutes', zero_line=True)
        _plot_hist(axs[0, 1], h_sunset, 't_open − sunset (0°)', 'Minutes', zero_line=True)
        _plot_hist(axs[0, 2], h_m12, 't_open − (−12°)', 'Minutes', zero_line=True)

        _plot_hist(axs[1, 0], h_open_alt, 'Sun altitude at open', 'Degrees')
        _plot_hist(axs[1, 1], h_dT, 'ΔT at open', '°C', zero_line=True)
        _plot_hist(axs[1, 2], h_dH, 'Δh at open', 'kJ/kg moist', zero_line=True)

        if show:
            plt.show()
        figs.append(fig)

        # Extra altitude panels (if any)
        if len(h_alt_extras) > 0:
            n = len(h_alt_extras)
            ncols = min(3, n)
            nrows = int(np.ceil(n / ncols))
            fig2, axs2 = plt.subplots(nrows, ncols, figsize=(5 * ncols, 3.8 * nrows), constrained_layout=True)
            axs2 = np.atleast_1d(axs2).ravel()
            for ax, (a, entry) in zip(axs2, sorted(h_alt_extras.items(), key=lambda kv: kv[0])):
                _plot_hist(ax, entry, f't_open − ({a:.0f}°)', 'Minutes', zero_line=True)
            for ax in axs2[n:]:
                ax.set_visible(False)
            if show:
                plt.show()
            figs.append(fig2)

        # Wind speed figure (1x2)
        fig3, axs3 = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
        _plot_hist(axs3[0], h_wind_speed_open, 'Wind speed at open', 'm/s')
        _plot_hist(axs3[1], h_wind_speed_iq, 'Wind speed at IQ start', 'm/s')
        if show:
            plt.show()
        figs.append(fig3)

        # Direction figure (1x3): wind @ open, wind @ IQ start, dome az @ open
        fig4, axs4 = plt.subplots(1, 3, figsize=(15, 4), constrained_layout=True)
        _plot_hist(axs4[0], h_wind_dir_open, 'Wind direction at open', 'Degress')
        _plot_hist(axs4[1], h_wind_dir_iq, 'Wind direction at IQ start', 'Degress')
        _plot_hist(axs4[2], h_dome_az_open, 'Dome azimuth at open', 'Degress')
        if show:
            plt.show()
        figs.append(fig4)

    return (summary, figs if return_figs else None)
```

```python
import numpy as np
import pandas as pd

def _series_stats(values: pd.Series | np.ndarray, mae_center='median'):
    v = pd.to_numeric(pd.Series(values), errors='coerce').dropna()
    n = int(v.size)
    if n == 0:
        return dict(n=0, median=np.nan, sd=np.nan, mae=np.nan, mean=np.nan,
                    p25=np.nan, p75=np.nan, vmin=np.nan, vmax=np.nan)
    med = float(v.median())
    mean = float(v.mean())
    sd = float(v.std(ddof=1)) if n > 1 else 0.0
    mae = float((v - (med if mae_center=='median' else mean)).abs().mean())
    return dict(n=n, median=med, sd=sd, mae=mae,
                mean=mean, p25=float(v.quantile(0.25)), p75=float(v.quantile(0.75)),
                vmin=float(v.min()), vmax=float(v.max()))

def compute_night_windows_from_features(features: pd.DataFrame, sun_col='SunAltitude') -> pd.DataFrame:
    feat = ensure_utc_index(features, 'features')
    if sun_col not in feat.columns:
        raise KeyError(f"{sun_col} not found in features.")
    t_night_min, night_id_series = time_since_minus_twelve_start(feat, sun_col)
    rows = []
    ids = [int(x) for x in pd.Series(night_id_series).dropna().unique()]
    ids.sort()
    for i, nid in enumerate(ids):
        mask = (night_id_series == nid)
        if not mask.any():
            continue
        t0 = feat.index[mask][0]
        if i+1 < len(ids):
            t1 = feat.index[night_id_series == ids[i+1]][0]
        else:
            t1 = t0 + pd.Timedelta(hours=24)
        rows.append({'night_id': nid, 't0_m12': t0, 't1_m12': t1})
    return pd.DataFrame(rows)

def collect_dh_dT_plot_points(
    features: pd.DataFrame,
    events: pd.DataFrame,
    *,
    dh_col='dhe_kJkg_moist_sm',
    dT_col='dTe_C_sm',
    wind_col='WindSpeed_301',
    scenario_col='scenario',
    sun_col='SunAltitude',
    mode='event',               # 'event' (one trajectory per event) or 'night'
    anchor='t_open',            # 't_open' or 't_m12'
    horizon_min=240,
    sample_every_min=5,
    initial_tol_min=10
):
    """
    Collect the exact points used in the Δh–ΔT plot:
      - points_df: rows for every sampled timestamp (per event or per night, per scenario)
                   columns: [scenario, event_open, night_id?, time, minutes_since_anchor, dT, dH, wind]
      - init_df:   initial-condition points at t_open
                   columns: [scenario, event_open, night_id?, dT0, dH0, wind0]
    The sampling mirrors the plot: nearest-with-tolerance reindexing on a 1D grid from anchor to anchor+horizon_min.
    """
    feat = ensure_utc_index(features, 'features')
    have_wind = (wind_col in feat.columns)
    points_rows, init_rows = [], []

    # Prepare night windows if needed
    nights = None
    if mode == 'night' or anchor == 't_m12':
        if 'night_id' not in events.columns:
            raise KeyError("events must have 'night_id' for mode='night' or anchor='t_m12'.")
        nights = compute_night_windows_from_features(feat, sun_col=sun_col)

    # Iterate per scenario
    scen_list = sorted(events[scenario_col].dropna().unique().tolist()) if scenario_col in events.columns else ['__all__']
    if not scen_list:
        scen_list = ['__all__']

    for scen in scen_list:
        ev_scen = events[events[scenario_col] == scen].copy() if scenario_col in events.columns else events.copy()
        ev_scen = ev_scen.sort_index()  # index is t_open

        if mode == 'event':
            # One path per event
            for t_open, row in ev_scen.iterrows():
                event_open = pd.to_datetime(t_open, utc=True)
                # Anchor time
                if anchor == 't_open':
                    t_start = event_open
                else:
                    # use nautical start from nights for this event
                    nid = row.get('night_id', np.nan)
                    if nights is None or pd.isna(nid):
                        continue
                    nn = nights[nights['night_id'] == int(nid)]
                    if nn.empty:
                        continue
                    t_start = nn.iloc[0]['t0_m12']

                t_end = t_start + pd.Timedelta(minutes=float(horizon_min))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce') if have_wind else pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 1:
                    continue
                dT = dT[mask]; dH = dH[mask]; W = W[mask]

                mins = (dT.index - t_start).total_seconds() / 60.0
                for ts, ms, x, y, wv in zip(dT.index, mins, dT.values, dH.values, W.values):
                    points_rows.append(dict(
                        scenario=scen, event_open=event_open, time=ts,
                        minutes_since_anchor=float(ms), dT=float(x), dH=float(y),
                        wind=(float(wv) if np.isfinite(wv) else np.nan)
                    ))

                # Initial-condition sample at t_open (larger tolerance)
                dT0 = feat[dT_col].reindex([event_open], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                dH0 = feat[dh_col].reindex([event_open], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                W0  = feat[wind_col].reindex([event_open], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min))) if have_wind else pd.Series([np.nan], index=[event_open])
                if dT0.notna().all() and dH0.notna().all():
                    init_rows.append(dict(
                        scenario=scen, event_open=event_open,
                        dT0=float(dT0.iloc[0]), dH0=float(dH0.iloc[0]),
                        wind0=(float(W0.iloc[0]) if np.isfinite(W0.iloc[0]) else np.nan)
                    ))

        else:
            # One path per night (earliest open shown as initial)
            ev = ev_scen.reset_index().rename(columns={'index':'t_open'}).sort_values('t_open')
            first = ev.groupby('night_id', as_index=False).first()[['night_id','t_open',scenario_col]]
            nn = nights.merge(first, on='night_id', how='left')
            nn = nn[nn[scenario_col] == scen]

            for _, r in nn.iterrows():
                t0_m12, t1_m12 = r['t0_m12'], r['t1_m12']
                t_open_first = r['t_open']
                t_start = t0_m12 if anchor == 't_m12' else (t_open_first if pd.notna(t_open_first) else t0_m12)
                t_end = min(t1_m12, t_start + pd.Timedelta(minutes=float(horizon_min)))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce') if have_wind else pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 1:
                    continue
                dT = dT[mask]; dH = dH[mask]; W = W[mask]

                mins = (dT.index - t_start).total_seconds() / 60.0
                for ts, ms, x, y, wv in zip(dT.index, mins, dT.values, dH.values, W.values):
                    points_rows.append(dict(
                        scenario=scen, night_id=int(r['night_id']), time=ts,
                        minutes_since_anchor=float(ms), dT=float(x), dH=float(y),
                        wind=(float(wv) if np.isfinite(wv) else np.nan)
                    ))

                if pd.notna(t_open_first):
                    t0 = pd.to_datetime(t_open_first, utc=True)
                    dT0 = feat[dT_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    dH0 = feat[dh_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    W0  = feat[wind_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min))) if have_wind else pd.Series([np.nan], index=[t0])
                    if dT0.notna().all() and dH0.notna().all():
                        init_rows.append(dict(
                            scenario=scen, night_id=int(r['night_id']),
                            dT0=float(dT0.iloc[0]), dH0=float(dH0.iloc[0]),
                            wind0=(float(W0.iloc[0]) if np.isfinite(W0.iloc[0]) else np.nan)
                        ))

    points_df = pd.DataFrame(points_rows) if points_rows else pd.DataFrame(columns=['scenario','event_open','time','minutes_since_anchor','dT','dH','wind'])
    init_df   = pd.DataFrame(init_rows)   if init_rows   else pd.DataFrame(columns=['scenario','event_open','dT0','dH0','wind0'])
    return points_df, init_df

def stats_for_plot_points(points_df: pd.DataFrame, group_col='scenario', units=None) -> pd.DataFrame:
    """
    Compute per-group stats (median, sd, mae, etc.) for dT, dH, and wind if present.
    Returns a tidy DataFrame with columns: [metric, group, unit, n, median, sd, mae, mean, p25, p75, vmin, vmax]
    """
    if units is None:
        units = {'dT':'degC', 'dH':'kJ/kg moist', 'wind':'m/s'}

    metrics = [c for c in ['dT','dH','wind'] if c in points_df.columns]
    groups = points_df[group_col].dropna().unique().tolist() if group_col in points_df.columns else ['__all__']
    rows = []
    for m in metrics:
        for g in sorted(groups):
            mask = (points_df[group_col] == g) if group_col in points_df.columns else np.ones(len(points_df), dtype=bool)
            st = _series_stats(points_df.loc[mask, m])
            rows.append(dict(metric=m, group=str(g), unit=units.get(m, ''), **st))
    return pd.DataFrame(rows).sort_values(['metric','group'])

def stats_for_initial_points(init_df: pd.DataFrame, group_col='scenario', units=None) -> pd.DataFrame:
    """
    Compute per-group stats for initial-condition values at t_open: dT0, dH0, wind0 if present.
    Returns tidy DataFrame like stats_for_plot_points.
    """
    if units is None:
        units = {'dT0':'degC', 'dH0':'kJ/kg moist', 'wind0':'m/s'}
    metrics = [c for c in ['dT0','dH0','wind0'] if c in init_df.columns]
    groups = init_df[group_col].dropna().unique().tolist() if group_col in init_df.columns else ['__all__']
    rows = []
    for m in metrics:
        for g in sorted(groups):
            mask = (init_df[group_col] == g) if group_col in init_df.columns else np.ones(len(init_df), dtype=bool)
            st = _series_stats(init_df.loc[mask, m])
            rows.append(dict(metric=m, group=str(g), unit=units.get(m, ''), **st))
    return pd.DataFrame(rows).sort_values(['metric','group'])
```

```python
import numpy as np
import pandas as pd

def ensure_delta_min_start(events: pd.DataFrame, prefer_existing=True) -> pd.DataFrame:
    ev = events.copy()
    if prefer_existing and 'delta_min_start' in ev.columns:
        return ev
    if not isinstance(ev.index, pd.DatetimeIndex):
        raise TypeError("events.index must be a DatetimeIndex (t_open).")
    # Normalize index to UTC-aware Series for safe subtraction
    idx = ev.index
    idx_utc = idx.tz_localize('UTC') if idx.tz is None else idx.tz_convert('UTC')
    idx_series_utc = pd.Series(idx_utc, index=ev.index)
    t_iq = pd.to_datetime(ev.get('t_iq_start', pd.NaT), errors='coerce', utc=True)
    ev['delta_min_start'] = (t_iq - idx_series_utc).dt.total_seconds() / 60.0
    return ev

def _knee_from_cdf(x: np.ndarray) -> float:
    """
    Simple 'elbow' on CDF via max vertical distance from diagonal.
    x must be sorted ascending.
    """
    n = len(x)
    if n < 3:
        return float(np.nan)
    y = np.arange(1, n+1) / n
    # Normalize x to [0,1]
    xmin, xmax = x[0], x[-1]
    rng = (xmax - xmin) if (xmax > xmin) else 1.0
    x_norm = (x - xmin) / rng
    # Distance above diagonal y=x
    dist = y - x_norm
    i = int(np.argmax(dist))
    return float(x[i])

def recommend_tau_pre_min(
    events: pd.DataFrame,
    strategy: str = 'knee',         # 'knee' | 'quantile' | 'coverage'
    quantile: float = 0.80,         # used for strategy='quantile'
    target_preventing_frac: float | None = None,  # used for 'coverage': fraction with Δ>=tau
    drop_negatives: bool = True,    # exclude Δ<0
    max_minutes: float | None = 300 # exclude Δ>max_minutes
) -> dict:
    """
    Recommend tau_pre_min (minutes) from Δstart distribution.
    Returns a dict with tau_pre_min and summary stats.
    """
    ev = ensure_delta_min_start(events, prefer_existing=True)

    d = pd.to_numeric(ev['delta_min_start'], errors='coerce').dropna().values
    n_total = len(d)
    if drop_negatives:
        d = d[d >= 0]
    if max_minutes is not None:
        d = d[d <= float(max_minutes)]
    n_used = len(d)

    if n_used == 0:
        return {
            'tau_pre_min': np.nan, 'method': strategy, 'note': 'No valid Δstart values',
            'n_total': n_total, 'n_used': n_used
        }

    x = np.sort(d)
    # Useful quantiles
    p50 = float(np.quantile(x, 0.50))
    p80 = float(np.quantile(x, 0.80))
    p90 = float(np.quantile(x, 0.90))

    if strategy == 'quantile':
        tau = float(np.quantile(x, float(quantile)))
        method = f'quantile_P{int(round(quantile*100))}'
    elif strategy == 'coverage':
        if target_preventing_frac is None:
            raise ValueError("target_preventing_frac must be provided for strategy='coverage'.")
        # We want P(Δ>=tau)=target_preventing_frac -> tau = Q(1 - target_preventing_frac)
        q = max(0.0, min(1.0, 1.0 - float(target_preventing_frac)))
        tau = float(np.quantile(x, q))
        method = f'coverage_preventing={target_preventing_frac:.2f}'
    elif strategy == 'knee':
        tau = _knee_from_cdf(x)
        # Fallback if knee fails (degenerate distribution)
        if not np.isfinite(tau):
            tau = p80
            method = 'knee_fallback_to_P80'
        else:
            method = 'knee'
    else:
        raise ValueError("strategy must be 'knee', 'quantile', or 'coverage'.")

    return {
        'tau_pre_min': float(tau),
        'method': method,
        'n_total': int(n_total),
        'n_used': int(n_used),
        'filters': {'drop_negatives': bool(drop_negatives), 'max_minutes': float(max_minutes) if max_minutes is not None else None},
        'quantiles': {'P50': p50, 'P80': p80, 'P90': p90}
    }
```

```python
def counts_by_scenario(events: pd.DataFrame,
                       scenario_col='scenario',
                       night_col='night_id') -> tuple[dict, dict]:
    """
    Returns:
      - event_counts: dict[scenario] -> total number of events
      - night_counts: dict[scenario] -> unique number of nights with at least one event
    """
    ev = events.copy()
    ev_scen = ev[scenario_col] if scenario_col in ev.columns else pd.Series(['__all__']*len(ev), index=ev.index)
    event_counts = ev_scen.value_counts().to_dict()
    if night_col in ev.columns:
        night_counts = ev.groupby(scenario_col)[night_col].nunique().to_dict()
    else:
        night_counts = {k: np.nan for k in event_counts.keys()}
    return event_counts, night_counts

def violin_stats(violin_tidy: pd.DataFrame) -> pd.DataFrame:
    """
    Compute per-(metric, scenario) stats on the violin 'value' column.
    Returns tidy DataFrame with columns: [metric, scenario, n, median, sd, mae, mean, p25, p75, vmin, vmax].
    """
    def _series_stats(values: pd.Series | np.ndarray, mae_center='median'):
        v = pd.to_numeric(pd.Series(values), errors='coerce').dropna()
        n = int(v.size)
        if n == 0:
            return dict(n=0, median=np.nan, sd=np.nan, mae=np.nan, mean=np.nan,
                        p25=np.nan, p75=np.nan, vmin=np.nan, vmax=np.nan)
        med = float(v.median()); mean = float(v.mean())
        sd = float(v.std(ddof=1)) if n > 1 else 0.0
        mae = float((v - (med if mae_center=='median' else mean)).abs().mean())
        return dict(n=n, median=med, sd=sd, mae=mae, mean=mean,
                    p25=float(v.quantile(0.25)), p75=float(v.quantile(0.75)),
                    vmin=float(v.min()), vmax=float(v.max()))

    rows = []
    for (metric, scen), g in violin_tidy.groupby(['metric','scenario'], sort=False):
        st = _series_stats(g['value'])
        rows.append(dict(metric=metric, scenario=scen, **st))
    return pd.DataFrame(rows).sort_values(['metric','scenario'])

def plot_violins_by_scenario_with_counts(
    violin_tidy: pd.DataFrame,
    *,
    scenario_order: list[str] | None = None,
    metric_order: list[str] | None = None,
    event_counts: dict | None = None,   # scenario -> Nevents
    nights_overall_counts: dict | None = None,  # scenario -> N nights overall (optional)
    colormap_name='tab10',
    figsize=(16, 4),
    ncols=None,
    titles_by_metric: dict | None = None,   # per-subplot custom titles
    suptitle: str | None = "Per-night distributions by scenario (violins)",
    show_medians=True,
    label_fmt="{scen}\nN={n_night} nights"  # can be "{scen}\nN={n_night} nights, Ne={n_event} ev"
):
    """
    Draw side-by-side violin plots by scenario (one subplot per metric), with:
      - one color per scenario (deprecation-safe colormap)
      - x labels annotated with N (nights used in violin) and optionally Nevents
      - custom titles per subplot and custom suptitle

    Returns (fig, axes, stats_df)
    """
    df = violin_tidy.copy()
    if df.empty:
        fig, ax = plt.subplots(figsize=figsize)
        ax.text(0.5,0.5,"No data for violin plots", ha='center', va='center'); ax.axis('off')
        return fig, [ax], pd.DataFrame()

    if metric_order is None:
        metric_order = list(df['metric'].unique())
    if scenario_order is None:
        scenario_order = list(sorted(df['scenario'].unique()))

    # Deprecation-safe colormap
    try:
        from matplotlib import colormaps as mpl_cmaps
        cmap = mpl_cmaps.get_cmap(colormap_name, len(scenario_order))
    except Exception:
        cmap = plt.get_cmap(colormap_name, len(scenario_order))  # fallback

    colors = {s: cmap(i % cmap.N) for i, s in enumerate(scenario_order)}

    # Layout
    m = len(metric_order)
    if ncols is None:
        ncols = min(m, 4)
    nrows = int(np.ceil(m / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize, squeeze=False)
    axes = axes.flatten()

    # Stats for what is plotted
    stats_rows = []

    for ax_idx, metric in enumerate(metric_order):
        ax = axes[ax_idx]
        sub = df[df['metric'] == metric]
        if sub.empty:
            ax.set_visible(False); continue

        # Build data, positions, colors, and per-scenario labels including counts
        data = []; pos = []; facecols = []; tick_labels = []
        for j, scen in enumerate(scenario_order):
            vals = sub.loc[sub['scenario'] == scen, 'value'].dropna().values
            # violin data (empty placeholder keeps positions consistent)
            data.append(vals if vals.size else np.array([np.nan]))
            pos.append(j+1)
            facecols.append(colors[scen])

            # Counts: nights used in this metric’s violin (post-filter)
            n_night_metric = int(np.isfinite(vals).sum())
            # Optional events overall (from events table)
            n_event = int(event_counts.get(scen, np.nan)) if event_counts else None

            # Build tick label from label_fmt
            label = label_fmt.format(
                scen=scen,
                n_night=n_night_metric,
                n_event=(n_event if n_event is not None and np.isfinite(n_event) else "")
            )
            tick_labels.append(label)

            # gather stats for this metric+scenario
            med = float(np.nanmedian(vals)) if vals.size else np.nan
            n = int(vals.size if np.isfinite(vals).any() else 0)
            if n > 1:
                sd = float(np.nanstd(vals, ddof=1))
                mae = float(np.nanmean(np.abs(vals - med)))
            elif n == 1:
                sd = 0.0; mae = 0.0
            else:
                sd = np.nan; mae = np.nan
            stats_rows.append(dict(metric=metric, scenario=scen, n=n, median=med, sd=sd, mae=mae))

        # Draw violins
        parts = ax.violinplot(data, positions=pos, showextrema=False, widths=0.85)
        for pc, scen in zip(parts['bodies'], scenario_order):
            pc.set_facecolor(colors[scen]); pc.set_alpha(0.6); pc.set_edgecolor('k'); pc.set_linewidth(0.6)

        # Overlay medians
        if show_medians:
            meds_y = []
            for vals in data:
                vals = vals[np.isfinite(vals)]
                meds_y.append(np.median(vals) if vals.size else np.nan)
            ax.scatter(pos, meds_y, c='k', s=18, zorder=3, label='median' if ax_idx == 0 else None)

        ax.set_xticks(pos)
        ax.set_xticklabels(tick_labels, rotation=20)
        # Titles per subplot
        if titles_by_metric and metric in titles_by_metric:
            ax.set_title(titles_by_metric[metric])
        else:
            ax.set_title(metric)
        ax.grid(True, axis='y', alpha=0.3)

    # Hide unused
    for j in range(ax_idx+1, len(axes)):
        axes[j].set_visible(False)

    if suptitle:
        fig.suptitle(suptitle, y=0.98)
    fig.tight_layout(rect=[0, 0, 1, 0.96])

    stats_df = pd.DataFrame(stats_rows).sort_values(['metric','scenario'])
    return fig, axes[:m], stats_df
```

The basic metrics to diagnose the validity of the model are summarized below:

An R² score telling how well all other sensors together can predict the target sensor. High values will indicate that the model explains a high percentage of the variance in the target sensor.
Root Mean Squared Error (RMSE) telling how far off the predictions typically are. Low MSE values will indicate high prediction accuracy.
Mean Absolute Error (MAE) and Mean Absolute Percentage Error (MAPE) with low values indicating the amount in degress in which the predictions are off from the data.
Residuals centered near zero with small standard deviation indicating no systematic bias.
It is also necesary to include Cross-Validation (CV) folding (k-fold spliting data into random train and test subsets) and handling regularization, to manage multicolinearity and to prevent overfitting. The basic metrics to diagnose the validity of the CV folding are summarized below:

An CV-R² score indicating the model overall performance across all folds.
An CV-std score with low values showing consistency in performance across the different data splits.
Folds showing similar error patterns will be indicative of a robust model.
The model also shows the feature importance coefficients (without lag features) telling how much each sensor contributes to predicting the target. The bar chart will show which sensors have the strongest influence (positive or negative) on the target sensor.

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import ElasticNet
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GroupKFold, KFold, GridSearchCV
from sklearn.metrics import make_scorer, mean_absolute_error
from math import radians, sin, cos

# ---------- Utilities ----------
def ensure_utc_index(df, name='df'):
    if not isinstance(df.index, pd.DatetimeIndex):
        raise TypeError(f"{name} must have a DatetimeIndex.")
    return df.tz_localize('UTC') if df.index.tz is None else df.tz_convert('UTC')

def _to_float(s):
    return pd.to_numeric(s, errors='coerce')

def time_since_minus_twelve_start(feat: pd.DataFrame, sun_col='SunAltitude'):
    sun = _to_float(feat[sun_col])
    mask_night = sun <= -12.0
    # Rising edge (entering nautical night)
    night_id = (mask_night.astype(int).diff() == 1).cumsum()
    night_id = night_id.where(mask_night, np.nan)
    t_night_min = pd.Series(np.nan, index=feat.index)
    for nid in sorted(night_id.dropna().unique()):
        m = (night_id == nid)
        if not m.any():
            continue
        t0 = feat.index[m][0]
        t_night_min.loc[m] = (feat.index[m] - t0).total_seconds() / 60.0
    return t_night_min, night_id

# ---------- Dataset builder ----------
def build_phase_b_dataset(
    feat: pd.DataFrame,
    iq_1m: pd.DataFrame,
    H_minutes=60,
    dT_col_candidates=('dTe_C_sm','dTe_C'),
    wind_speed_col='WindSpeed_301',
    wind_dir_col='WindDirection_301',
    sun_col='SunAltitude',
    use_signed_time=True,
    iq_col='IQ',
    reindex_tolerance='90s',
    target='delta'  # 'delta' => IQ(t+H)-IQ(t); 'level' => IQ(t+H)
):
    """
    Returns X, y, groups, meta dict with:
      feature_names, index (timestamps), night_id
    Uses features: ΔT, wind speed, wind direction (sin/cos), Sun altitude, time-of-night (hours).
    """
    feat = ensure_utc_index(feat.copy(), 'feat')
    iq_1m = ensure_utc_index(iq_1m.copy(), 'iq_1m')
    for c in (wind_speed_col, wind_dir_col, sun_col):
        if c not in feat.columns:
            raise KeyError(f"feat missing required column '{c}'")

    # Pick ΔT column
    dT_col = next((c for c in dT_col_candidates if c in feat.columns), None)
    if dT_col is None:
        raise KeyError(f"feat missing ΔT columns; tried {dT_col_candidates}")

    # Time-of-night and night labels
    t_night_min, night_id = time_since_minus_twelve_start(feat, sun_col=sun_col)
    if use_signed_time:
        # Signed minutes relative to nearest nautical start:
        # Positive inside night, negative before next night start.
        signed = t_night_min.copy()
        if signed.isna().any():
            # Approximate next night start for non-night times
            mask_night = (_to_float(feat[sun_col]) <= -12.0)
            starts = feat.index[(mask_night.astype(int).diff() == 1).fillna(False)]
            last_start = pd.Series(index=feat.index, dtype='datetime64[ns, UTC]')
            last_start.loc[starts] = starts
            last_start = last_start.ffill()
            next_start = pd.Series(index=feat.index, dtype='datetime64[ns, UTC]')
            next_start.loc[starts] = starts
            next_start = next_start[::-1].ffill()[::-1]
            for t in feat.index[signed.isna()]:
                t1 = next_start.loc[t]
                if pd.notna(t1):
                    signed.loc[t] = -(t1 - t).total_seconds()/60.0
        t_night_min = signed
    t_night_hr = _to_float(t_night_min) / 60.0

    base = pd.DataFrame(index=feat.index)
    base['dT_C'] = _to_float(feat[dT_col])
    base['wind_mps'] = _to_float(feat[wind_speed_col])
    # Wind direction to sin/cos (radians)
    wdir_deg = _to_float(feat[wind_dir_col]) % 360.0
    wdir_rad = np.deg2rad(wdir_deg)
    base['wind_dir_sin'] = np.sin(wdir_rad)
    base['wind_dir_cos'] = np.cos(wdir_rad)
    base['sun_deg'] = _to_float(feat[sun_col])
    base['t_night_hr'] = _to_float(t_night_hr)

    # Align IQ to feature timestamps (nearest-with-tolerance)
    tol = pd.Timedelta(reindex_tolerance)
    iq_at_t = _to_float(iq_1m[iq_col].reindex(base.index, method='nearest', tolerance=tol))
    # IQ at t+H
    shifted_index = base.index + pd.Timedelta(minutes=int(H_minutes))
    iq_at_th = _to_float(iq_1m[iq_col].reindex(shifted_index, method='nearest', tolerance=tol).values)
    # Target
    if target == 'delta':
        y = iq_at_th - iq_at_t
    elif target == 'level':
        y = iq_at_th
    else:
        raise ValueError("target must be 'delta' or 'level'")

    # Assemble dataset
    X = base.dropna()
    y = pd.Series(y, index=base.index).loc[X.index]
    # Drop rows with NaN target
    m = np.isfinite(y.values)
    X = X.loc[m]
    y = y.loc[m]

    # Groups by night_id (align to X)
    night = night_id.reindex(X.index)
    # Make groups integers 0..K-1 (NaN -> new group per row to avoid leakage; but better to drop)
    # Drop rows without night label to enable grouped CV correctly
    good = night.notna()
    X = X.loc[good]
    y = y.loc[good]
    night = night.loc[good].astype(int)

    meta = {
        'feature_names': X.columns.tolist(),
        'index': X.index,
        'night_id': night
    }
    return X, y, night.values, meta

# ---------- Model fit ----------
def fit_phase_b_elasticnet(
    X, y, groups,
    degrees=(1, 2),
    alphas=None,
    l1_ratios=None,
    mae_within_tol=0.01,   # arcsec tolerance to prefer simpler degree
    random_state=0
):
    if alphas is None:
        alphas = np.logspace(-4, 2, 10)
    if l1_ratios is None:
        l1_ratios = np.linspace(0.0, 0.7, 8)

    pipe = Pipeline(steps=[
        ('poly', PolynomialFeatures(include_bias=False)),
        ('scaler', StandardScaler(with_mean=True, with_std=True)),
        ('enet', ElasticNet(max_iter=5000, random_state=random_state))
    ])

    param_grid = {
        'poly__degree': list(degrees),
        'enet__alpha': list(alphas),
        'enet__l1_ratio': list(l1_ratios),
    }

    # Grouped CV by night; fallback to KFold if fewer than 2 groups
    uniq_groups = np.unique(groups)
    if uniq_groups.size >= 2:
        cv = GroupKFold(n_splits=min(5, uniq_groups.size))
        cv_groups = groups
    else:
        cv = KFold(n_splits=min(5, len(y)), shuffle=True, random_state=random_state)
        cv_groups = None

    scoring = {
        'mae': 'neg_mean_absolute_error',
        'rmse': 'neg_root_mean_squared_error'
    }

    grid = GridSearchCV(
        pipe, param_grid=param_grid, scoring=scoring,
        refit='mae', cv=cv, n_jobs=-1, verbose=0
    )
    grid.fit(X, y, groups=cv_groups)

    # Prefer simpler degree within tolerance on MAE
    res = pd.DataFrame(grid.cv_results_)
    # Convert to positive MAE (lower is better)
    res['MAE'] = -res['mean_test_mae']
    best_mae = res['MAE'].min()
    cand = res[res['MAE'] <= (best_mae + mae_within_tol)].copy()
    # Choose smallest degree, then smallest alpha, then smallest l1_ratio
    cand = cand.sort_values(by=['param_poly__degree','param_enet__alpha','param_enet__l1_ratio','MAE'])
    chosen = cand.iloc[0]
    best_params = {
        'degree': int(chosen['param_poly__degree']),
        'alpha': float(chosen['param_enet__alpha']),
        'l1_ratio': float(chosen['param_enet__l1_ratio'])
    }

    # Refit final model with chosen params on all data
    final = Pipeline(steps=[
        ('poly', PolynomialFeatures(include_bias=False, degree=best_params['degree'])),
        ('scaler', StandardScaler(with_mean=True, with_std=True)),
        ('enet', ElasticNet(
            alpha=best_params['alpha'],
            l1_ratio=best_params['l1_ratio'],
            max_iter=5000,
            random_state=random_state
        ))
    ])
    final.fit(X, y)

    # Summaries
    summary = {
        'best_params': best_params,
        'cv_best_mae': float(best_mae),
        'cv_best_rmse': float((-res.loc[cand.index, 'mean_test_rmse']).iloc[0]) if 'mean_test_rmse' in res else np.nan,
        'cv_results': res[['param_poly__degree','param_enet__alpha','param_enet__l1_ratio','MAE','mean_test_rmse']].rename(columns={'mean_test_rmse':'negRMSE'}) if 'mean_test_rmse' in res else res
    }
    return final, summary

# ---------- Decision helper ----------
def recommend_open(delta_iq_hat, improve_threshold_arcsec=-0.1):
    """
    delta_iq_hat: predicted IQ(t+H) - IQ(t).
      Negative is improvement (smaller FWHM).
    improve_threshold_arcsec: e.g., -0.1 means open if predicted improvement <= -0.1".
    """
    return bool(delta_iq_hat <= float(improve_threshold_arcsec))
```

```python
# Inputs:
# - feat: minute-level features on a DatetimeIndex with columns:
#   ['dTe_C_sm' or 'dTe_C', 'WindSpeed_301', 'WindDirection_301', 'SunAltitude']
# - iq_1m: minute-level IQ series DataFrame with column 'IQ' on a DatetimeIndex

H = 60  # prediction horizon in minutes

# 1) Build dataset
X, y, groups, meta = build_phase_b_dataset(
    feat=features,
    iq_1m=iq_1m,
    H_minutes=H,
    dT_col_candidates=('dTe_C_sm','dTe_C'),
    wind_speed_col='WindSpeed_301',
    wind_dir_col='WindDirection_301',
    sun_col='SunAltitude',
    target='delta'  # predict ΔIQ over H minutes
)
print("Dataset:", X.shape, "Target samples:", y.shape, "N nights:", len(np.unique(groups)))

# 2) Fit model with grouped CV
model, info = fit_phase_b_elasticnet(
    X, y, groups,
    degrees=(1,2),
    alphas=np.logspace(-4, 2, 10),
    l1_ratios=np.linspace(0.0, 0.7, 8),
    mae_within_tol=0.01,
    random_state=0
)
print("Best params:", info['best_params'])
print("CV best MAE (arcsec):", info['cv_best_mae'])

# 3) Make a decision for the latest timestamp
x_latest = X.iloc[[-1]]          # last row
delta_hat = float(model.predict(x_latest)[0])
do_open = recommend_open(delta_hat, improve_threshold_arcsec=-0.1)
print(f"ΔIQ_hat@{H}m = {delta_hat:.3f} arcsec -> Open? {do_open}")

# 4) Batch predictions if needed
yhat = model.predict(X)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import ceil
from matplotlib import cm
from matplotlib.colors import Normalize

# Assumes your ensure_utc_index(...) and time_since_minus_twelve_start(...) exist.

def compute_night_windows_from_features(features: pd.DataFrame, sun_col='SunAltitude') -> pd.DataFrame:
    feat = ensure_utc_index(features, 'features')
    if sun_col not in feat.columns:
        raise KeyError(f"{sun_col} not found in features.")
    t_night_min, night_id_series = time_since_minus_twelve_start(feat, sun_col)
    rows = []
    ids = [int(x) for x in pd.Series(night_id_series).dropna().unique()]
    ids.sort()
    for i, nid in enumerate(ids):
        mask = (night_id_series == nid)
        if not mask.any():
            continue
        t0 = feat.index[mask][0]
        if i+1 < len(ids):
            t1 = feat.index[night_id_series == ids[i+1]][0]
        else:
            t1 = t0 + pd.Timedelta(hours=24)
        rows.append({'night_id': nid, 't0_m12': t0, 't1_m12': t1})
    return pd.DataFrame(rows)

def plot_dh_dT_vectors_by_scenario_v2(
    features: pd.DataFrame,
    events: pd.DataFrame,
    dh_col='dhe_kJkg_moist_sm',
    dT_col='dTe_C_sm',
    wind_col='WindSpeed_301',
    sun_col='SunAltitude',
    scenario_col='scenario',
    mode='event',                # 'event' -> one trajectory per event; 'night' -> one per night
    anchor='t_open',             # 't_open' or 't_m12' (m12 requires night mapping)
    horizon_min=240,             # window length after anchor
    sample_every_min=5,
    invert_x=True,
    invert_y=False,
    initial_marker='D',          # diamond for initial conditions
    initial_tol_min=10,          # tolerance to sample ΔT/Δh at t_open for the marker
    figsize=(12, 5),
    alpha_points=0.7,
    arrow_alpha=0.35,
    title="Δh vs ΔT (by scenario; diamonds = initial at open; color = wind)"
):
    """
    Event-mode (default): titles show event totals per scenario and drawn events. Initial markers match drawn events.
    Night-mode: one path per night (like before), titles show per-night counts.
    Colors points by wind with a shared colorbar. Inverts axes if requested.
    Returns (fig, axes, counts_dict).
    """
    feat = ensure_utc_index(features, 'features')
    for col in [dh_col, dT_col]:
        if col not in feat.columns:
            raise KeyError(f"features missing '{col}'")

    have_wind = (wind_col in feat.columns)
    # Wind color mapping (shared)
    if have_wind:
        w_all = pd.to_numeric(feat[wind_col], errors='coerce').clip(lower=0)
        wmin = float(np.nanpercentile(w_all, 2)) if w_all.notna().any() else 0.0
        wmax = float(np.nanpercentile(w_all, 98)) if w_all.notna().any() else (wmin + 1.0)
        if not np.isfinite(wmax) or wmax <= wmin:
            wmax = wmin + 1.0
        norm = Normalize(vmin=wmin, vmax=wmax)
        cmap = cm.viridis
        sm = cm.ScalarMappable(norm=norm, cmap=cmap)
        sm.set_array([])
    else:
        norm = None; cmap = None; sm = None

    scen_list = sorted(events[scenario_col].dropna().unique().tolist()) if scenario_col in events.columns else ['__all__']
    if not scen_list:
        scen_list = ['__all__']

    # Build night windows if needed (for m12 anchoring or night-mode)
    nights = None
    if mode == 'night' or anchor == 't_m12':
        nights = compute_night_windows_from_features(feat, sun_col=sun_col)
        if 'night_id' not in events.columns:
            raise KeyError("events must have 'night_id' to use mode='night' or anchor='t_m12'.")

    # Panel layout
    n_scen = len(scen_list)
    ncols = min(2, n_scen)
    nrows = int(ceil(n_scen / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize, sharex=True, sharey=True)
    if nrows * ncols == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    counts = {}
    for i, scen in enumerate(scen_list):
        ax = axes[i]
        drawn = 0
        total = 0

        if mode == 'event':
            # One trajectory per event
            ev_scen = events[events[scenario_col] == scen].copy()
            ev_scen = ev_scen.sort_index()  # index is t_open
            total = len(ev_scen)
            for t_open, row in ev_scen.iterrows():
                # Anchor time
                if anchor == 't_open':
                    t_start = t_open
                else:
                    # use nautical start from nights for this event
                    nid = row.get('night_id', np.nan)
                    if nights is None or pd.isna(nid):
                        continue
                    nn = nights[nights['night_id'] == int(nid)]
                    if nn.empty: 
                        continue
                    t_start = nn.iloc[0]['t0_m12']

                t_end = t_start + pd.Timedelta(minutes=float(horizon_min))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                if have_wind:
                    W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                else:
                    W  = pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 2:
                    continue
                drawn += 1

                dT = dT[mask]; dH = dH[mask]; W = W[mask]
                if have_wind:
                    colors = cmap(norm(W.values))
                    ax.scatter(dT.values, dH.values, s=40, c=colors, edgecolor='k', linewidth=0.3, alpha=alpha_points)
                else:
                    ax.scatter(dT.values, dH.values, s=40, color='tab:blue', edgecolor='k', linewidth=0.3, alpha=alpha_points)

                x = dT.values; y = dH.values
                dx = x[1:] - x[:-1]; dy = y[1:] - y[:-1]
                ax.quiver(x[:-1], y[:-1], dx, dy, angles='xy', scale_units='xy', scale=1.0,
                          color='gray', alpha=arrow_alpha, width=0.002, headwidth=3, headlength=4)

                # Initial marker at t_open (sample ΔT/Δh at t_open with larger tolerance)
                t0 = pd.to_datetime(t_open, utc=True)
                t_window = t0 + pd.to_timedelta([0], unit='min')
                dT0 = feat[dT_col].reindex(t_window, method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                dH0 = feat[dh_col].reindex(t_window, method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                if dT0.notna().all() and dH0.notna().all():
                    ax.scatter([float(dT0.iloc[0])], [float(dH0.iloc[0])],
                               s=90, marker=initial_marker, color='gold', edgecolor='k', linewidth=0.6, zorder=5)

        else:
            # Night-mode: one trajectory per night (earliest open shown with marker)
            if nights is None:
                nights = compute_night_windows_from_features(feat, sun_col=sun_col)
            # Map earliest open per night for marker and scenario filtering
            ev = events.reset_index().rename(columns={'index':'t_open'}).sort_values('t_open')
            first = ev.groupby('night_id', as_index=False).first()[['night_id','t_open',scenario_col]]
            # Filter to nights of this scenario
            nn = nights.merge(first, on='night_id', how='left')
            nn = nn[nn[scenario_col] == scen]
            total = nn['night_id'].nunique()

            for _, r in nn.iterrows():
                t0_m12, t1_m12 = r['t0_m12'], r['t1_m12']
                t_open_first = r['t_open']
                t_start = t0_m12 if anchor == 't_m12' else (t_open_first if pd.notna(t_open_first) else t0_m12)
                t_end = min(t1_m12, t_start + pd.Timedelta(minutes=float(horizon_min)))
                grid = pd.date_range(t_start, t_end, freq=f'{int(sample_every_min)}min', tz='UTC')

                tol = pd.Timedelta(minutes=max(1, sample_every_min))
                dT = pd.to_numeric(feat[dT_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                dH = pd.to_numeric(feat[dh_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce')
                W  = pd.to_numeric(feat[wind_col].reindex(grid, method='nearest', tolerance=tol), errors='coerce') if have_wind else pd.Series(np.nan, index=grid)

                mask = dT.notna() & dH.notna()
                if mask.sum() < 2:
                    continue
                drawn += 1

                dT = dT[mask]; dH = dH[mask]; W = W[mask]
                if have_wind:
                    colors = cmap(norm(W.values))
                    ax.scatter(dT.values, dH.values, s=40, c=colors, edgecolor='k', linewidth=0.3, alpha=alpha_points)
                else:
                    ax.scatter(dT.values, dH.values, s=40, color='tab:blue', edgecolor='k', linewidth=0.3, alpha=alpha_points)

                x = dT.values; y = dH.values
                dx = x[1:] - x[:-1]; dy = y[1:] - y[:-1]
                ax.quiver(x[:-1], y[:-1], dx, dy, angles='xy', scale_units='xy', scale=1.0,
                          color='gray', alpha=arrow_alpha, width=0.002, headwidth=3, headlength=4)

                # Initial marker at earliest open in night
                if pd.notna(t_open_first):
                    t0 = pd.to_datetime(t_open_first, utc=True)
                    dT0 = feat[dT_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    dH0 = feat[dh_col].reindex([t0], method='nearest', tolerance=pd.Timedelta(minutes=float(initial_tol_min)))
                    if dT0.notna().all() and dH0.notna().all():
                        ax.scatter([float(dT0.iloc[0])], [float(dH0.iloc[0])],
                                   s=90, marker=initial_marker, color='gold', edgecolor='k', linewidth=0.6, zorder=5)

        # Axes cosmetics and titles
        if invert_x: ax.invert_xaxis()
        if invert_y: ax.invert_yaxis()
        ax.axhline(0.0, color='k', lw=0.8, alpha=0.6)
        ax.axvline(0.0, color='k', lw=0.8, alpha=0.6)
        ax.grid(True, alpha=0.3)

        # Title
        # For event mode, show event totals; for night mode, show nights
        noun = 'events' if mode == 'event' else 'nights'
        ax.set_title(f"{scen} ({noun} total={int(total)}, drawn={int(drawn)})")
        counts[scen] = dict(total=int(total), drawn=int(drawn))

    # Axis labels and shared colorbar
    for ax in axes[:len(scen_list)]:
        ax.set_xlabel("ΔT (°C)")
        ax.set_ylabel("Δh (kJ/kg moist)")
    fig.suptitle(title, y=0.98)
    fig.tight_layout(rect=[0, 0, 1, 0.96])

    for j in range(len(scen_list), len(axes)):
        axes[j].set_visible(False)

    if have_wind and sm is not None:
        cbar = fig.colorbar(sm, ax=axes[:len(scen_list)], fraction=0.02, pad=0.02)
        cbar.set_label("wind (m/s)")

    # Legend for initial marker
    axes[0].scatter([], [], s=90, marker=initial_marker, color='gold', edgecolor='k', linewidth=0.6, label='Initial (open)')
    axes[0].legend(loc='best', frameon=True)

    return fig, axes[:len(scen_list)], counts
```
